***

## 🔹 Key Concept: Mutable Default Argument

* The parameter `x=[]` is a **default argument**.
* In Python, **default arguments are evaluated only once**, when the function is defined (not every time it's called).
* Lists are **mutable objects** → meaning they can be changed.

***

## 🔹 Step-by-step execution

### ✅ First call → `test()`

* `x` is not passed → uses default `[]`

* Inside function:
  ```python
  x.append(1)
  ```
  Now `x = [1]`

* Output:
  ```
  [1]
  ```

***

### ✅ Second call → `test()`

* Again `x` is not passed → BUT Python reuses the same default list (not a new one)

* So now:
  * Previous `x` was already `[1]`
  * `x.append(1)` → now `[1, 1]`

* Output:
  ```
  [1, 1]
  ```

***

## 🔹 Final Output

```
[1]
[1, 1]
```

***

## 🚨 Why this matters (important!)

This behavior can cause **bugs** because you might expect a fresh list each time, but Python keeps modifying the same list.

***

## ✅ Correct way (if you want a new list each time)

```python
def test(x=None):
    if x is None:
        x = []
    x.append(1)
    print(x)

test()
test()
```

### ✅ Output now:

```
[1]
[1]
```

***

## 🧠 Summary

* Default arguments are created **once**.
* Mutable objects (like lists, dicts) will **persist changes** across function calls.
* Use `None` as a safe default and initialize inside the function.

## 🔹 Key Concept: Variable Scope

There are **two different `x` variables here**:

1. **Global variable** → `x = 10`
2. **Local variable inside function** → `x = 5`

***

## 🔹 Step-by-step execution

### ✅ 1. Global variable

```python
x = 10
```

* This `x` exists **outside** the function
* Called **global scope**

***

### ✅ 2. Function definition

```python
def demo():
    x = 5
    print(x)
```

* Inside the function, a **new variable `x` is created**
* This does **NOT affect the global `x`**

***

### ✅ 3. Function call

```python
demo()
```

* Inside function → `x = 5`
* So it prints:

```
5
```

***

### ✅ 4. After function call

```python
print(x)
```

* This refers to **global `x`**
* Which is still:

```
10
```

***

## ✅ Final Output

```
5
10
```

***

## 🧠 Important Concepts

### 🔸 Local Scope

* Variables defined inside a function
* Exist only while the function runs

### 🔸 Global Scope

* Variables defined outside functions
* Available everywhere (unless shadowed)

### 🔸 Variable Shadowing

* The local `x` (5) **hides** the global `x` (10) inside the function

***

## 🚀 Extra (Very Important)

If you wanted to modify the global `x`, you must use `global`:

```python
x = 10

def demo():
    global x
    x = 5

demo()
print(x)
```

✅ Output:

```
5
```

***

## 🔥 Summary

* Inside function → Python uses **local variables first**
* Global `x` remains unchanged unless explicitly modified
* This is called **scope isolation**

def rec():

***

## 🔹 What is happening?

* This function **calls itself recursively**
* There is **no stopping condition (base case)**

***

## 🔹 Step-by-step execution

1. `rec()` is called
2. Inside the function:
   ```python
   return rec()
   ```
3. It calls `rec()` again
4. That new call again calls `rec()`
5. This keeps going **forever**

***

## 🔥 Problem: Infinite Recursion

Since there is **no termination condition**, Python keeps creating new function calls **until memory is exhausted**

***

## ❌ Final Result

Python throws an error:

```
RecursionError: maximum recursion depth exceeded
```

***

## 🧠 Key Concept: Recursion must have a base case

A recursive function should always have:

1. ✅ A **base case** → where it stops
2. ✅ A **recursive case** → where it calls itself

***

## ✅ Fixed Example

```python
def rec(n):
    if n == 0:        # base case
        return 0
    return n + rec(n-1)

print(rec(3))
```

### Execution:

```
rec(3) → 3 + rec(2)
rec(2) → 2 + rec(1)
rec(1) → 1 + rec(0)
rec(0) → 0  ✅ stops
```

### Output:

```
6
```

***

## 🚀 Summary

* Your function → **infinite recursion**
* Missing → **base case**
* Result → `RecursionError`
