# Run the Read-Only Network Pytests

This notebook runs the existing pytest suite manually and displays the results in each cell. The tests use only `show` and `ping` commands and do not modify router configuration.

Each test cell is independent, so it can be run without executing earlier cells. Save the notebook after execution to preserve the visible pytest output. Detailed command evidence is also written under `results/`.

## 1. Check Python, pytest, and test discovery

```python
from pathlib import Path
import subprocess
import sys

candidates = [
    Path.cwd(),
    Path.cwd() / "read_only_pytests",
    Path.cwd() / "1.Python Learning" / "Network Coding" / "EVE-NG_3_Router_Automation" / "read_only_pytests",
]
TEST_DIR = next((path.resolve() for path in candidates if (path / "pytest.ini").exists()), None)
if TEST_DIR is None:
    raise RuntimeError("Could not locate the read_only_pytests folder")

print(f"Python: {sys.executable}")
print(f"Test directory: {TEST_DIR}")
result = subprocess.run(
    [sys.executable, "-m", "pytest", "--collect-only", "-q"],
    cwd=TEST_DIR,
    text=True,
    capture_output=True,
)
print(result.stdout)
if result.stderr:
    print(result.stderr)
print(f"Exit code: {result.returncode}")
```

## 11. View the latest saved test summary and artifact location

```python
from pathlib import Path
import json

candidates = [Path.cwd(), Path.cwd() / "read_only_pytests", Path.cwd() / "1.Python Learning" / "Network Coding" / "EVE-NG_3_Router_Automation" / "read_only_pytests"]
TEST_DIR = next(path.resolve() for path in candidates if (path / "pytest.ini").exists())
summary_path = TEST_DIR / "results" / "latest_summary.json"
if summary_path.exists():
    print(json.dumps(json.loads(summary_path.read_text(encoding="utf-8")), indent=2))
else:
    print("No result summary exists yet. Run one of the pytest cells first.")

run_directories = sorted(path for path in (TEST_DIR / "results").glob("20*") if path.is_dir())
if run_directories:
    latest = run_directories[-1]
    artifacts = list(latest.rglob("*.json"))
    print(f"Latest evidence directory: {latest}")
    print(f"Saved command artifacts: {len(artifacts)}")
```
