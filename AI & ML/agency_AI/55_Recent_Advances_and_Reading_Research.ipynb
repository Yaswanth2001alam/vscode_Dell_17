## Worked example 1: Budgeted candidate verification and a simple MoE router

**Purpose:** Inspect two modern mechanisms without claiming to reproduce a frontier model.

### Read the code in this order
1. Evaluate candidate answers with a deterministic verifier.
2. Stop at a hard candidate budget.
3. Route a tiny vector to the highest-scoring synthetic expert.
4. Assert verified correctness and route choice, not general reasoning ability.

### Syntax and behavior to notice
- `assert` checks the stated behavior; a failed check raises AssertionError.
- `raise` reports invalid input or a failed requirement instead of fabricating success.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
verified answer: 21 candidates checked: 3
toy selected expert: 1 output: [2.0, 4.0]
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import numpy as np

candidates = [18, 20, 21]
budget, checked, accepted = 3, 0, None
for candidate in candidates[:budget]:
    checked += 1
    if candidate == 3 * 7:
        accepted = candidate
        break
if accepted is None:
    raise RuntimeError("no verified candidate within budget")
assert accepted == 21 and checked <= budget
gate_logits = np.array([0.1, 2.0, -0.5])
expert = int(np.argmax(gate_logits))
experts = [lambda x: x, lambda x: 2*x, lambda x: -x]
output = experts[expert](np.array([1., 2.]))
assert expert == 1 and output.tolist() == [2., 4.]
print("verified answer:", accepted, "candidates checked:", checked)
print("toy selected expert:", expert, "output:", output.tolist())
````
