## Concept notes and mental model

A variable is a name bound to an object. Assignment does not copy a list: two names may refer
to the same mutable object. Integers and strings are immutable; lists and dictionaries are
mutable. A string is Unicode text, and f-strings insert evaluated values into readable output.

Use a list for an ordered collection that changes, a tuple for a fixed positional record,
a set for distinct unordered membership, and a dictionary for key-to-value lookup. Sets do not
preserve a meaningful presentation order, so sort them when a report needs repeatability.
Dictionary access with [] raises KeyError for an unknown required field. get is appropriate
when absence has a deliberate meaning, not when it would hide a malformed record.

AI data pipelines often begin as lists of dictionaries before conversion to a table or tensor.
Keep units in field names and separate numeric values from formatted labels. Converting a
number to a string for display too early makes numeric sorting and arithmetic error-prone.


**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
hostname = " EDGE-01 ".strip().lower()
record = {"hostname": hostname, "latency_ms": 12.5, "role": "edge"}
records = [record, {"hostname": "core-01", "latency_ms": 4.0, "role": "core"}]
coordinates = (1, 2)
roles = {row["role"] for row in records}
alias = records
separate = list(records)
separate.append({"hostname": "lab-01", "latency_ms": 1.0, "role": "lab"})
assert alias is records and separate is not records
assert len(records) == 2 and len(separate) == 3
assert coordinates[0] == 1 and hostname == "edge-01"
print(f"{record['hostname']}: {record['latency_ms']:.1f} ms")
print("roles:", sorted(roles))
````
