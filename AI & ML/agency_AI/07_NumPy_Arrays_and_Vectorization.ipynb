## Concept notes and mental model

A NumPy array stores values with a shape and dtype. A feature matrix X usually has shape
(n_samples, n_features). A target vector often has shape (n_samples,). Shape mistakes can
broadcast into a plausible but wrong calculation, so inspect shapes before trusting outputs.

Elementwise multiplication uses *, matrix multiplication uses @. Broadcasting matches trailing
dimensions that agree or have size one: subtracting a (n_features,) mean from X centers each
column. A (n_samples, 1) vector can broadcast differently from a flat vector.

Vectorization expresses batch arithmetic in array operations rather than Python-level loops.
It can be faster and clearer, but materializing a huge intermediate array can use too much memory.
Use an explicit random generator with a seed for local reproducibility. A slice may be a view;
call copy when you need independent data.

## Mathematics or data contract

For X of shape (N,D), weights w of shape (D,), X @ w produces (N,).

## Worked example 1: Batch predictions and column standardization

**Purpose:** Make array dimensions and standardization visible.

### Read the code in this order
1. Create a 3x2 feature matrix.
2. Compute one prediction per row with matrix multiplication.
3. Center and scale columns using training data statistics.
4. Assert shapes, arithmetic, and centered columns.

### Syntax and behavior to notice
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
shape: (3, 2) predictions: [8.0, 15.0, 24.0]
column means after scaling: [0.0, -0.0]
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import numpy as np

X = np.array([[10., 1.], [20., 2.], [30., 4.]])
weights = np.array([0.5, 2.0])
predictions = X @ weights + 1.0
assert predictions.shape == (3,)
np.testing.assert_allclose(predictions, [8., 15., 24.])
means, scales = X.mean(axis=0), X.std(axis=0)
standardized = (X - means) / scales
np.testing.assert_allclose(standardized.mean(axis=0), [0., 0.], atol=1e-12)
independent = X[:1].copy()
independent[0, 0] = 999
assert X[0, 0] == 10
print("shape:", X.shape, "predictions:", predictions.tolist())
print("column means after scaling:", np.round(standardized.mean(axis=0), 6).tolist())
````
