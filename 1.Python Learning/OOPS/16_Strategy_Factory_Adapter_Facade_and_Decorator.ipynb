## Concept notes and mental model

**Level:** Intermediate

**Learning goal:** Apply strategy, factory, adapter, facade, and decorator only where they simplify a real boundary.

**Patterns are names for recurring design choices, not rules requiring more classes.**
A strategy supplies interchangeable decision logic; a callable is enough for many strategies.
A factory centralizes construction when callers should not select concrete classes themselves.
An adapter translates one interface into another. A facade presents a simple operation over
several collaborators. A decorator object wraps another object to add behavior while keeping
the relevant public contract.

**A networking example.** An inventory parser produces a platform key. A factory picks an
offline adapter. A transport adapter translates a library's run method into the service's send
method. A strategy chooses a compliance threshold. A facade coordinates collection and
evaluation. A metrics decorator records how many reads occurred. Each component has one reason
to change, so tests can replace exactly the dependency they need.

**Object decorator versus @decorator syntax.** Both can wrap behavior, but the syntax and object
lifecycle differ. A function decorator transforms a callable at definition time; preserve
function metadata with functools.wraps when writing one. A wrapper object can instead implement
send and delegate to another transport.

**Avoid pattern inflation.** If a dictionary of callables solves construction, do not add a
metaclass-driven service locator. A factory must reject unknown keys explicitly. An adapter
should translate behavior without lying about error semantics. A facade should not conceal
operational failure behind an empty success report. Composition remains the common mechanism.

**Pause and explain:** Identify which pattern changes an interface and which one preserves it while adding behavior.

## Pattern 051: Strategy object for a threshold policy

**Core concept:** Swap an algorithm through an injected strategy object.

**Network-engineering purpose:** A health evaluator can use stricter or looser policies without branching on environment names.

### Read the implementation in this order
1. Define a policy with evaluate.
2. Store the limit in the policy instance.
3. Inject it into Monitor.
4. Check behavior for the scenario's load.

**Common mistake / boundary:** A strategy should encapsulate one decision, not become a second orchestration service.

**Practice after the ten worked scenarios:** Add a policy that checks a minimum headroom instead of maximum load.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0501 - Strategy object for a threshold policy: branch scenario

**Scenario 01/10; pattern 051; Intermediate.** This is a self-contained offline exercise. A health evaluator can use stricter or looser policies without branching on environment names.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `load` | `20` |
| `threshold` | `80` |

**Logic to trace:** Define a policy with evaluate. Check behavior for the scenario's load. Follow the class-specific steps under Pattern 051 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text

```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A strategy should encapsulate one decision, not become a second orchestration service.

````python
# Example 0501: Strategy object for a threshold policy - branch
case = {'load': 20, 'threshold': 80}

class ThresholdPolicy:
    def __init__(self, limit):
        self.limit = limit

    def evaluate(self, load):
        return load <= self.limit

class Monitor:
    def __init__(self, policy):
        self.policy = policy

    def check(self, load):
        return self.policy.evaluate(load)

result = Monitor(ThresholdPolicy(case["threshold"])).check(case["load"])
assert result == (case["load"] <= case["threshold"])
print(result)
````

## Pattern 052: Factory with an explicit supported set

**Core concept:** Construction can be selected by a validated external key.

**Network-engineering purpose:** A platform key can select one of a few known offline command renderers.

### Read the implementation in this order
1. Declare concrete renderers.
2. Store their constructors in a registry.
3. Look up the requested key.
4. Convert an unsupported key to an explicit domain-facing ValueError.

**Common mistake / boundary:** Do not import or execute arbitrary names supplied by an untrusted inventory file.

**Practice after the ten worked scenarios:** Inject the registry into the factory to make supported plugins application-controlled.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0511 - Factory with an explicit supported set: branch scenario

**Scenario 01/10; pattern 052; Intermediate.** This is a self-contained offline exercise. A platform key can select one of a few known offline command renderers.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `style` | `'compact'` |

**Logic to trace:** Declare concrete renderers. Convert an unsupported key to an explicit domain-facing ValueError. Follow the class-specific steps under Pattern 052 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
status brief
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not import or execute arbitrary names supplied by an untrusted inventory file.

````python
# Example 0511: Factory with an explicit supported set - branch
case = {'style': 'compact'}

class Compact:
    def render(self):
        return "status brief"

class Verbose:
    def render(self):
        return "status detail"

class Factory:
    types = {"compact": Compact, "verbose": Verbose}

    @classmethod
    def create(cls, key):
        try:
            constructor = cls.types[key]
        except KeyError as error:
            raise ValueError(f"unsupported style: {key}") from error
        return constructor()

key = case["style"]
result = Factory.create(key).render()
assert result == ("status brief" if key == "compact" else "status detail")
print(result)
````

## Pattern 055: Decorator object counts transport calls

**Core concept:** A wrapper preserves send while adding instrumentation.

**Network-engineering purpose:** A service can observe read volume without changing its replay transport.

### Read the implementation in this order
1. Create a minimal transport.
2. Wrap it with a counting transport.
3. Increment then delegate each call.
4. Check the count and returned value.

**Common mistake / boundary:** Decide whether the counter tracks attempts or successful calls; this one tracks attempts.

**Practice after the ten worked scenarios:** Track successes and failures separately while re-raising transport errors.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0541 - Decorator object counts transport calls: branch scenario

**Scenario 01/10; pattern 055; Intermediate.** This is a self-contained offline exercise. A service can observe read volume without changing its replay transport.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Create a minimal transport. Check the count and returned value. Follow the class-specific steps under Pattern 055 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
ETHERNET1/1 calls: 2
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Decide whether the counter tracks attempts or successful calls; this one tracks attempts.

````python
# Example 0541: Decorator object counts transport calls - branch
case = {'interface': 'Ethernet1/1'}

class ReplayTransport:
    def send(self, command):
        return command.upper()

class CountingTransport:
    def __init__(self, wrapped):
        self.wrapped, self.calls = wrapped, 0

    def send(self, command):
        self.calls += 1
        return self.wrapped.send(command)

transport = CountingTransport(ReplayTransport())
result = transport.send(case["interface"])
transport.send("status")
assert transport.calls == 2
assert result == case["interface"].upper()
print(result, "calls:", transport.calls)
````
