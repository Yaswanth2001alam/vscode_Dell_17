In WAN networking, **Camel Case** and **Snake Case** are **not networking protocols or WAN technologies**. They are **naming conventions** used in automation scripts, Python code, APIs, configuration generators, monitoring tools, and network source code.

Since you're learning Python and WAN automation, you'll see them often in scripts that interact with routers, switches, telemetry, and APIs.

***

# 1. Camel Case

Camel Case writes multiple words **without spaces**, and each new word starts with a capital letter.

Examples:

```text
loopbackInterface
routerHostname
neighborIPAddress
bgpPeerStatus
interfaceDescription
```

The "humps" look like a camel's back:

```text
bgpPeerStatus
   ^   ^
```

### Network Example

Instead of:

```text
bgp peer status
```

Use:

```python
bgpPeerStatus = "Established"
```

***

# 2. Snake Case

Words are separated by underscores (\_).

Examples:

```text
loopback_interface
router_hostname
neighbor_ip_address
bgp_peer_status
interface_description
```

### Network Example

```python
bgp_peer_status = "Established"
```

# 4. Which One Does Python Use?

Python recommends **snake\_case**.

PEP-8 (Python coding standard):

✅ Preferred

```python
router_name = "rwa01"
isis_neighbor_count = 8
```

❌ Not preferred

```python
routerName = "rwa01"
isisNeighborCount = 8
```

This is why most network automation scripts use snake\_case.

## Interview/Automation Takeaway

In network automation:

1. Receive JSON from device/API.
2. Normalize all keys to `snake_case`.
3. Validate expected values:
   * BGP → `Established`
   * IS-IS → Neighbor count > 0
   * LLDP → Neighbor present
   * MACsec → `Secured`
   * Interface → `up`
4. Generate PASS/FAIL output.
5. Use the failures in pytest assertions, reports, or dashboards.

This pattern is used extensively in pyATS, Nornir, Netmiko workflows, and WAN interop test automation.

````python
import re

def camel_to_snake(name):
    return re.sub(r'(?<!^)(?=[A-Z])', '_', name).lower()

def normalize_dict(data):
    result = {}

    for key, value in data.items():

        new_key = camel_to_snake(key)

        if isinstance(value, dict):
            result[new_key] = normalize_dict(value)
        else:
            result[new_key] = value

    return result
````
