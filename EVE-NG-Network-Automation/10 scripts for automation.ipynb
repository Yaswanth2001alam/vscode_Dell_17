Script 1: Backup Configurations
What it Does: Connects to a device and saves the running configuration to a file.

Real World Use: Automate daily config backups across your entire network.

python
from netmiko import ConnectHandler
from datetime import datetime

device = {
    'device_type': 'cisco_ios',
    'host': '192.168.1.1',
    'username': 'admin',
    'password': 'cisco123'
}

connection = ConnectHandler(**device)
config = connection.send_command('show running-config')

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
filename = f"backup_{timestamp}.txt"

with open(filename, 'w') as f:
    f.write(config)

print(f"Backup saved to {filename}")
connection.disconnect()

Script 3: Ping Multiple Devices
What it Does: Pings multiple devices and reports which are reachable.

Real World Use: Check network connectivity across all devices after maintenance.

python
import subprocess

devices = ['192.168.1.1', '192.168.1.2', '192.168.1.3']

for ip in devices:
    response = subprocess.run(['ping', '-c', '1', ip], capture_output=True)
    if response.returncode == 0:
        print(f"{ip} is reachable")
    else:
        print(f"{ip} is not reachable")

Script 10: Login to 10 Devices at Once
What it Does: Connects to multiple devices and runs a command on each.

Real World Use: Run the same command across your entire network in seconds. This is where Netmiko's send_command shines.

python
from netmiko import ConnectHandler

devices = [
    {'device_type': 'cisco_ios', 'host': '192.168.1.1', 'username': 'admin', 'password': 'cisco123'},
    {'device_type': 'cisco_ios', 'host': '192.168.1.2', 'username': 'admin', 'password': 'cisco123'},
    {'device_type': 'cisco_ios', 'host': '192.168.1.3', 'username': 'admin', 'password': 'cisco123'}
]

for device in devices:
    try:
        connection = ConnectHandler(**device)
        output = connection.send_command('show version | include uptime')
        print(f"{device['host']}: {output}")
        connection.disconnect()
    except Exception as e:
        print(f"Failed to connect to {device['host']}: {e}")

Will Python Replace CLI in CCNA Exam?
No. But 10% of the exam is automation and programmability.

The CCNA 200-301 exam covers REST APIs, JSON, HTTP codes, Ansible, and Python basics. But you are not expected to write code on the exam. You are expected to understand concepts and interpret scripts.

In the real world, CLI is still essential for troubleshooting. When a network goes down, you need deep understanding of routing protocols and CLI to diagnose the issue. Python is the layer on top of CLI that handles the repetitive tasks.
