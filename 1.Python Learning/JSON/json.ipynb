```python
import json

with open('library.json', 'r') as file:
    catalog23 = json.load(file)

for book in catalog23:
    title = book["title"]
    available = "available" if book["available"] else "not available"
    rating = book["rating"]

    print(f"{title} is { available} and has a {rating} rating")
```

```python
import json

with open('library.1.json', 'r') as file:
    catalog33 = json.load(file)

new_book = {
    "title" : "New Book",
    "available" : True,
    "rating" : 8.5
}

catalog33.append(new_book)

with open("library.1.json", "w") as file:
    json.dump(catalog33, file, indent=2)

with open("library.1.json", 'r') as file:
    updated = json.load(file)

print(updated)
```
