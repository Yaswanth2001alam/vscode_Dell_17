## Concept notes and mental model

Principal component analysis finds orthogonal directions of maximal variance in centered data.
Reducing dimensions projects observations onto a smaller set of those directions. This can
compress correlated features and sometimes help downstream modeling, but high variance is not
necessarily task-relevant information.

PCA centers inputs; feature scaling is a separate choice. A high-magnitude measurement can
dominate components if units are not made comparable. Fit PCA only on training data in an
evaluation pipeline. Reconstruction maps the lower-dimensional representation back to the
original space approximately.

Explained variance measures retained variation, not classification accuracy or causal
importance. Component signs are arbitrary: an equally valid fit can flip a direction and its
scores. For very sparse text, truncated SVD can avoid centering a huge sparse matrix.

## Mathematics or data contract

For centered X, a projection is Z = X @ components.T; inverse transformation approximates X.

## Worked example 1: Compress and reconstruct correlated features

**Purpose:** Inspect retained variance and reconstruction error.

### Read the code in this order
1. Generate a shared latent factor plus small noise.
2. Standardize before PCA.
3. Keep one component.
4. Check dimensions and compare reconstruction error with the standardized variance scale.

### Syntax and behavior to notice
- `fit(...)` learns model or preprocessing state from the supplied training data.
- `transform(...)` applies already-fitted preprocessing; it does not refit it.
- `assert` checks the stated behavior; a failed check raises AssertionError.
- A fixed random seed makes the chosen sampling reproducible within the environment.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
retained variance: 0.9983
reconstruction MSE: 0.0017
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(42)
latent = rng.normal(size=(160, 1))
X = latent @ np.array([[1., 2., -1.]]) + rng.normal(0, 0.05, (160, 3))
standardized = StandardScaler().fit_transform(X)
pca = PCA(n_components=1).fit(standardized)
Z = pca.transform(standardized)
reconstruction = pca.inverse_transform(Z)
error = np.mean((standardized - reconstruction) ** 2)
assert Z.shape == (160, 1) and pca.explained_variance_ratio_[0] > 0.95
assert error < 0.05
print("retained variance:", round(float(pca.explained_variance_ratio_[0]), 4))
print("reconstruction MSE:", round(float(error), 4))
````
