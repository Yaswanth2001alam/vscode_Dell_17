## Concept notes and mental model

**Level:** Intermediate

**Learning goal:** Make domain objects work naturally with repr, bool, len, iteration, lookup, and with.

**Special methods** connect objects to Python syntax. `repr(obj)` uses `__repr__`; `str(obj)`
uses `__str__` when supplied. A useful repr is unambiguous and helps debugging, but should not
disclose credentials or tokens. A string representation is not a serialization contract.

**Truth and size.** `bool(obj)` uses `__bool__`, or falls back to `__len__` when appropriate.
Return an actual boolean from __bool__ and a nonnegative integer from __len__. A domain
choice such as 'healthy results are truthy' must be documented; otherwise write an explicit
`result.healthy` query so readers do not have to guess.

**Iteration** normally works by __iter__ returning an iterator. A collection should usually
create a fresh iterator each time, rather than store one exhausted iterator on the object.
**Subscription** uses __getitem__. A mapping-like API should preserve KeyError for missing
keys rather than silently return a plausible empty record.

**Context managers** model resource lifetime. `with` calls __enter__, runs the block, and calls
__exit__ even when the body raises. __exit__ receives exception information. Returning false
(or None) allows the exception to propagate; true suppresses it. Do not suppress an operational
failure merely to make a notebook cell look successful.

**Equality and hashing.** If objects compare equal, their hashes must agree. A value used as a
set member or dictionary key must not change in a way that changes equality/hash semantics.
For unsupported operand types, an equality method should normally return NotImplemented.
Not every possible operator belongs on a network object: prefer a clear named method when
overloaded syntax would surprise an engineer reading the automation.

**Pause and explain:** Explain what __exit__ should return on failure and why a mutable hash key is dangerous.

## Pattern 040: Context-managed simulated session

**Core concept:** A context manager provides deterministic cleanup even when a block fails.

**Network-engineering purpose:** Session-like resources need explicit lifetime management rather than reliance on __del__.

### Read the implementation in this order
1. Open the simulated state in __enter__.
2. Return self for the with target.
3. Close state in __exit__ and return False.
4. Raise a known body error and verify both propagation and cleanup.

**Common mistake / boundary:** __del__ timing is not a reliable resource cleanup contract.

**Practice after the ten worked scenarios:** Record open/close events and assert their order.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0391 - Context-managed simulated session: branch scenario

**Scenario 01/10; pattern 040; Intermediate.** This is a self-contained offline exercise. Session-like resources need explicit lifetime management rather than reliance on __del__.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Open the simulated state in __enter__. Raise a known body error and verify both propagation and cleanup. Follow the class-specific steps under Pattern 040 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01 session closed after failure
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** __del__ timing is not a reliable resource cleanup contract.

````python
# Example 0391: Context-managed simulated session - branch
case = {'hostname': 'branch-edge-01'}

class Session:
    def __enter__(self):
        self.open = True
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        self.open = False
        return False

session = Session()
try:
    with session:
        assert session.open
        raise RuntimeError("simulated body failure")
except RuntimeError as error:
    assert str(error) == "simulated body failure"
else:
    raise AssertionError("body failure suppressed")
assert session.open is False
print(case["hostname"], "session closed after failure")
````
