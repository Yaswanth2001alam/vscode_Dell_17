## Concept notes and mental model

**Level:** Intermediate

**Learning goal:** Use substitutable subclasses and trace cooperative multiple inheritance.

**Inheritance** creates a derived type from an existing base type. Single inheritance has one
direct base. Multilevel inheritance extends a derived class again. Hierarchical inheritance
has several children of one base. Multiple inheritance has more than one direct base; hybrid
hierarchies combine those shapes. The shape is less important than a coherent public contract.

**Overriding** replaces a base method with a method of the same name in a subclass. Python
dispatches at runtime using the actual object. A list of different device adapters can therefore
answer the same method call differently. This is subtype polymorphism. Duck typing can achieve
interchangeable behavior without an inherited base.

**super() is not simply 'call my parent'.** It creates a proxy that continues lookup after the
current class in the method resolution order (MRO) for the actual instance. In cooperative
multiple inheritance, every participating method calls super consistently and accepts
compatible arguments. Inspect `SomeClass.__mro__` when reasoning about a diamond.

**Mixins** provide a small reusable capability, not a second unrelated domain identity.
Keep their dependencies explicit and avoid conflicting constructor conventions. Prefer
composition if capabilities need complex ownership or independent configuration.

**Substitutability.** A subclass should honor the promises of its base. A read method that
suddenly deletes configuration, or a child that rejects inputs the base promised to support,
violates that expectation even if the code runs. Do not use inheritance merely to reuse a
few lines. Python does not overload methods by parameter signature: writing a second method
with the same name replaces the first definition. Use defaults, keyword-only parameters,
separate names, or explicit dispatch instead.

**Pause and explain:** Trace a diamond MRO; explain why runtime overriding is not signature-based overloading.

## Pattern 021: Specialize a base device

**Core concept:** A derived class inherits existing methods and adds a focused capability.

**Network-engineering purpose:** A router can reuse a common identity operation while adding a routing-related label.

### Read the implementation in this order
1. Define the shared constructor and identity method on Device.
2. Inherit Router from Device.
3. Add a capability without duplicating identity.
4. Check both inherited and specialized methods.

**Common mistake / boundary:** Adding a subclass is useful only if the distinction changes meaningful behavior.

**Practice after the ten worked scenarios:** Create a Switch subclass and process both through the shared identity() contract.

### Example 0201 - Specialize a base device: branch scenario

**Scenario 01/10; pattern 021; Intermediate.** This is a self-contained offline exercise. A router can reuse a common identity operation while adding a routing-related label.

**Local inputs** (all are defined again in the code cell):

````python
# Example 0201: Specialize a base device - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    def identity(self):
        return self.hostname

class Router(Device):
    def role(self):
        return "routing"

router = Router(case["hostname"])
assert router.identity() == case["hostname"]
assert isinstance(router, Device)
print(router.identity(), router.role())
````

## Pattern 022: Extend initialization with super

**Core concept:** A subclass can preserve base invariants while adding fields.

**Network-engineering purpose:** A router identity and local ASN can be initialized through one consistent construction path.

### Read the implementation in this order
1. Keep hostname initialization in the base.
2. Call super().__init__ from the child.
3. Store the child's additional ASN.
4. Check that both layers initialized correctly.

**Common mistake / boundary:** Skipping base initialization can leave inherited methods with missing fields.

**Practice after the ten worked scenarios:** Validate ASN separately and distinguish a lab value from a complete BGP configuration.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

````python
# Example 0211: Extend initialization with super - branch
case = {'hostname': 'branch-edge-01', 'asn': 64512}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

class Router(Device):
    def __init__(self, hostname, asn):
        super().__init__(hostname)
        self.asn = asn

router = Router(case["hostname"], case["asn"])
assert router.hostname == case["hostname"]
assert router.asn == case["asn"]
print(router.hostname, router.asn)
````
