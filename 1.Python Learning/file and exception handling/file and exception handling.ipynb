🧠 1. What is File Handling?  
File handling allows Python to:  
✔️ Read data from files  
✔️ Write data into files  
✔️ Store information permanently  

📂 2. Opening a File  
Use open() to work with files.  

Example:
file = open("data.txt", "r")

📌 "r" means read mode.  

📖 3. Reading a File  

📌 Read Entire File  
Example:
file = open("data.txt", "r")
content = file.read()
print(content)
file.close()

📄 4. Read Line by Line  
Example:
file = open("data.txt", "r")
for line in file:
    print(line)
file.close()

✍️ 5. Writing to a File  
Use "w" mode to write data.  

Example:
file = open("notes.txt", "w")
file.write("Python File Handling")
file.close()

⚠️ "w" overwrites existing data.  

➕ 6. Appending Data  
Use "a" mode to add data without deleting old content.  

Example:
file = open("notes.txt", "a")
file.write("\nNew Line Added")
file.close()

🔒 7. Using with Statement  
Best way to handle files.  

Example:
with open("data.txt", "r") as file:
    content = file.read()
    print(content)

✔️ Automatically closes the file  
✔️ Cleaner and safer code  

📊 8. Working with CSV Files  
CSV files are used for tabular data.  

Example:
import csv
with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)

⚠️ 9. What is Exception Handling?  
Errors can crash programs.  
Exception handling prevents that.  

🛡️ 10. try-except Block  
Example:
try:
    num = int(input("Enter Number: "))
    print(num)
except:
    print("Invalid Input")

📌 If the user enters text instead of number, the program won’t crash.  

⚡ 11. Handling Specific Errors  

Example:
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")

🔄 12. finally Block  
finally always runs whether error occurs or not.  

Example:
try:
    print("Start")
except:
    print("Error")
finally:
    print("Finished")

````python
import csv

with open("network_devices.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["Status"] == "Down":
            print("ALERT:", row["Device Name"], "is DOWN")
````

````python
import csv

vendor_count = {}

with open("network_devices.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        vendor = row["Vendor"]
        vendor_count[vendor] = vendor_count.get(vendor, 0) + 1

print(vendor_count)
````

````python
import csv

with open("network_devices.csv", "r") as infile, open("down_devices.csv", "w", newline="") as outfile:
    reader = csv.DictReader(infile)
    writer = csv.DictWriter(outfile, fieldnames=reader.fieldnames)
    
    writer.writeheader()
    
    for row in reader:
        if row["Status"] == "Down":
            writer.writerow(row)
````
