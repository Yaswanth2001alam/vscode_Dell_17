# 1. ValueError

## What it means

Python got the **correct data type**, but the **value itself is invalid**.

### Network Example: IP Validation

```python
from ipaddress import ip_address

ip_address("10.1.1.300")
```

Output:

```python
ValueError: '10.1.1.300' does not appear to be an IPv4 or IPv6 address
```

### Why?

* String is correct type ✅
* IP value is invalid ❌

Think:

```text
Data Type = Correct
Value = Wrong
```

***

### Another Example

```python
int("router")
```

Output:

```python
ValueError: invalid literal for int()
```

Python expected:

```python
"100"
```

but got:

```python
"router"
```

***

### Handling ValueError

```python
try:
    vlans = int(input("Enter VLAN Number: "))
except ValueError:
    print("Invalid VLAN ID")
```

***

# 2. KeyError

## What it means

You tried to access a dictionary key that doesn't exist.

### Network Example

```python
device = {
    "hostname": "rwa01",
    "ip": "10.1.1.1"
}

print(device["vendor"])
```

Output:

```python
KeyError: 'vendor'
```

***

### Why?

Dictionary contains:

```python
hostname
ip
```

But you're asking for:

```python
vendor
```

which doesn't exist.

***

### Real Network Automation Example

Imagine API result:

```python
interface = {
    "name": "Ethernet1",
    "status": "up"
}
```

You run:

```python
print(interface["speed"])
```

Output:

```python
KeyError: 'speed'
```

because speed isn't returned.

***

### Safe Method

```python
print(interface.get("speed", "Not Found"))
```

Output:

```python
Not Found
```

***

# 3. TypeError

## What it means

Operation is performed on an incompatible data type.

### Example

```python
vlan = "100"

print(vlan + 1)
```

Output:

```python
TypeError
```

***

### Why?

```python
"100"   -> string
1       -> integer
```

Python can't add them directly.

***

### Fix

```python
vlan = int(vlan)

print(vlan + 1)
```

Output:

```python
101
```

***

# 4. NameError

## What it means

Variable does not exist.

### Example

```python
print(hostname)
```

Output:

```python
NameError: name 'hostname' is not defined
```

***

### Fix

```python
hostname = "ewa01"

print(hostname)
```

***

# 5. IndexError

## What it means

Trying to access an index that doesn't exist.

### Network Example

```python
neighbors = ["ewa01", "ewa02"]

print(neighbors[5])
```

Output:

```python
IndexError
```

***

### Why?

Available indexes:

```python
0 -> ewa01
1 -> ewa02
```

No index 5 exists.

***

### Fix

```python
if len(neighbors) > 5:
    print(neighbors[5])
```

***

# 6. FileNotFoundError

Very common for automation scripts.

### Example

```python
with open("bgp_output.txt") as file:
    print(file.read())
```

Output:

```python
FileNotFoundError
```

***

### Why?

File does not exist in current directory.

Check:

```python
pwd
ls
```

before running script.

***

# 7. AttributeError

## What it means

Object doesn't have the method you're trying to use.

***

### Example

```python
hostname = "ewa01"

hostname.append("test")
```

Output:

```python
AttributeError
```

***

### Why?

`.append()`

works on:

```python
list
```

not on:

```python
string
```

***

### Correct

```python
devices = ["ewa01"]

devices.append("ewa02")
```

***

# 10. SyntaxError

## What it means

Python cannot understand your code.

### Example

```python
if True
    print("Hello")
```

Output:

```python

```

***

### Fix


# 11. IndentationError

Extremely common.

### Example

# 12. ConnectionError (Network Automation)

### Example

```python
from netmiko import ConnectHandler

device = {
    "host": "10.1.1.1"
}
```

If device is unreachable:

```python
ConnectionError
```

or

```python
NetmikoTimeoutException
```

***

### Common Causes

* Wrong IP
* Device down
* ACL blocking
* SSH disabled
* Routing issue

***

# Most Important Errors for Network Engineers

If you're automating Cisco, Juniper, Arista, IXIA, Azure networking, focus on these first:

| Error               | Where Seen                  |
| ------------------- | --------------------------- |
| ValueError          | IP validation, VLAN input   |
| KeyError            | JSON/API parsing            |
| TypeError           | String vs Integer issues    |
| FileNotFoundError   | Reading configs/logs        |
| IndexError          | Parsing outputs             |
| NameError           | Variable mistakes           |
| AttributeError      | Working with strings/lists  |
| ModuleNotFoundError | Netmiko/NAPALM installation |
| TimeoutError        | SSH/API connectivity        |
| ConnectionError     | Devices unreachable         |
| SyntaxError         | Coding mistakes             |
| IndentationError    | Python formatting           |

# Real Interview Question

```python
devices = {
    "rwa01": "10.1.1.1"
}

print(devices["ewa01"])
```

**Question:** What error occurs?

✅ Answer:

```python
KeyError
```

Reason: `"ewa01"` key doesn't exist in the dictionary.

***

### Easy way to remember

```text
ValueError     -> Value is wrong
KeyError       -> Dictionary key missing
TypeError      -> Wrong data types
NameError      -> Variable missing
IndexError     -> List index missing
FileNotFound   -> File missing
AttributeError -> Method missing
SyntaxError    -> Code grammar wrong
IndentationError -> Spacing wrong
ConnectionError -> Device unreachable
```

For network automation interviews (Netmiko, NAPALM, REST APIs, JSON parsing), **ValueError, KeyError, TypeError, FileNotFoundError, TimeoutError, and ConnectionError** are the six errors you will encounter most often in real-world scripts.
