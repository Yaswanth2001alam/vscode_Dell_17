## Concept notes and mental model

Boosting builds an additive model in stages. Each stage responds to the current model's errors,
often by fitting a tree to negative loss gradients. This differs from a random forest's largely
independent randomized trees.

learning_rate shrinks each stage's contribution, while the number and complexity of trees
determine capacity. These settings interact: a smaller rate may require more stages. Tune
with validation data or cross-validation and keep the final test set separate.

Gradient boosting is a strong family for tabular problems, not a universal winner. It can
overfit noisy targets or leaked features. Modern libraries implement different objectives,
histogram algorithms, category handling, hardware support, and missing-value policies; do not
assume identical behavior across implementations.

## Mathematics or data contract

An additive update is F_next(x)=F_current(x)+learning_rate*tree(x).

## Worked example 1: Fit a nonlinear numeric relationship

**Purpose:** See boosting outperform a constant baseline on held-out data.

### Expected output

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(42)
X = rng.uniform(-2, 2, (240, 2))
y = X[:, 0] ** 2 + np.sin(X[:, 1]) + rng.normal(0, 0.1, 240)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
model = GradientBoostingRegressor(n_estimators=70, max_depth=2, random_state=42)
model.fit(X_train, y_train)
error = mean_absolute_error(y_test, model.predict(X_test))
baseline = mean_absolute_error(y_test, np.full_like(y_test, y_train.mean()))
assert error < baseline
print("test MAE:", round(error, 3), "baseline:", round(baseline, 3))
````
