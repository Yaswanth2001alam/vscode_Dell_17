When you're connected to **192.168.32.138**, `ip` is `192.168.32.138`, so it runs:

```bash
ping 192.168.32.138
```

from router **192.168.32.138** itself.

When connected to **192.168.32.139**, it runs:

```bash
ping 192.168.32.139
```

from router **192.168.32.139** itself.

So both routers are pinging themselves.

````python
from pathlib import Path

import yaml
from netmiko import ConnectHandler
from netmiko.exceptions import (
    NetmikoAuthenticationException,
    NetmikoTimeoutException,
)


def load_inventory(filename: str) -> list[dict]:
    inventory_path = Path(filename)

    if not inventory_path.exists():
        raise FileNotFoundError(f"Inventory file not found: {filename}")

    with inventory_path.open("r", encoding="utf-8") as file:
        inventory = yaml.safe_load(file)

    return inventory["routers"]


def main() -> None:
    routers = load_inventory("devices.yaml")

    for router in routers:
        device_name = router.pop("name")

        print(f"\n{'=' * 60}")
        print(f"Connecting to {device_name}: {router['host']}")
        print(f"{'=' * 60}")

        try:
            connection = ConnectHandler(**router)
            connection.enable()

            hostname = connection.send_command("show running-config | include hostname")
            interfaces = connection.send_command("show ip interface brief")
            version = connection.send_command("show version | include IOS|uptime")

            print(hostname)
            print(version)
            print(interfaces)

            connection.disconnect()

        except NetmikoAuthenticationException:
            print(f"Authentication failed for {device_name}")

        except NetmikoTimeoutException:
            print(f"Connection timed out for {device_name}")

        except Exception as error:
            print(f"Unexpected error on {device_name}: {error}")


if __name__ == "__main__":
    main()
````

# 2. Understanding `devices.yaml`

Your inventory is:

```yaml
routers:
  - name: R1
    device_type: cisco_ios
    host: 192.168.32.138
    username: admin
    password: admin123
    #secret: EnablePassword123

  - name: R2
    device_type: cisco_ios
    host: 192.168.32.139
    username: admin
    password: admin123
    #secret: EnablePassword123
```

## `routers:`

```yaml
routers:
```

This creates a YAML key named `routers`.

Under it is a list containing R1 and R2.

Python reads this as approximately:

```python
{
    "routers": [
        {
            "name": "R1",
            "device_type": "cisco_ios",
            "host": "192.168.32.138",
            "username": "admin",
            "password": "admin123"
        },
        {
            "name": "R2",
            "device_type": "cisco_ios",
            "host": "192.168.32.139",
            "username": "admin",
            "password": "admin123"
        }
    ]
}
```

## `name`

```yaml
name: R1
```

This is your friendly name for the router. Netmiko does not require it. Your Python program uses it when printing messages.

Example:

```text
Connecting to R1: 192.168.32.138
```

## `device_type`

```yaml
device_type: cisco_ios
```

This tells Netmiko what network operating system it is connecting to.

For these Cisco 7200 IOS routers:

```yaml
device_type: cisco_ios
```

is correct.

Netmiko uses this value to determine:

* The expected command prompt
* How to enter enable mode
* How commands should be sent
* How configuration mode works
* When command output has finished

## `host`

```yaml
host: 192.168.32.138
```

This is the IP address used for SSH.

In your topology:

```text
R1 = 192.168.32.138
R2 = 192.168.32.139
```

## `username` and `password`

```yaml
username: admin
password: admin123
```

These are used for SSH authentication.

They must match the router configuration, such as:

```cisco
username admin privilege 15 secret admin123
```

## `secret`

```yaml
#secret: EnablePassword123
```

The `#` makes this a comment, so Python does not receive a `secret` value.

A Netmiko `secret` is the router's enable password:

```cisco
enable secret EnablePassword123
```

Your connection still succeeded because the `admin` user is probably privilege level 15 and arrives directly at the privileged prompt:

```text
R1#
```

rather than:

```text
R1>
```

You can verify this on the router:

```cisco
show privilege
```

Expected result:

```text
Current privilege level is 15
```

When the user already has privilege 15, `connection.enable()` may determine that it is already in enable mode.

# 17. Improvements to your existing program

The main concern is:

```python
device_name = router.pop("name")
```

`pop()` modifies the original router dictionary.

A safer version is:

```python
device_name = router["name"]

connection_parameters = {
    key: value
    for key, value in router.items()
    if key != "name"
}
```

Here is an improved version:

```python
from pathlib import Path
from typing import Any

import yaml
from netmiko import ConnectHandler
from netmiko.exceptions import (
    NetmikoAuthenticationException,
    NetmikoTimeoutException,
)


def load_inventory(filename: str) -> list[dict[str, Any]]:
    inventory_path = Path(filename)

    if not inventory_path.exists():
        raise FileNotFoundError(
            f"Inventory file not found: {inventory_path.resolve()}"
        )

    with inventory_path.open("r", encoding="utf-8") as file:
        inventory = yaml.safe_load(file)

    if not inventory:
        raise ValueError("The inventory file is empty.")

    if "routers" not in inventory:
        raise KeyError("The YAML file does not contain a 'routers' section.")

    if not isinstance(inventory["routers"], list):
        raise TypeError("'routers' must be a YAML list.")

    return inventory["routers"]


def main() -> None:
    routers = load_inventory("devices.yaml")

    commands = {
        "Hostname": "show running-config | include hostname",
        "Version": "show version | include IOS|uptime",
        "Interfaces": "show ip interface brief",
    }

    for router in routers:
        device_name = router.get("name", router.get("host", "Unknown"))
        connection = None

        connection_parameters = {
            key: value
            for key, value in router.items()
            if key != "name"
        }

        print(f"\n{'=' * 70}")
        print(f"Connecting to {device_name}: {router.get('host')}")
        print(f"{'=' * 70}")

        try:
            connection = ConnectHandler(**connection_parameters)

            if not connection.check_enable_mode():
                connection.enable()

            for section_name, command in commands.items():
                print(f"\n--- {section_name} ---")
                output = connection.send_command(command)
                print(output)

        except NetmikoAuthenticationException:
            print(f"FAIL: Authentication failed for {device_name}")

        except NetmikoTimeoutException:
            print(f"FAIL: Connection timed out for {device_name}")

        except Exception as error:
            print(
                f"FAIL: Unexpected error on {device_name}: "
                f"{type(error).__name__}: {error}"
            )

        finally:
            if connection is not None:
                connection.disconnect()
                print(f"\nDisconnected from {device_name}")


if __name__ == "__main__":
    main()
```

This version adds:

* YAML validation
* No mutation with `pop()`
* Conditional enable mode
* Guaranteed disconnect
* Better section formatting
* Error type printing
* Safer access using `.get()`

# 19. Safe code 2: Back up router configurations

Reading and saving a configuration locally does not alter the router.

```python
from datetime import datetime
from pathlib import Path

import yaml
from netmiko import ConnectHandler


def load_routers(filename: str) -> list[dict]:
    with Path(filename).open("r", encoding="utf-8") as file:
        return yaml.safe_load(file)["routers"]


def create_backup_filename(device_name: str) -> Path:
    timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    backup_directory = Path("backups")
    backup_directory.mkdir(exist_ok=True)

    return backup_directory / f"{device_name}_{timestamp}.cfg"


def main() -> None:
    for router in load_routers("devices.yaml"):
        device_name = router["name"]

        connection_parameters = {
            key: value
            for key, value in router.items()
            if key != "name"
        }

        connection = None

        try:
            print(f"Backing up {device_name}...")

            connection = ConnectHandler(**connection_parameters)

            running_configuration = connection.send_command(
                "show running-config",
                read_timeout=60,
            )

            backup_file = create_backup_filename(device_name)

            backup_file.write_text(
                running_configuration,
                encoding="utf-8",
            )

            print(f"Backup saved: {backup_file.resolve()}")

        except Exception as error:
            print(f"Backup failed for {device_name}: {error}")

        finally:
            if connection:
                connection.disconnect()


if __name__ == "__main__":
    main()
```

Files will be created like:

```text
backups/
├── R1_2026-07-27_23-40-10.cfg
└── R2_2026-07-27_23-40-13.cfg
```

Be careful with backup files because configurations may contain:

* Password hashes
* SNMP communities
* Pre-shared keys
* Internal IP addresses

# 20. Safe code 3: Automatically check interface pass/fail

This creates a simple status report without changing anything.

```python
from pathlib import Path

import yaml
from netmiko import ConnectHandler


def load_routers(filename: str) -> list[dict]:
    with Path(filename).open("r", encoding="utf-8") as file:
        return yaml.safe_load(file)["routers"]


def evaluate_interface_line(line: str) -> str:
    words = line.split()

    if len(words) < 6:
        return "UNKNOWN"

    if "administratively down" in line:
        return "ADMIN-DOWN"

    status = words[-2]
    protocol = words[-1]

    if status == "up" and protocol == "up":
        return "PASS"

    return "CHECK"


def main() -> None:
    for router in load_routers("devices.yaml"):
        device_name = router["name"]

        connection_parameters = {
            key: value
            for key, value in router.items()
            if key != "name"
        }

        connection = None

        try:
            connection = ConnectHandler(**connection_parameters)

            output = connection.send_command("show ip interface brief")

            print(f"\nInterface report for {device_name}")
            print("=" * 75)

            lines = output.splitlines()

            for line in lines[1:]:
                if not line.strip():
                    continue

                result = evaluate_interface_line(line)
                print(f"[{result:10}] {line}")

        except Exception as error:
            print(f"{device_name}: ERROR — {error}")

        finally:
            if connection:
                connection.disconnect()


if __name__ == "__main__":
    main()
```
