```python
from netmiko import ConnectHandler
import time

# =========================
# Device Definitions
# =========================

R1 = {
    "device_type": "cisco_ios",
    "host": "192.168.32.130",
    "username": "admin",
    "password": "Cisco123",
}

R2 = {
    "device_type": "cisco_ios",
    "host": "192.168.32.131",
    "username": "admin",
    "password": "Cisco123",
}

# =========================
# Interface Configurations
# =========================

R1_CONFIG = [
    "interface ethernet1/1",
    "ip address 10.10.10.1 255.255.255.252",
    "no shutdown",
    "exit",
    "router ospf 1",
    "router-id 1.1.1.1",
    "network 10.10.10.0 0.0.0.3 area 0",
]

R2_CONFIG = [
    "interface ethernet1/1",
    "ip address 10.10.10.2 255.255.255.252",
    "no shutdown",
    "exit",
    "router ospf 1",
    "router-id 2.2.2.2",
    "network 10.10.10.0 0.0.0.3 area 0",
]

# =========================
# Configure Router Function
# =========================

def configure_router(device, config, name):
    print(f"\nConnecting to {name}...")

    connection = ConnectHandler(**device)

    output = connection.send_config_set(config)

    print(f"\nConfiguration pushed to {name}:")
    print(output)

    connection.save_config()

    print(f"\nSaved configuration on {name}")

    connection.disconnect()

# =========================
# Verification Function
# =========================

def verify_router(device, name):

    print(f"\nVerifying {name}...")

    connection = ConnectHandler(**device)

    print("\n========== Interface Status ==========")
    interface_output = connection.send_command(
        "show ip interface brief"
    )
    print(interface_output)

    print("\n========== LLDP Neighbors ==========")
    lldp_output = connection.send_command(
        "show lldp neighbors"
    )
    print(lldp_output)

    print("\n========== OSPF Neighbors ==========")
    ospf_output = connection.send_command(
        "show ip ospf neighbor"
    )
    print(ospf_output)

    print("\n========== Ping Test ==========")
    if name == "R1":
        ping_output = connection.send_command(
            "ping 10.10.10.2 repeat 3"
        )
    else:
        ping_output = connection.send_command(
            "ping 10.10.10.1 repeat 3"
        )

    print(ping_output)

    connection.disconnect()

# =========================
# Main Program
# =========================

configure_router(R1, R1_CONFIG, "R1")
configure_router(R2, R2_CONFIG, "R2")

# Give OSPF time to establish
time.sleep(10)

verify_router(R1, "R1")
verify_router(R2, "R2")

print("\nAutomation Completed Successfully")
```
