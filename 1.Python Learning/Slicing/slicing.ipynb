Great question. Understanding **positive vs negative indexing** is one of the most important Python fundamentals because it affects lists, strings, tuples, and many other sequences.

***

# 1. What Is Indexing?

An **index** is the position of an element in a sequence.

Consider this list:

```python
devices = ["R1", "R2", "R3", "R4", "R5"]
```

Python assigns **two sets of indexes**:

```text
Positive Indexes
       0     1     2     3     4
       ↓     ↓     ↓     ↓     ↓
devices = ["R1","R2","R3","R4","R5"]
       ↑     ↑     ↑     ↑     ↑
      -5    -4    -3    -2    -1

Negative Indexes
```

***

# 2. Positive Indexing

Positive indexing starts from the **left side**.

```python
devices = ["R1", "R2", "R3", "R4", "R5"]

print(devices[0])
print(devices[1])
print(devices[2])
```

Output:

```python
R1
R2
R3
```

### Visualization

```text
Index:   0    1    2    3    4

List:   [R1, R2, R3, R4, R5]
          ^
          |
      devices[0]
```

The first element is always index **0**.

***

# 3. Negative Indexing

Negative indexing starts from the **right side**.

```python
devices = ["R1", "R2", "R3", "R4", "R5"]

print(devices[-1])
print(devices[-2])
print(devices[-3])
```

Output:

```python
R5
R4
R3
```

### Visualization

```text
Index:  -5   -4   -3   -2   -1

List:   [R1, R2, R3, R4, R5]
                           ^
                           |
                     devices[-1]
```

The last element is always **-1**.

***

# 4. Side-By-Side Comparison

For:

```python
a = [10, 20, 30, 40, 50]
```

| Positive | Negative | Value |
| -------- | -------- | ----- |
| a\[0]    | a\[-5]   | 10    |
| a\[1]    | a\[-4]   | 20    |
| a\[2]    | a\[-3]   | 30    |
| a\[3]    | a\[-2]   | 40    |
| a\[4]    | a\[-1]   | 50    |

Notice:

```text
0 ↔ -5
1 ↔ -4
2 ↔ -3
3 ↔ -2
4 ↔ -1
```

***

# 5. Why Negative Indexing Exists

Suppose you want the last device.

### Without Negative Indexing

```python
devices[len(devices)-1]
```

Python must:

1. Calculate length
2. Subtract 1
3. Access index

### With Negative Indexing

```python
devices[-1]
```

Much simpler.

# 7. Positive Slicing

### Syntax

```python
list[start:end]
```

Example:

```python
a = [10,20,30,40,50]

print(a[1:4])
```

Output:

```python
[20, 30, 40]
```

Visualization:

```text
Index: 0  1  2  3  4
Value:10 20 30 40 50

      [1---------4)
```

Start included, stop excluded.

# 8. Negative Slicing

Example:

```python
a = [10,20,30,40,50]

print(a[-3:])
```

Output:

```python
[30, 40, 50]
```

Explanation:

```text
-3 represents 30

[30,40,50]
```

# 12. Memory Trick

Think of a train:

```text
[R1][R2][R3][R4][R5]
```

Positive indexing starts from the **front**:

```text
0 → 1 → 2 → 3 → 4
```

Negative indexing starts from the **back**:

```text
-5 ← -4 ← -3 ← -2 ← -1
```

### Quick Rule

```python
a[0]   # first element
a[-1]  # last element

a[1]   # second element
a[-2]  # second-last element

a[2]   # third element
a[-3]  # third-last element
```

For real-world network automation scripts, you'll use **negative indexing far more often** when grabbing the latest log entry, last route learned, last interface in a list, or most recent event without needing to calculate the list length first.
