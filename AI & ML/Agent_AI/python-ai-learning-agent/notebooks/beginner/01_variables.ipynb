## 1. Names refer to objects

A variable is a label, not a box that owns a value exclusively. `backup = count` adds a binding. Reassigning `count` moves only that binding.

**Predict:** will `backup` become 4? Expected output: `4 3`.

```python
count = 3
backup = count
count = 4
print(count, backup)
assert (count, backup) == (4, 3)
```

## 2. Types describe supported operations

`int` represents whole numbers; `float` represents approximate real values; `str` stores text; `bool` stores a truth value; `None` marks absence. Lists collect values in order.

The loop prints one value and its type at a time. Expected type order: int, float, str, bool, NoneType.

```python
values = [24, 0.25, "lab-router", True, None]
for value in values:
    print(repr(value), type(value).__name__)
assert [type(value).__name__ for value in values] == ["int", "float", "str", "bool", "NoneType"]
```

## 3. Parse first, then calculate

Input files often contain text, even when it looks numeric. `int` parses integer text; it raises `ValueError` for invalid text. Parsing is separate from checking whether the result is a valid nonnegative count.

Expected output: 25, then an explained conversion error. Catching the specific expected exception keeps this demonstration executable.

```python
raw_count = "24"
print(int(raw_count) + 1)
assert int(raw_count) + 1 == 25
try:
    int("twenty-four")
except ValueError:
    print("Conversion failed: use integer text such as 24.")
```

## 4. Mutation versus rebinding

`alias = devices` does not copy the list. Appending changes the shared object. Expected output: `['r1', 'r2']`.

**Experiment:** after this cell, try `alias = []`. Predict whether `devices` changes; only the alias binding moves.

```python
devices = ["r1"]
alias = devices
alias.append("r2")
print(devices)
assert devices == ["r1", "r2"]
assert alias is devices
```

## 5. Copying and identity

A shallow copy creates a different outer list. Initially equal contents do not imply shared identity. Expected comparisons before appending: `True False`. After appending, only the copied list contains r3.

For nested lists, inner objects remain shared. Never assume a shallow copy duplicates a whole object graph.

```python
independent = devices.copy()
print(devices == independent, devices is independent)
assert devices == independent and devices is not independent
independent.append("r3")
print(devices, independent)
assert devices == ["r1", "r2"]
assert independent == ["r1", "r2", "r3"]
```

## 6. A numerical foundation

Binary floating-point approximates many decimal fractions. Expected output on this interpreter: equality is False, but `math.isclose` is True. Approximate comparison is useful for measurements; choosing tolerances is a domain decision.

```python
measurement = 0.1 + 0.2
print(measurement, measurement == 0.3, math.isclose(measurement, 0.3))
assert math.isclose(measurement, 0.3)
```

## 7. Visualize bindings

Follow the arrows: devices and alias point to the same list; independent points to another list. This is a conceptual graph, not a diagram of physical RAM addresses.

```python
figure, axes = plt.subplots(figsize=(8, 3))
axes.axis("off")
for label, y in [("devices", 0.8), ("alias", 0.5), ("independent", 0.2)]:
    target = (0.65, 0.65) if label != "independent" else (0.65, 0.2)
    axes.annotate(label, xy=target, xytext=(0.1, y), arrowprops={"arrowstyle": "->"})
axes.text(0.65, 0.65, "['r1', 'r2']", bbox={"facecolor": "lightblue"})
axes.text(0.65, 0.2, "['r1', 'r2', 'r3']", bbox={"facecolor": "lightgreen"})
axes.set_title("Names refer to objects; assignment is not copying")
plt.show()
plt.close(figure)
```

## 8. Measure direct object sizes carefully

The chart uses `sys.getsizeof`, which measures direct object sizes on the current interpreter, not recursive memory consumption. A list's referenced integers are not included in its direct size. Exact bytes depend on Python implementation and build; there is no universal expected bar height.

**Read the axes:** x identifies an object; y is direct bytes. Do not interpret this as total application memory or optimize code based on this tiny experiment alone.

```python
samples = {"integer": 24, "float": 0.25, "text": "router", "list": [1, 2, 3]}
sizes = [sys.getsizeof(value) for value in samples.values()]
assert all(size > 0 for size in sizes)
figure, axes = plt.subplots(figsize=(7, 3))
axes.bar(list(samples), sizes)
axes.set_ylabel("Direct size in bytes (not recursive)")
axes.set_title("Object sizes on this Python interpreter")
figure.tight_layout()
plt.show()
plt.close(figure)
```
