What Each Module Does

platform.system() → Shows operating system (Windows/Linux/macOS)
platform.release() → Shows OS version/release
platform.processor() → Shows CPU information
psutil.cpu_count() → Shows number of CPU cores
psutil.virtual_memory().total → Shows total RAM
psutil.disk_usage('/') → Shows total disk size

This is a good beginner Python project for learning:

Modules and imports
Function calls
System administration scripting
Basic hardware information gathering

# 10. Wi-Fi Network Viewer Without pywifi

Since pywifi is giving errors, use Windows' built-in command:

```python
import subprocess

result = subprocess.run(
    ["netsh", "wlan", "show", "networks"],
    capture_output=True,
    text=True
)

print(result.stdout)
```

No extra packages needed.

# 12. Network Engineer Project: Interface Health Checker

This combines many concepts.

```python
import socket
import psutil
import platform

print("=" * 50)
print("DEVICE HEALTH CHECK")
print("=" * 50)

print(
    "Hostname:",
    socket.gethostname()
)

print(
    "OS:",
    platform.system()
)

print(
    "CPU:",
    psutil.cpu_percent(),
    "%"
)

print(
    "RAM:",
    psutil.virtual_memory().percent,
    "%"
)

print(
    "Disk:",
    psutil.disk_usage('/').percent,
    "%"
)

print("\nINTERFACES")

for name in psutil.net_if_addrs():

    print("-", name)
```
