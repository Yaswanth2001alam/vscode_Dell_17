## Concept notes and mental model

A virtual environment isolates a project's Python packages. It does not provide operating-system
security isolation or make downloaded code trustworthy. Use python -m pip so installation targets
the interpreter you selected. In PowerShell, quoting paths matters when a directory contains spaces.

Suggested commands from this course folder:
```powershell
py -3.13 -m venv .venv
.\.venv\Scripts\python.exe -m pip install -r requirements.txt
.\.venv\Scripts\python.exe -m ipykernel install --user --name agency-ai --display-name "Agency AI"
.\.venv\Scripts\python.exe -m jupyter lab
```
Choose the Agency AI kernel in the notebook UI. A kernel has its own interpreter and in-memory
state; installing packages in another interpreter will not fix its imports. Restart and Run All
exposes accidental dependence on earlier notebook executions.

requirements.txt declares compatible dependency ranges; requirements-tested.txt records the
direct package versions used for this course's validation. Neither is a cryptographic lock of
every transitive dependency. PyTorch wheel support depends on Python, OS, and accelerator
platform; use the official installation selector if the plain CPU learning setup needs adjustment.
All lessons here are CPU-sized and do not download model weights or datasets.

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import importlib.metadata
import sys

packages = ["numpy", "pandas", "scikit-learn", "torch"]
versions = {name: importlib.metadata.version(name) for name in packages}
assert all(versions.values())
print("Python:", sys.version.split()[0])
print("Virtual environment:", sys.prefix != sys.base_prefix)
print("Packages:", versions)
````
