## Concept notes and mental model

A plot makes patterns and mistakes easier to inspect, but its axes and sampling choices affect
interpretation. Scatter plots show relationships; histograms show distributions; line plots
imply an ordering, often time. Label units and avoid connecting unrelated shuffled records.

In regression, residual = actual minus predicted. A residual plot can reveal curvature,
heteroscedasticity, or subgroup structure that a single error average hides. A small residual
does not prove a causal relationship or out-of-sample generalization.

Use the object-oriented Matplotlib API: fig, ax = plt.subplots(), then call methods on ax.
Close figures you no longer need. The example uses a memory buffer so a standalone script
does not open an interactive window or write files into the repository.

## Worked example 1: Render a labeled prediction plot in memory

**Purpose:** Create a reproducible image without external data.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.


**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
from io import BytesIO
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np

X = np.array([10, 20, 30, 40])
y = np.array([5, 8, 10, 14])
prediction = 0.3 * X + 2
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(X, y, label="observed")
ax.plot(X, prediction, label="linear model")
ax.set(xlabel="Utilization (%)", ylabel="Latency (ms)", title="Synthetic latency example")
ax.legend()
buffer = BytesIO()
fig.savefig(buffer, format="png", dpi=90, bbox_inches="tight")
assert buffer.getvalue().startswith(b"\x89PNG")
assert len(ax.lines) == 1
plt.close(fig)
print("PNG generated in memory; MAE:", round(float(np.abs(y - prediction).mean()), 3))
````
