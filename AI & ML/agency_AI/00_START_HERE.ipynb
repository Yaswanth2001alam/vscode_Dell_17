# Start here: AI from Python foundations to modern systems

**55 topic notebooks, 55 matching Python scripts, 70 worked examples, and 5 original diagrams.** Each notebook is intentionally small enough for this course's 256 KiB per-file limit. GitHub rendering depends on GitHub's service behavior, so the matching plain Python file remains a readable fallback.

This is a broad learning curriculum, not an exhaustive account of every AI algorithm, model, research paper, or current product. Simplified simulations are labeled. Core training examples really fit small estimators/networks; no lesson silently substitutes a fake for a claimed pretrained LLM.

## Study route
1. **Beginner foundations, 01-10:** AI vocabulary, Python, HTTP/async, environments, arrays, tables, visualization, and essential math.
2. **Classical ML, 11-24:** datasets/splits, preprocessing, regression, classification algorithms, metrics, model selection, clustering, PCA, and anomalies.
3. **RL and deep learning, 25-31:** bandits, Q-learning, autograd, an actual MLP training loop, CNNs, recurrent sequence models, and local transfer learning.
4. **NLP and language models, 32-45:** normalization, TF-IDF, embeddings/entities, attention, transformers, tokens, next-token training, decoding, prompts, tool schemas, retrieval, evaluation, LoRA/distillation/preferences, quantization, and cache costs.
5. **Advanced systems, 46-55:** generative/multimodal mechanisms, serving, MLOps, responsible evaluation, forecasting, bounded workflows, capstones, and a dated research snapshot.

Follow a notebook's prerequisite list when jumping between topics. No previous notebook execution is required for a worked code cell.

## Run the course
Open this folder in Jupyter/VS Code and choose a Python 3.11+ kernel with the course dependencies. See README.md for the PowerShell setup commands. A notebook can run each example independently. From this folder, a matching script can run as:
```powershell
python ".\examples\13_Linear_Regression.py"
```
Use a quoted path because `AI & ML` contains spaces and an ampersand. Async examples use top-level await in notebooks and asyncio.run at script entry. All examples are local and CPU-sized; there are no paid API calls, credentials, model downloads, remote datasets, listening servers, or real network-device operations.

## Learning method
Read -> predict -> run -> inspect assertions -> modify one input -> explain the result -> complete the extension. Keep a clean train/validation/test boundary, write units and shapes, and say unknown when evidence is missing. A printed result or successful tensor shape is not automatically evidence of model quality.

**Recent-research boundary:** the reading snapshot is dated 2026-10-17. It cites primary papers and provider documentation and does not promise a complete live list of the newest models. Recheck model contracts, licenses, availability, and your own evaluations before deployment.
