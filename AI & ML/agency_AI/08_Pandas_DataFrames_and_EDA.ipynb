## Concept notes and mental model

A DataFrame is a labeled table with columns that may have different types. Inspect shape,
dtypes, missingness, duplicates, units, and target frequency before fitting a model. Exploratory
analysis is investigation, not permission to repeatedly optimize against a held-out test set.

loc selects by labels/conditions; iloc selects positional rows/columns. Filtering does not
guarantee a new independent object; use .copy before deliberate separate mutation. Groupby
summaries can reveal site-level differences hidden by an overall mean.

Missing values are not necessarily zero. A missing latency may mean no measurement or an outage,
and filling it with zero may falsely imply excellent performance. Define a policy, fit numerical
imputers on training data only, and preserve useful missingness indicators where appropriate.
These small descriptive summaries do not train a deployable imputer.

## Mathematics or data contract

A group mean is sum(valid observations)/count(valid observations), not automatically a population truth.

## Important limitation

Tiny descriptive samples are not representative evaluations or causal evidence.

Distinguish what is implemented and asserted below from what would need additional data, training, independent evaluation, authorization, or operational controls.

## Worked example 1: Inspect and summarize a telemetry table

**Purpose:** Practice missingness, filtering, and grouped aggregation.

### Read the code in this order
1. Construct a DataFrame from local records.
2. Count missing measurements explicitly.
3. Group valid latency values by site.
4. Verify the values and keep labels separate from features.

### Syntax and behavior to notice

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
missing: {'site': 0, 'latency_ms': 1, 'outage': 0}

```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import pandas as pd

frame = pd.DataFrame({
    "site": ["a", "a", "b", "b"],
    "latency_ms": [10., 20., 40., None],
    "outage": [0, 0, 1, 1],
})
assert frame.shape == (4, 3)
assert frame["latency_ms"].isna().sum() == 1
summary = frame.groupby("site")["latency_ms"].mean()
assert summary.to_dict() == {"a": 15.0, "b": 40.0}
X = frame[["latency_ms"]].copy()
y = frame["outage"].copy()
assert X.shape == (4, 1) and y.shape == (4,)
print("missing:", frame.isna().sum().to_dict())
print("site means:", summary.to_dict())
````
