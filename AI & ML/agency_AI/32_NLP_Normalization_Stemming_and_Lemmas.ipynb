## Concept notes and mental model

Natural language processing models text or speech structure and meaning for tasks such as
classification, extraction, search, translation, and generation. Preprocessing choices depend
on the task: lowercasing can help lexical matching but remove information useful to names or acronyms.

Tokenization splits input into units. A whitespace or regex tokenizer is simple but does not
reproduce a production language model's subword vocabulary. Stemming removes affixes heuristically
and may produce nonwords; lemmatization maps forms to dictionary-like base forms, often using
part-of-speech context. Neither is universally required for transformer inputs.

Keep negation, numbers, punctuation, and domain identifiers when they carry task meaning.
Removing 'not' can reverse a fault message. A hostname such as edge-01 may need to stay intact.
This exercise uses explicitly labeled toy rules rather than pretending to implement a complete
linguistic stemmer or lemmatizer. Real systems may use language-specific libraries and models.

## Worked example 1: Compare minimal normalization with toy morphology

**Purpose:** Make every text transformation explicit.

### Read the code in this order
1. Normalize case and tokenize while preserving hyphenated identifiers.
2. Apply a deliberately limited suffix rule.
3. Use a tiny explicit irregular lemma dictionary.
4. Assert that negation and host identity survive.

### Syntax and behavior to notice

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
tokens: ['edge-01', 'devices', 'were', 'running', 'not', 'failed']
toy stems: ['edge-01', 'device', 'were', 'running', 'not', 'failed']
dictionary lemmas: ['edge-01', 'device', 'be', 'run', 'not', 'failed']
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import re

text = "EDGE-01 devices were running, not failed."
tokens = re.findall(r"[a-z0-9]+(?:-[a-z0-9]+)*", text.lower())
def toy_stem(token):
    return token[:-1] if token.endswith("s") and len(token) > 3 else token
lemma_dictionary = {"were": "be", "running": "run", "devices": "device"}
stems = [toy_stem(token) for token in tokens]
lemmas = [lemma_dictionary.get(token, token) for token in tokens]
assert "edge-01" in tokens and "not" in tokens
assert "run" in lemmas and "running" in stems
print("tokens:", tokens)
print("toy stems:", stems)
print("dictionary lemmas:", lemmas)
````
