# Alternative Method 2: Measure Download Time Yourself

This is a good learning exercise.

```python
import requests
import time

url = "https://proof.ovh.net/files/100Mb.dat"

start = time.time()

response = requests.get(url)

elapsed = time.time() - start

size_mb = len(response.content) / (1024 * 1024)

speed_mbps = (size_mb * 8) / elapsed

print(f"Download Speed: {speed_mbps:.2f} Mbps")
```

### How it works

1. Downloads a large file.
2. Measures the time taken.
3. Calculates:

```text
Speed = (File Size × 8) / Time
```

This is exactly how many speed tests estimate throughput.

## Fix 2: Import Directly

Sometimes with Python 3.13:

```python
from speedtest import Speedtest

s = Speedtest()

print(s.download() / 1_000_000)
```

works even when:

```python
speedtest.Speedtest()
```

does not.

Try:

```python
from speedtest import Speedtest

s = Speedtest()

download = s.download() / 1_000_000
```
