## Concept notes and mental model

**Level:** Intermediate

**Learning goal:** Model has-a relationships and inject offline collaborators instead of hard-coding I/O.

**Composition** builds a larger behavior from smaller objects: a Device has Interfaces, and a
Collector has a Transport. In ownership terminology, composition often implies that the owner
controls a component's lifecycle; aggregation often describes a reference to an independently
existing component. Python references do not enforce those lifecycle distinctions for you.

**Prefer clear responsibility boundaries.** An interface can validate its own desired MTU.
A transport can send or simulate commands. A service coordinates them. An inventory repository
stores records. A policy decides whether the observed state is acceptable. Combining all of
this in one giant Device class makes testing, reuse, and error reporting harder.

**Dependency injection** means providing a collaborator from outside, often through __init__.
It is not a special library. Passing a fake transport makes tests deterministic and keeps these
notebooks offline. A real adapter would be responsible for authentication, timeouts, host-key
verification, cleanup, and platform-specific behavior. The service should not create a hidden
global SSH connection just because someone instantiates it.

**Ownership matters with mutable inputs.** If the owner promises a snapshot, copy the container.
If it intentionally shares a live collection, document that shared state and its synchronization
rules. Defining ownership is more useful than mechanically copying everything.

**Design question.** Use inheritance for a genuine substitutable is-a relationship. Use
composition for has-a, uses-a, and policies that may vary independently. A router with a
transport and a policy is usually easier to evolve than separate subclasses for every
combination of router platform, transport, region, and validation policy.

**Pause and explain:** Describe which object owns each dependency and how you would replace transport in a unit test.

**Core concept:** Composition groups related objects behind one useful query.

**Network-engineering purpose:** A device-level report can summarize interface state without duplicating interface logic.

### Read the implementation in this order
1. Create Interface objects with names and admin states.
2. Give Device its own list container.

### Example 0151 - A device owns interface objects: branch scenario

**Scenario 01/10; pattern 016; Intermediate.** This is a self-contained offline exercise. A device-level report can summarize interface state without duplicating interface logic.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Create Interface objects with names and admin states. Check the resulting list of enabled names. Follow the class-specific steps under Pattern 016 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['Ethernet1/1']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Copying the container does not clone its Interface objects.

````python
# Example 0151: A device owns interface objects - branch
case = {'interface': 'Ethernet1/1'}

class Interface:
    def __init__(self, name, admin_up):
        self.name, self.admin_up = name, admin_up

class Device:
    def __init__(self, interfaces):
        self.interfaces = list(interfaces)

    def enabled_names(self):
        return [item.name for item in self.interfaces if item.admin_up]

device = Device([Interface(case["interface"], True), Interface("Loopback9", False)])
assert device.enabled_names() == [case["interface"]]
print(device.enabled_names())
````

## Pattern 017: Inject a fake transport

**Core concept:** A service delegates command execution to an injected collaborator.

**Network-engineering purpose:** Collection logic can be exercised without credentials, sockets, or a real device.

### Read the implementation in this order
1. Make a fake whose send method records the request.
2. Inject it through the service constructor.
3. Ask the service to collect a status.
4. Assert both returned data and the exact recorded command.

**Common mistake / boundary:** A fake proves local behavior only; it cannot establish compatibility with a real network OS.

**Practice after the ten worked scenarios:** Make the fake raise TimeoutError and design a caller that reports it explicitly.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0161 - Inject a fake transport: branch scenario

**Scenario 01/10; pattern 017; Intermediate.** This is a self-contained offline exercise. Collection logic can be exercised without credentials, sockets, or a real device.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Make a fake whose send method records the request. Assert both returned data and the exact recorded command. Follow the class-specific steps under Pattern 017 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
up ['show interface Ethernet1/1']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A fake proves local behavior only; it cannot establish compatibility with a real network OS.

````python
# Example 0161: Inject a fake transport - branch
case = {'interface': 'Ethernet1/1'}

class FakeTransport:
    def __init__(self):
        self.calls = []

    def send(self, command):
        self.calls.append(command)
        return "up"

class Collector:
    def __init__(self, transport):
        self.transport = transport

    def status(self, interface):
        return self.transport.send(f"show interface {interface}")

transport = FakeTransport()
result = Collector(transport).status(case["interface"])
assert result == "up"
assert transport.calls == [f"show interface {case['interface']}"]
print(result, transport.calls)
````

## Pattern 020: Repository separates storage from lookup

**Core concept:** A repository exposes a domain-oriented access operation.

**Network-engineering purpose:** Consumers can request a hostname without depending on the backing dictionary's organization.

### Read the implementation in this order
1. Copy the incoming record mapping.
2. Expose a get method with explicit missing-key behavior.
3. Retrieve a known record.
4. Try an unknown hostname and verify KeyError is not swallowed.

**Common mistake / boundary:** Returning a mutable stored record can expose internal state; this example stores immutable strings.

**Practice after the ten worked scenarios:** Add list_by_site without exposing the internal dictionary.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0191 - Repository separates storage from lookup: branch scenario

**Scenario 01/10; pattern 020; Intermediate.** This is a self-contained offline exercise. Consumers can request a hostname without depending on the backing dictionary's organization.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |
| `platform` | `'offline-compact'` |

**Logic to trace:** Copy the incoming record mapping. Try an unknown hostname and verify KeyError is not swallowed. Follow the class-specific steps under Pattern 020 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
missing record reported
offline-compact
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Returning a mutable stored record can expose internal state; this example stores immutable strings.

````python
# Example 0191: Repository separates storage from lookup - branch
case = {'hostname': 'branch-edge-01', 'platform': 'offline-compact'}

class InventoryRepository:
    def __init__(self, records):
        self._records = dict(records)

    def get(self, hostname):
        return self._records[hostname]

repository = InventoryRepository({case["hostname"]: case["platform"]})
assert repository.get(case["hostname"]) == case["platform"]
try:
    repository.get("missing-lab-device")
except KeyError:
    print("missing record reported")
else:
    raise AssertionError("missing record was hidden")
print(repository.get(case["hostname"]))
````
