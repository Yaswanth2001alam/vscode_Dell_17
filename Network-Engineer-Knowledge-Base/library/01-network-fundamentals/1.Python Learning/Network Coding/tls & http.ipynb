````python
import requests
import time

urls = [
    "http://example.com",
    "https://example.com",
    "https://www.google.com",
    "https://www.microsoft.com"
]

for url in urls:
    start = time.perf_counter()

    try:
        response = requests.get(
            url,
            timeout=5
        )

        elapsed = (time.perf_counter() - start) * 1000

        print("\n----------------------------")
        print("URL        :", url)
        print("Status     :", response.status_code)
        print("Final URL  :", response.url)
        print(f"Time       : {elapsed:.2f} ms")
        print("Size       :", len(response.content), "bytes")

    except requests.RequestException as error:
        print(url, "FAILED:", error)
````

````python
import socket
import ssl
from datetime import datetime, timezone

hostname = "www.google.com"

context = ssl.create_default_context()

with socket.create_connection(
    (hostname, 443),
    timeout=5
) as sock:

    with context.wrap_socket(
        sock,
        server_hostname=hostname
    ) as tls:

        cert = tls.getpeercert()

        expiry_string = cert["notAfter"]

        expiry = datetime.strptime(
            expiry_string,
            "%b %d %H:%M:%S %Y %Z"
        ).replace(tzinfo=timezone.utc)

        now = datetime.now(timezone.utc)

        remaining = expiry - now

        print("Hostname     :", hostname)
        print("Expires      :", expiry)
        print("Days Left    :", remaining.days)

        if remaining.days < 30:
            print("Status       : WARNING")
        else:
            print("Status       : OK")
````

````python
import socket
import ssl
````
````python
import time

hostname = "www.google.com"
port = 443

# DNS
start = time.perf_counter()

ip = socket.gethostbyname(hostname)

dns_time = (time.perf_counter() - start) * 1000


# TCP
start = time.perf_counter()

sock = socket.create_connection(
    (ip, port),
    timeout=5
)

tcp_time = (time.perf_counter() - start) * 1000


# TLS
context = ssl.create_default_context()

start = time.perf_counter()

tls = context.wrap_socket(
    sock,
    server_hostname=hostname
)

tls_time = (time.perf_counter() - start) * 1000


print("===== CONNECTION TEST =====")

print("Hostname :", hostname)
print("IP       :", ip)

print(f"DNS      : {dns_time:.2f} ms")
print(f"TCP      : {tcp_time:.2f} ms")
print(f"TLS      : {tls_time:.2f} ms")

print("TLS Ver  :", tls.version())
print("Cipher   :", tls.cipher()[0])

tls.close()
````

````python
import socket
import ssl

hostname = "www.google.com"

context = ssl.create_default_context()

with socket.create_connection(
    (hostname, 443),
    timeout=5
) as sock:

    with context.wrap_socket(
        sock,
        server_hostname=hostname
    ) as tls:

        request = (
            f"GET / HTTP/1.1\r\n"
            f"Host: {hostname}\r\n"
            f"Connection: close\r\n"
            f"\r\n"
        )

        tls.sendall(request.encode())

        while True:
            data = tls.recv(4096)

            if not data:
                break

            print(
                data.decode(
                    errors="replace"
                ),
                end=""
            )
````

### 1. Basic HTTP/HTTPS health check

```python
import requests
import time

urls = [
    "http://example.com",
    "https://example.com",
    "https://www.google.com",
    "https://www.microsoft.com"
]

        print("\n----------------------------")

        print("Status     :", response.status_code)
        print("Final URL  :", response.url)

    except requests.RequestException as error:
        print(url, "FAILED:", error)
```


Example:

```text
===== CONNECTION TEST =====

Hostname : www.google.com
```

```text

  ↓

  ↓

```

If DNS suddenly takes 2 seconds, you know the TCP connection itself isn't necessarily the problem.

### Measure DNS, TCP, TLS and HTTP timing

This is one of the best curl tests to automate:

```python
import subprocess

url = "https://www.google.com"

format_string = (
    "DNS Lookup      : %{time_namelookup}s\n"
    "TCP Connect     : %{time_connect}s\n"
    "TLS Handshake   : %{time_appconnect}s\n"
    "First Byte      : %{time_starttransfer}s\n"
    "Total Time      : %{time_total}s\n"
    "HTTP Status     : %{http_code}\n"
    "Remote IP       : %{remote_ip}\n"
    "Remote Port     : %{remote_port}\n"
)

result = subprocess.run(
    [
        "curl.exe",
        "-s",
        "-o", "NUL",
        "-w", format_string,
        url
    ],
    capture_output=True,
    text=True
)

print(result.stdout)
```

You might get:

```text
DNS Lookup      : 0.012341s
TCP Connect     : 0.031284s
TLS Handshake   : 0.078432s
First Byte      : 0.143221s
Total Time      : 0.167843s
HTTP Status     : 200
Remote IP       : 142.250.x.x
Remote Port     : 443
```

This is extremely useful for isolating where a web connection is slow.

### Test IPv4 vs IPv6

Force IPv4:

```python
import subprocess

result = subprocess.run(
    ["curl.exe", "-4", "-I", "https://www.google.com"],
    capture_output=True,
    text=True
)

print(result.stdout)
```

Force IPv6:

```python
import subprocess

result = subprocess.run(
    ["curl.exe", "-6", "-I", "https://www.google.com"],
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
```

This can expose situations where:

```text
IPv4     PASS
IPv6     FAIL
```

### Test a specific TCP port

```python
import subprocess

targets = [
    "https://www.google.com:443",
    "https://www.microsoft.com:443",
    "https://github.com:443"
]

for target in targets:

    result = subprocess.run(
        [
            "curl.exe",
            "-s",
            "-o", "NUL",
            "-w",
            "%{http_code} %{remote_ip} %{time_total}",
            "--connect-timeout", "5",
            target
        ],
        capture_output=True,
        text=True
    )

    print(target)
    print(result.stdout)
    print()
```

### Test TLS version

Force TLS 1.2:

```python
import subprocess

result = subprocess.run(
    [
        "curl.exe",
        "-v",
        "--tlsv1.2",
        "https://www.google.com"
    ],
    capture_output=True,
    text=True
)

print(result.stderr)
```

On Windows curl, the exact TLS negotiation details depend on the TLS backend shown by `curl.exe --version` (often Schannel).

### Test HTTP methods

GET:

```python
["curl.exe", "-X", "GET", "https://example.com"]
```

HEAD:

```python
["curl.exe", "-I", "https://example.com"]
```

POST against an API **you control or are authorized to use**:

```python
import subprocess

result = subprocess.run(
    [
        "curl.exe",
        "-X", "POST",
        "-H", "Content-Type: application/json",
        "-d", '{"name":"test","status":"ok"}',
        "https://your-api.example/test"
    ],
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
```

### A useful curl diagnostic script

I'd use something like this for network troubleshooting:

```python
import subprocess

url = "https://www.google.com"

curl_format = """
==============================
 CURL NETWORK TEST
==============================
HTTP Status      : %{http_code}
Remote IP        : %{remote_ip}
Remote Port      : %{remote_port}

DNS Lookup       : %{time_namelookup}s
TCP Connection   : %{time_connect}s
TLS Handshake    : %{time_appconnect}s
First Byte       : %{time_starttransfer}s
Total Time       : %{time_total}s

Download Size    : %{size_download} bytes
Download Speed   : %{speed_download} bytes/sec
==============================
"""

result = subprocess.run(
    [
        "curl.exe",
        "-L",
        "-s",
        "-o", "NUL",
        "--connect-timeout", "5",
        "--max-time", "15",
        "-w", curl_format,
        url
    ],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print("CURL TEST FAILED")
    print("Exit Code:", result.returncode)
    print("Error:", result.stderr)
```

That gives you a compact **Layer 4–7 diagnostic**:

```text
==============================
 CURL NETWORK TEST
==============================
HTTP Status      : 200
Remote IP        : 142.250.x.x
Remote Port      : 443

DNS Lookup       : 0.013s
TCP Connection   : 0.032s
TLS Handshake    : 0.081s
First Byte       : 0.147s
Total Time       : 0.169s

Download Size    : 18243 bytes
Download Speed   : 107946 bytes/sec
==============================
```

For network troubleshooting, the most valuable curl tests to learn next are **proxy testing, custom DNS/IP testing (`--resolve`), source-interface testing, certificate validation, HTTP/2 vs HTTP/1.1 comparison, and repeated curl latency tests**.
