🧠 1. What is OOP?  
OOP is a programming approach that organizes code into objects.

Think of: 
🚗 Car
📱 Mobile
🏦 Bank Account  

Each object has:  
✔️ Properties (Data)  
✔️ Behaviors (Functions)

📦 2. What is a Class?  
A class is a blueprint for creating objects.

Example:
class Car:
    pass
Here, "Car" is a class.

🚗 3. Creating an Object  
An object is an instance of a class.

Example:
class Car:
    pass

car1 = Car()
print(type(car1))

📌 Output: <class '*main*.Car'>

⚙️ 4. Attributes in a Class  
Attributes store information.

Example:
class Car:
    brand = "Toyota"

car1 = Car()
print(car1.brand)

📌 Output: Toyota

🔨 5. Constructor ("*init")*  
A constructor runs automatically when an object is created.

Example:
class Car:
    def __init__(self, brand):
        self.brand = brand

car1 = Car("Toyota")
print(car1.brand)
📌 Output: Toyota

🎯 6. Instance Methods  
Methods define object behavior.

Example:
class Car:
    def start(self):
        print("Car Started")

car1 = Car()
car1.start()

📌 Output: Car Started

👨‍👩‍👧‍👦 7. Inheritance  
Inheritance allows one class to use features of another class.

Example:
class Vehicle:
    def move(self):
        print("Vehicle Moving")

class Car(Vehicle):
    pass

car1 = Car()
car1.move()

📌 Output: Vehicle Moving  

✔️ Reuse existing code  
✔️ Reduce duplication

🔒 8. Encapsulation  
Encapsulation hides data from direct access.

Example:
class Account:
    def __init__(self):
        self.__balance = 1000

account = Account()
📌 "__balance" becomes private.

🔄 9. Polymorphism  
Same method name, different behavior.

Example:
class Bird:
    def sound(self):
        print("Bird Sound")

class Dog:
    def sound(self):
        print("Dog Sound")

bird = Bird()
dog = Dog()

bird.sound()
dog.sound()

📌 Output:  
Bird Sound  
Dog Sound

🎭 10. Method Overriding  
Child class can redefine a parent method.

Example:
class Vehicle:
    def move(self):
        print("Vehicle Moving")

class Bike(Vehicle):
    def move(self):
        print("Bike Moving")

bike = Bike()
bike.move()

📌 Output: Bike Moving

```python
class Interface:
    def __init__(self, name, mtu=1500):
        self.name = name
        self.mtu = mtu


    @property
    def mtu(self):
        return self._mtu

    @mtu.setter
    def mtu(self, value):
        if type(value) is not int:
            raise TypeError("MTU must be an integer")

        if not 576 <= value <= 9216:
            raise ValueError("MTU must be between 576 and 9216")

        self._mtu = value

interface = Interface("Ethernet1")

print(interface.mtu)

interface.mtu = 9000
print(interface.mtu)

try:
    interface.mtu = -100
except ValueError as error:
    print(error)

print(interface.mtu)
```

```python
class NetworkDevice:
    def __init__(self, hostname):
        self.hostname = hostname

    def connect(self):
        return f"Connecting to {self.hostname}"


class Router(NetworkDevice):
    def show_routes(self):
        return f"Showing routing table on {self.hostname}"


router = Router("R1")

print(router.connect())
print(router.show_routes())
print(isinstance(router, NetworkDevice))
```
