In Python, an **attribute is a value or variable that belongs to an object or a class**.

For example:

```python
class Router:
    vendor = "Cisco"

r1 = Router()

print(r1.vendor)
```

Here:

- `Router` → class
- `r1` → object/instance
- `vendor` → **attribute**
- `"Cisco"` → value stored in the attribute

You access attributes using the **dot `.` operator**:

```python
object.attribute
```

### 1. Instance attributes

Most commonly, attributes are created inside `__init__()` using `self`.

```python
class Router:
    def __init__(self, hostname, ip):
        self.hostname = hostname
        self.ip = ip

r1 = Router("R1", "10.1.1.1")

print(r1.hostname)
print(r1.ip)
```

Output:

```text
R1
10.1.1.1
```

Here, `hostname` and `ip` are attributes of `r1`.

Think of the object roughly like:

```text
r1
 ├── hostname = "R1"
 └── ip       = "10.1.1.1"
```

You can see these attributes with:

```python
print(r1.__dict__)
```

which gives approximately:

```python
{
    'hostname': 'R1',
    'ip': '10.1.1.1'
}
```

### 2. Attributes vs methods

An **attribute stores information**, while a **method performs an action**.

```python
class Router:
    def __init__(self, hostname):
        self.hostname = hostname

    def connect(self):
        print("Connecting to", self.hostname)

r1 = Router("R1")

print(r1.hostname)
r1.connect()
```

So:

```text
r1.hostname
   ↑
attribute

r1.connect()
   ↑
method
```

### 3. Python can treat attributes dynamically

Unlike some languages, Python usually lets you change attributes:

```python
r1.hostname = "R2"
```

You can even create a new one:

```python
r1.vendor = "Cisco"
```

Now:

```python
print(r1.__dict__)
```

could show:

```python
{
    'hostname': 'R2',
    'vendor': 'Cisco'
}
```

### 4. How Python finds an attribute

When you write:

```python
r1.hostname
```

conceptually Python asks:

> "Does object `r1` have something called `hostname`?"

If it finds it, Python returns its value.

If it doesn't exist:

```python
print(r1.location)
```

you normally get:

```text
AttributeError:
'Router' object has no attribute 'location'
```

This is why you often see errors like:

```text
AttributeError: 'Device' object has no attribute 'ip'
```

It means your code tried to access:

```python
device.ip
```

but Python couldn't find an `ip` attribute on that object.

### Network-engineering example

```python
class NetworkDevice:
    def __init__(self, hostname, ip, vendor):
        self.hostname = hostname
        self.ip = ip
        self.vendor = vendor

device = NetworkDevice(
    "rwa01.str15",
    "10.3.159.10",
    "Cisco"
)

print(device.hostname)
print(device.ip)
print(device.vendor)
```

Here the object represents one network device:

```text
device
│
├── hostname ──> "rwa01.str15"
├── ip       ──> "10.3.159.10"
└── vendor   ──> "Cisco"
```

So the easiest definition to remember is:

> **Attribute = data/information associated with an object.**

And:

```python
object.attribute
```

means **"give me this piece of information belonging to this object."**
