## Concept notes and mental model

A neuron computes a weighted sum plus bias followed by an activation. Stacking linear layers
without nonlinear activations still represents a linear transformation; nonlinearities allow
richer decision boundaries. Common activations include ReLU, sigmoid, and tanh, with different
gradient behavior.

Backpropagation applies the chain rule through a computation graph to calculate parameter
gradients. Automatic differentiation computes these derivatives from recorded operations; it
does not choose a useful objective or guarantee learning success.

In PyTorch, requires_grad tracks differentiable tensors, backward accumulates gradients, and
an optimizer or no_grad update changes parameters. Clear gradients between independent updates.
Calling eval changes some module behaviors but does not itself turn off gradient tracking;
use no_grad or inference_mode for inference as appropriate.

## Worked example 1: Check an automatic gradient and reduce loss

**Purpose:** Make one learning update mathematically transparent.

### Read the code in this order
1. Create a trainable scalar tensor.
2. Build an explicit prediction and squared loss.
3. Compare backward's gradient to the analytical answer.
4. Update without recording the update itself and verify lower loss.

### Syntax and behavior to notice
- `backward()` computes/accumulates gradients through the recorded computation graph.
- `assert` checks the stated behavior; a failed check raises AssertionError.
- `with`/`async with` bounds a resource lifetime or a temporary execution mode.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
gradient: -24.0 loss before/after: 36.0 1.44
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import torch

torch.set_num_threads(1)
weight = torch.tensor(0.0, requires_grad=True)
x, target = torch.tensor(2.0), torch.tensor(6.0)
loss = (weight * x - target) ** 2
loss.backward()
assert weight.grad.item() == -24.0
before = loss.item()
with torch.no_grad():
    weight -= 0.1 * weight.grad
weight.grad = None
after = ((weight * x - target) ** 2).item()
assert after < before
print("gradient:", -24.0, "loss before/after:", round(before, 3), round(after, 3))
````
