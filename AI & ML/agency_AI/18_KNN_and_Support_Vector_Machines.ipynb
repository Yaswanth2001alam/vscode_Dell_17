## Concept notes and mental model

KNN predicts from nearby training examples. It stores the training representation rather than
learning a compact global decision boundary. Choice of k, distance metric, and feature scaling
matter; large datasets can make naive neighbor search expensive.

An SVM seeks a separating boundary with a margin, trading margin violations against regularization.
Kernels allow nonlinear decision boundaries without explicitly constructing every transformed
feature. The RBF kernel depends on a distance scale controlled by gamma.

Standardize numeric features inside a pipeline for both algorithms. Units such as bytes and
milliseconds can otherwise dominate distances differently. Tune k, C, and gamma on development
data. A model's high score on one synthetic geometry does not prove it is best for every dataset.

## Mathematics or data contract

An RBF similarity is exp(-gamma*||x-x_prime||^2); larger gamma makes influence more local.

## Worked example 1: Compare two nonlinear classifiers on the same split

**Purpose:** Keep preprocessing and evaluation conditions consistent.

### Read the code in this order
1. Generate two-moon data with a fixed seed.
2. Reserve a stratified test set.
3. Use a separate scaling pipeline for each classifier.
4. Check held-out scores without selecting a winner from the test set.

### Syntax and behavior to notice

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
held-out accuracies: {'knn': 0.983, 'svm': 0.983}
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X, y = make_moons(n_samples=240, noise=0.12, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)
estimators = {"knn": KNeighborsClassifier(n_neighbors=5), "svm": SVC(C=2, gamma="scale")}
scores = {}
for name, estimator in estimators.items():
    model = make_pipeline(StandardScaler(), estimator).fit(X_train, y_train)
    scores[name] = round(model.score(X_test, y_test), 3)
assert all(score > 0.8 for score in scores.values())
print("held-out accuracies:", scores)
````
