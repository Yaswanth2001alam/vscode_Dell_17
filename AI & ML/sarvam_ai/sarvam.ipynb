### 6. Generate Structured JSON Instead of Normal Text

This is where AI becomes much easier to combine with automation.

Sarvam currently supports structured JSON output using `response_format`, including JSON Schema-constrained responses. 

For a beginner, you can first ask the model for JSON:

```python
import os
import json

from dotenv import load_dotenv
from sarvamai import SarvamAI

load_dotenv()

client = SarvamAI(
    api_subscription_key=os.getenv("SARVAM_API_KEY")
)

prompt = """
Analyze this network event:

Interface Ethernet1/1 changed state to DOWN.
BFD neighbor 10.0.0.2 went DOWN.
BGP neighbor 10.0.0.2 went DOWN.

Return ONLY JSON.

Format:

{
    "severity": "",
    "device_issue": "",
    "affected_protocols": [],
    "probable_root_cause": "",
    "recommended_action": ""
}
"""

response = client.chat.completions(
    model="sarvam-105b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

result = response.choices[0].message.content

print(result)
```

Once you understand that, move to Sarvam's actual structured-output feature with a JSON Schema. That's better than relying only on prompt wording because the API can constrain the output to the schema. 

This becomes useful for automation:

```text
Network event
      ↓
Sarvam AI
      ↓
JSON

{
   "severity": "critical",
   "protocol": "BGP",
   "root_cause": "physical interface failure"
}

      ↓
Python
      ↓
Ticket / Slack / Dashboard / Database
```

```python

    }
]

    try:

        response = client.chat.completions(

            messages=messages
        )

        answer = response.choices[0].message.content

        print("\nNetwork AI:")

            {
                "role": "assistant",
                "content": answer
            }
        )

    except Exception as e:

        print("\nERROR:")
        print(e)
```

Sarvam explicitly supports multi-turn chat by passing the conversation history in `messages`. 

This means you can have a conversation such as:

```text
You:
Explain BGP

AI:
BGP is...

You:
What is the FSM?

AI:
The BGP finite state machine...

You:
My neighbor is stuck in Active. Why?

AI:
Based on what we discussed...

You:
Give me IOS XR commands.

AI:
show bgp neighbors...
```

The model receives the previous conversation because you're continually adding messages to `messages`.
