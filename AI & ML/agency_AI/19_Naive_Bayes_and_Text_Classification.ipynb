## Concept notes and mental model

Bayes' rule relates a class posterior to a class prior and the likelihood of observed features.
Naive Bayes assumes features are conditionally independent given the class. That assumption is
often imperfect, but the method can be a useful fast baseline.

Multinomial Naive Bayes suits nonnegative count-like text features; Gaussian Naive Bayes models
continuous features with class-conditional Gaussian assumptions. They are different likelihood
models, not interchangeable preprocessing choices. Smoothing prevents an unseen token from
forcing an entire class likelihood to zero.

Fit the vocabulary using training text only. Words from held-out documents must not influence
feature construction during evaluation. Tiny hand-written phrases below demonstrate mechanics,
not sentiment understanding, semantic reasoning, or real-world reliability.

## Worked example 1: Classify synthetic operational messages

**Purpose:** Train a vocabulary and probabilistic classifier together.

### Read the code in this order
1. Create training messages and labels.
2. Fit CountVectorizer inside a Pipeline.
3. Predict on separate messages.
4. Assert expected toy outcomes and probability normalization.

### Syntax and behavior to notice
- `fit(...)` learns model or preprocessing state from the supplied training data.
- `predict(...)` returns the estimator's chosen output, not a guarantee of correctness.
- `predict_proba(...)` returns estimated class probabilities in the estimator's class order.
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
labels: [1, 0]
probabilities: [[0.143, 0.857], [0.947, 0.053]]
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

messages = ["link down outage", "router failed down", "packet loss outage",
            "link healthy stable", "router healthy normal", "traffic stable normal"]
labels = [1, 1, 1, 0, 0, 0]
model = make_pipeline(CountVectorizer(), MultinomialNB(alpha=1.0))
model.fit(messages, labels)
test = ["link down failed", "healthy stable traffic"]
predictions = model.predict(test)
probabilities = model.predict_proba(test)
assert predictions.tolist() == [1, 0]
np.testing.assert_allclose(probabilities.sum(axis=1), 1)
print("labels:", predictions.tolist())
print("probabilities:", np.round(probabilities, 3).tolist())
````
