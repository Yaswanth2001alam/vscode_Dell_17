Python Tip: Remove Duplicates While Keeping Order

A common task in Python is removing duplicates from a list.
Example list:
[1, 2, 2, 3, 4, 3, 5, 1]

Most of us immediately think of using a set:
 - list(set(numbers))

But there is a problem.
A set does not preserve order.

Python provides a clean trick using dict.fromkeys().
unique_numbers = list(dict.fromkeys(numbers))

How it works:
 • Dictionaries cannot have duplicate keys.
 • Python dictionaries preserve insertion order.
 • fromkeys() creates dictionary keys from list values.

So duplicates are automatically removed while keeping the original order.
Result:
 [1, 2, 3, 4, 5]

Simple trick.
Very useful when cleaning data or preparing lists.








```python
numbers = [1,2,3,4,5,3,1,2,6,7,0,9,8,0,345]

unique_numbers = list(dict.fromkeys(numbers))
print(unique_numbers)
```
