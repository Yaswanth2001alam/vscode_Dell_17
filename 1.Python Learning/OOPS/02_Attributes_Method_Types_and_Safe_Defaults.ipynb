## Concept notes and mental model

**Level:** Beginner

**Learning goal:** Choose instance/class/static methods and avoid accidentally shared mutable state.

**Instance methods** receive `self` and are appropriate when behavior depends on one object's
state. `device.describe()` binds the instance to the method. Conceptually, calling an ordinary
method through an instance resembles `Device.describe(device)`.

**Class methods** use `@classmethod` and receive `cls`, the class through which they were called.
Alternative constructors such as `Device.from_record(record)` are a good use: return `cls(...)`
so a subclass can inherit the constructor. **Static methods** use `@staticmethod`, receive
neither implicit instance nor class, and group a small utility with the concept it belongs to.
A module-level function is often simpler if no such grouping adds value.

**Class versus instance attributes.** A class attribute can represent shared metadata such as
a platform family. Reading usually falls back from the instance to its class. Assigning
`obj.family = ...` normally creates an instance attribute, shadowing the class value. Mutating
a list found on the class changes that very list for all readers. Those are different actions.

**Defaults are evaluated at function definition.** A list used as a default argument is reused
across calls. Use `None` as the default and allocate inside the method, or use a dataclass
`default_factory`. Also copy an incoming list if the object should own its own container.
A shallow copy separates the outer list but still shares nested objects.

**Scope and debugging.** A parameter is local to its call. `self.name` is an instance attribute,
and `Device.family` is an explicit class lookup. Avoid globals that quietly link notebook
examples. Every example in this course defines its dependencies again so it can be rerun
without relying on earlier device objects.

**Pause and explain:** Choose the method type for a parser, a per-device operation, and an alternative constructor.

## Pattern 006: Class metadata and instance shadowing

**Core concept:** A shared class value and an instance override have different lifetimes.

**Network-engineering purpose:** A platform default can be overridden for one inventory record without changing every record.

### Read the implementation in this order
1. Put family on the class.
2. Create two instances.
3. Override family on only the first.
4. Read the class and second instance to prove the default remains unchanged.

**Common mistake / boundary:** Mutating a shared collection is not the same as assigning an instance attribute.

**Practice after the ten worked scenarios:** Delete the override and predict which value will be read next.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

## Pattern 007: Alternative constructor with classmethod

**Core concept:** A named constructor turns external data into a domain object.

**Network-engineering purpose:** Inventory imports often contain strings that need normalization at a clear boundary.

### Read the implementation in this order
1. Decorate from_record with classmethod.
2. Read and normalize the external hostname.
3. Return cls rather than hard-coding Device.
4. Check the initialized object.

**Common mistake / boundary:** Do not trust a parsed record merely because it came from a JSON or CSV file.

**Practice after the ten worked scenarios:** Extend from_record to raise ValueError for an empty normalized hostname.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0061 - Alternative constructor with classmethod: branch scenario

**Scenario 01/10; pattern 007; Beginner.** This is a self-contained offline exercise. Inventory imports often contain strings that need normalization at a clear boundary.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Decorate from_record with classmethod. Check the initialized object. Follow the class-specific steps under Pattern 007 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not trust a parsed record merely because it came from a JSON or CSV file.

````python
# Example 0061: Alternative constructor with classmethod - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    @classmethod
    def from_record(cls, record):
        return cls(record["hostname"].strip().lower())

device = Device.from_record({"hostname": "  " + case["hostname"].upper() + "  "})
assert device.hostname == case["hostname"]
print(device.hostname)
````

## Pattern 010: Keyword-only operational options

**Core concept:** A star in a signature forces named arguments for important options.

**Network-engineering purpose:** A dry-run flag should not be confused with an unrelated positional argument.

### Read the implementation in this order
1. Accept the device identity positionally.
2. Require dry_run after the star.
3. Record a plan without applying a change.
4. Confirm the result explicitly says plan or apply.

**Common mistake / boundary:** A flag name alone is not a safety system; production needs approval, authorization, and verification.

**Practice after the ten worked scenarios:** Add a keyword-only change_id and validate it for apply mode.
