## Beginner -> intermediate -> advanced: notes and 1,000 worked scenario examples

**Course layout:** 20 chapters, 100 distinct worked patterns, and **10 explicit scenario datasets per
pattern = 1,000 runnable examples**. The ten versions of a pattern are deliberately related practice,
not 1,000 unrelated algorithms. Each version contains its full implementation and local input data.
Scenario variations cover empty/full capacity, threshold equality, above/below limits, valid boundary
values, default/overridden ports, IPv4/IPv6 assignments, approval/denial, dry-run/apply simulation,
different retry counts, and different concurrency budgets. Not every topic uses every field.

### How to use these notes
1. Read a chapter's concept notes, then its five worked-pattern explanations.
2. Run the first scenario under a pattern and compare the stored output with the reasoning.
3. Change one input, predict what will happen, and rerun that cell. Assertions stop on unexpected behavior.
4. Try the remaining scenarios, especially boundary and negative cases. Complete the extension prompt
   yourself; those prompts are practice tasks, not additional counted examples.
5. Use **Restart Kernel and Run All** for a clean sequential pass, or execute an example independently.
   A code cell's `case` dictionary contains the local inputs needed for that example.

**Prerequisites:** variables, strings, integers, booleans, lists, dictionaries, loops, functions,
indentation, and basic try/except. Read `x["key"]` as dictionary lookup, `obj.name` as attribute access,
`f"{value}"` as string interpolation, and `assert condition` as a learning/test check that raises when
false. `@name` is decorator syntax. An annotation such as `value: int` describes intended type, not
automatic runtime enforcement. Follow the beginner chapters in order before the advanced sections.

**Environment:** Python **3.11 or newer**, using only the standard library in the learning cells.
`TaskGroup` requires 3.11+. Async examples use notebook top-level `await`; a normal script instead
needs an async entry point called with `asyncio.run(...)`. No pip installation is required for the
examples in a working Python Jupyter kernel. Generated outputs were checked on Python 3.13.14.

**Safety and scope:** all adapters, commands, approvals, and device states are offline simulations.
No SSH, HTTP requests, sockets, subprocesses, credentials, real device writes, real sleeps, or external
files are used by the learning examples. Documentation addresses are synthetic lab data. Simplified
MTU/VLAN policies and command strings are not vendor specifications. Expected exception paths are
caught narrowly and checked; unexpected errors must still stop execution. Assertions here are tests,
not a replacement for runtime validation in production.

### OOP's four common teaching pillars
| Concept | Meaning | Networking illustration | Important qualification |
| --- | --- | --- | --- |
| Encapsulation | Put state behind a deliberate public interface | Validate an interface MTU through a setter | Underscores are conventions, not security |
| Abstraction | Expose only the required contract | A collector calls transport.send() | A contract also needs error and lifecycle semantics |
| Inheritance | Extend a substitutable base type | An offline router model extends a device model | Prefer composition for independent capabilities |
| Polymorphism | Different objects satisfy the same operation | Different adapters render through command() | A shared base is not always required |

## Glossary and decision guide

| Term | Practical meaning |
| --- | --- |
| Class / instance | A type definition / one object created from it |
| Attribute / method | A named value on an object / behavior obtained through an object or class |
| State / invariant | Current values / rules that must remain true |
| self / cls | Conventional instance parameter / class parameter |
| Identity / equality | The same object (`is`) / values considered equivalent (`==`) |
| Public / internal | Supported caller contract / implementation detail subject to change |
| Composition / aggregation | Owned collaboration / references to independently owned objects |
| Inheritance / overriding | Deriving a type / replacing an inherited operation |
| Duck typing / Protocol | Using compatible behavior / describing that shape for static checking |
| ABC | Nominal contract that can prevent incomplete instantiation |
| Descriptor | Object controlling attribute access through the descriptor protocol |
| MRO | Ordered class lookup path used by cooperative super() |
| Dataclass / value object | Generated record methods / object whose meaning is its value |
| Idempotency | Repetition does not add another intended final effect |
| Dependency injection | Supply collaborators externally rather than creating hidden dependencies |
| Context manager | Structured entry/exit and cleanup around a block |
| Coroutine / task | Awaitable async operation / scheduled execution of a coroutine |
| Serialization / schema | Data representation / rules for interpreting that representation |
| Precondition / postcondition | Requirement before an operation / guarantee checked after it |

**Choose the smallest tool:** plain function for a stateless conversion; dataclass for structured
values; ordinary class for state plus invariants; composition for interchangeable collaborators;
Protocol for structural contracts; ABC for enforced nominal extension points; inheritance for real
substitutability; descriptor for repeated attribute-validation machinery; metaclass only if ordinary
construction and __init_subclass__ cannot solve a demonstrated requirement.

**Common mistakes to recognize:** forgetting self; returning a value from __init__; confusing print
with return; mutating a class-level list; using a list as a default argument; assigning a property from
inside its own setter; treating `__private` as secure; confusing is with ==; implementing equality
without considering hashability; expecting signature-based method overloading; using deep copy on a
session; suppressing exceptions in __exit__; treating annotations as validators; making a blocking
call inside async def; forgetting to inspect Future results; calling asyncio.run in a running notebook
loop; retrying a non-idempotent write; classifying missing telemetry as healthy.



## Slow walkthrough: how to read your first class

Use Example 0001 as your reference. Do not try to memorize every special method before you understand
what happens in this small program.

1. **`class Device:`** executes a class definition and binds the resulting class object to the name
   Device. Indentation places the following functions in that class's namespace. The class is not a
   physical network device and defining it does not create a connection.
2. **`def __init__(self, hostname):`** defines the initializer. After a normal instance is created,
   Python calls this method with the instance and your supplied hostname. You do not supply self
   separately when calling Device(...). The object may receive many initial values; this first
   example intentionally starts with just one.
3. **`self.hostname = hostname`** reads the parameter on the right and stores it as an attribute on
   the instance on the left. The two uses of hostname have different roles. Calling the parameter
   name something else would work too; consistent naming simply makes the relationship clear.
4. **`def label(self):`** defines an instance method that needs no additional explicit arguments.
   It can read the instance's hostname through self. Local variables from an earlier method call are
   not automatically available here.
5. **`return f"device:{self.hostname}"`** produces a new string. The f-string evaluates the expression
   inside braces. The return exits this method and sends the string to the caller; it does not itself
   display anything.
6. **`device = Device(case["hostname"])`** first looks up a string in the local input dictionary, then
   calls the class, then binds the resulting instance to the variable device. The lowercase variable
   name and the uppercase class name are a convention, not two unrelated kinds of syntax.
7. **`device.label()`** retrieves a bound method and calls it. Python supplies device as self. The
   same class can serve many objects because each call uses the particular instance's attributes.
8. **`assert ...`** compares the actual returned value with the expected one. A false assertion raises
   AssertionError. It is an executable learning check, not an operational input-validation mechanism.
9. **`print(device.label())`** calls the method again and displays its returned string. Nothing about
   that print changes the stored hostname. A method that prints instead of returning would return
   None unless it has an explicit return statement.

### Follow state, not just syntax
Before construction there is no device instance in this example. After construction, its hostname is
the input string. Calling label reads that state without changing it. By contrast, the Interface
shutdown example intentionally changes admin_up from True to False. Classify each method you write as
a query, a command that mutates state, a conversion, or an I/O boundary. Clear intent is more valuable
than making every method look alike.

### Procedural code versus an object model
A dictionary plus a function can solve the first example. OOP becomes useful when related operations
share state and must protect common rules: constructing an interface, changing its MTU, rendering its
desired state, and checking a postcondition all need consistent values. Do not wrap every standalone
function in a class merely to call the result object-oriented. It is normal for good Python programs
to combine functions, classes, comprehensions, and immutable records.

### Four pillars in one simulated change
Encapsulation validates a proposed value before storing it. Abstraction lets a service call a transport
without knowing its library. Inheritance can specialize a true device subtype. Polymorphism lets the
same service call different compatible transports. Composition connects a service, a policy, and a
transport without requiring one subclass for every combination. None of these ideas requires
credentials or live I/O to learn and test.

## Design principles: use these as questions, not slogans

| Principle | Question to ask in a network automation review | Concrete application |
| --- | --- | --- |
| Single responsibility | Does this class have one coherent reason to change? | Separate inventory parsing, transport, policy, and reporting |
| Open/closed | Can a new behavior be supplied without editing a stable client? | Inject a new policy implementation into the existing evaluator |
| Liskov substitution | Can this child honor what callers of the base were promised? | Do not make a read-only reader mutate configuration |
| Interface segregation | Does a client depend only on operations it needs? | A Collector needs read/send, not deployment and billing methods |
| Dependency inversion | Does orchestration depend on a small contract instead of a concrete library? | Type against a Transport protocol and supply an adapter at startup |

These are the SOLID principles in practical form. They do not require five extra classes in every
script. A function can satisfy a dependency; a dataclass can be a value without a repository; two
well-named methods may be clearer than a generalized framework. A good abstraction removes real
duplication or isolates a real change boundary. An abstraction that has only speculative future uses
adds cognitive cost without present benefit.

### Equality, hashing, and operator decisions
Use `is` to ask whether two references point to the same object, and `==` to ask whether two values
compare equal. The frozen RouteKey example demonstrates equal independently created values, matching
hashes, and set deduplication. Ordinary mutable dataclasses are generally not hashable by default
because mutation could invalidate a hash-based collection's assumptions.

If you write `__eq__` manually, compare the intended domain key and normally return NotImplemented
for an unsupported other type. NotImplemented is a protocol signal, not the NotImplementedError
exception. If you supply `__hash__`, equal values must have equal hashes and the compared key must
remain stable while stored in a set or dictionary. Do not use a hash as a unique identifier or assume
its value is stable across Python processes.

Operator overloading gives syntax such as addition meaning through methods such as `__add__` and
`__radd__`. Use it only for natural domain arithmetic with defined units and compatible types:
adding two compatible traffic totals could make sense, while adding two live SSH sessions probably
does not. Return a new value rather than surprising callers by mutating an operand. Prefer a named
method when a symbol would hide units, network scope, or operational side effects. `__call__` instead
lets an instance be invoked like a function; the backoff policy shows this directly with policy(index).

### Overloading, overriding, and constructors are different
Overriding changes an inherited method in a subclass. Operator overloading implements a special
method for Python syntax. Signature-based method overloading, as seen in some other languages, is not
how ordinary Python class definitions work: a second same-named def replaces the earlier binding.
Use defaults or distinct method names, and remember that typing overload declarations are static
descriptions, not multiple runtime implementations. Named classmethod constructors express multiple
ways to build an object without redefining __init__ repeatedly.

### Lifecycle and errors
Creating an object should normally validate its configuration, not secretly perform a slow login.
Make external actions explicit. Resource cleanup belongs in with/async with or a deliberate close
method. An object's destructor is not a dependable place to commit changes or close a production
session, because finalization timing and shutdown conditions can differ.

Distinguish programming errors, invalid input, policy rejection, transient transport failure, and
failed postchecks. A message that says success while returning no data after a timeout is misleading.
These examples display expected rejections explicitly and assert the failure behavior. Their fakes
cannot prove a real device supports a command or that a real rollout is safe.

# Chapter 01: Objects, classes, and state

**Level:** Beginner

**Learning goal:** Read a class definition, build independent objects, and trace changes to instance state.

**What OOP means.** Object-oriented programming organizes a program around objects that combine
information (attributes) with operations (methods). A network device is a useful model: it has
a hostname and interface state, and it can produce a description or validate a proposed change.
An object here represents a device; it is not the physical device and cannot reach one unless
you deliberately give it an I/O implementation. These lessons only use in-memory simulations.

**Essential vocabulary.** A class defines a type and its behavior. An instance is one object of
that type. An attribute is a name accessed with a dot. A method is a function obtained through
an object or class. State is the collection of current values. An invariant is a rule that must
remain true, such as a nonnegative counter. An interface is the public contract callers use.

**Read the syntax slowly.** `class Device:` starts an indented class body. A `def` within it
defines a method. Calling `Device("edge-1")` creates an instance and normally initializes it
through `__init__`. More precisely, `__new__` creates the instance and `__init__` initializes
it; `__init__` must return `None`. `self` is the conventional name for the instance passed to a
bound method. It is not a keyword. `self.hostname = hostname` stores the argument on that object.
`return` hands a value to the caller; `print` displays a value. They are not interchangeable.

**A trace to keep in mind.** Create an object -> initialize its fields -> call a method -> read
or update its state -> return a result -> check the result. Two calls to a class normally
create two independent instances. Assigning an existing object to another variable does not
create a copy; it creates another reference to the same object.

**When a class helps.** Use one when a concept has meaningful state, invariants, interchangeable
behavior, or a resource lifetime. A one-line conversion usually deserves a plain function, not
a class with no purpose. Start with a small model, not a hierarchy of Router/Platform/Region/
Organization subclasses. Model the behavior the automation actually needs.

**Pause and explain:** Explain class versus instance; trace `self`; distinguish a returned result from printed output.

## Pattern 001: Create a device instance

**Core concept:** A class bundles identity and a small public operation.

**Network-engineering purpose:** An inventory object gives later checks one consistent hostname instead of unrelated variables.

### Read the implementation in this order
1. Define Device before calling it.
2. Store hostname on self in __init__.
3. Call label() through the instance; Python supplies self.
4. Assert the returned string and print it.

**Common mistake / boundary:** Leaving off self when assigning hostname creates only a temporary local variable.

**Practice after the ten worked scenarios:** Add a role field and include it in label() without adding a second class.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0001 - Create a device instance: branch scenario

**Scenario 01/10; pattern 001; Beginner.** This is a self-contained offline exercise. An inventory object gives later checks one consistent hostname instead of unrelated variables.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Define Device before calling it. Assert the returned string and print it. Follow the class-specific steps under Pattern 001 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
device:branch-edge-01
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Leaving off self when assigning hostname creates only a temporary local variable.

````python
# Example 0001: Create a device instance - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    def label(self):
        return f"device:{self.hostname}"

device = Device(case["hostname"])
assert device.label() == "device:" + case["hostname"]
print(device.label())
````

## Pattern 003: An instance method changes state

**Core concept:** Methods can enforce a controlled state transition.

**Network-engineering purpose:** An administrative state model separates intended shutdown state from real operational telemetry.

### Read the implementation in this order
1. Initialize the interface as administratively up.
2. Use shutdown() to update that instance.
3. Return the new state for a caller to inspect.
4. Check the state before and after the operation.

**Common mistake / boundary:** Changing this Python attribute does not configure a switch; admin state is also not link state.

**Practice after the ten worked scenarios:** Add no_shutdown() and demonstrate that repeated calls are idempotent.

**Expected printed output:**


**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. **Remember:** Changing this Python attribute does not configure a switch; admin state is also not link state.

````python
# Example 0021: An instance method changes state - branch
case = {'interface': 'Ethernet1/1'}

class Interface:
    def __init__(self, name):
        self.name = name
        self.admin_up = True

    def shutdown(self):
        self.admin_up = False
        return self.admin_up

interface = Interface(case["interface"])
assert interface.admin_up is True
assert interface.shutdown() is False
print(interface.name, interface.admin_up)
````

## Pattern 005: Return a calculated result

**Core concept:** An object can compute a value from state without mutating it.

**Network-engineering purpose:** A port-capacity model reports remaining lab capacity and makes its arithmetic testable.

### Read the implementation in this order
1. Store total and used capacity.
2. Compute remaining capacity in a method.
3. Check the arithmetic independently.
4. Verify the query did not change used capacity.

**Common mistake / boundary:** A query that unexpectedly changes state is difficult to reason about and rerun.

**Practice after the ten worked scenarios:** Validate that used is between zero and total, then add a utilization percentage query.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0041 - Return a calculated result: branch scenario

**Scenario 01/10; pattern 005; Beginner.** This is a self-contained offline exercise. A port-capacity model reports remaining lab capacity and makes its arithmetic testable.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `capacity` | `24` |
| `used` | `0` |

**Logic to trace:** Store total and used capacity. Verify the query did not change used capacity. Follow the class-specific steps under Pattern 005 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
free ports: 24
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A query that unexpectedly changes state is difficult to reason about and rerun.

````python
# Example 0041: Return a calculated result - branch
case = {'capacity': 24, 'used': 0}

class PortCapacity:
    def __init__(self, total, used):
        self.total = total
        self.used = used

    def remaining(self):
        return self.total - self.used

ports = PortCapacity(case["capacity"], case["used"])
assert ports.remaining() == case["capacity"] - case["used"]
assert ports.used == case["used"]
print("free ports:", ports.remaining())
````

# Chapter 02: Method types, scope, and safe defaults

**Level:** Beginner

**Learning goal:** Choose instance/class/static methods and avoid accidentally shared mutable state.

**Instance methods** receive `self` and are appropriate when behavior depends on one object's
state. `device.describe()` binds the instance to the method. Conceptually, calling an ordinary
method through an instance resembles `Device.describe(device)`.

**Class methods** use `@classmethod` and receive `cls`, the class through which they were called.
Alternative constructors such as `Device.from_record(record)` are a good use: return `cls(...)`
so a subclass can inherit the constructor. **Static methods** use `@staticmethod`, receive
neither implicit instance nor class, and group a small utility with the concept it belongs to.
A module-level function is often simpler if no such grouping adds value.

**Class versus instance attributes.** A class attribute can represent shared metadata such as
a platform family. Reading usually falls back from the instance to its class. Assigning
`obj.family = ...` normally creates an instance attribute, shadowing the class value. Mutating
a list found on the class changes that very list for all readers. Those are different actions.

**Defaults are evaluated at function definition.** A list used as a default argument is reused
across calls. Use `None` as the default and allocate inside the method, or use a dataclass
`default_factory`. Also copy an incoming list if the object should own its own container.
A shallow copy separates the outer list but still shares nested objects.

**Scope and debugging.** A parameter is local to its call. `self.name` is an instance attribute,
and `Device.family` is an explicit class lookup. Avoid globals that quietly link notebook
examples. Every example in this course defines its dependencies again so it can be rerun
without relying on earlier device objects.

**Pause and explain:** Choose the method type for a parser, a per-device operation, and an alternative constructor.

## Pattern 006: Class metadata and instance shadowing

**Core concept:** A shared class value and an instance override have different lifetimes.

**Network-engineering purpose:** A platform default can be overridden for one inventory record without changing every record.

### Read the implementation in this order
1. Put family on the class.
2. Create two instances.
3. Override family on only the first.
4. Read the class and second instance to prove the default remains unchanged.

**Common mistake / boundary:** Mutating a shared collection is not the same as assigning an instance attribute.

## Pattern 007: Alternative constructor with classmethod

**Core concept:** A named constructor turns external data into a domain object.

**Network-engineering purpose:** Inventory imports often contain strings that need normalization at a clear boundary.

### Read the implementation in this order
1. Decorate from_record with classmethod.
2. Read and normalize the external hostname.
3. Return cls rather than hard-coding Device.
4. Check the initialized object.

**Common mistake / boundary:** Do not trust a parsed record merely because it came from a JSON or CSV file.

**Practice after the ten worked scenarios:** Extend from_record to raise ValueError for an empty normalized hostname.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0061 - Alternative constructor with classmethod: branch scenario

**Scenario 01/10; pattern 007; Beginner.** This is a self-contained offline exercise. Inventory imports often contain strings that need normalization at a clear boundary.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Decorate from_record with classmethod. Check the initialized object. Follow the class-specific steps under Pattern 007 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not trust a parsed record merely because it came from a JSON or CSV file.

````python
# Example 0061: Alternative constructor with classmethod - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    @classmethod
    def from_record(cls, record):
        return cls(record["hostname"].strip().lower())

device = Device.from_record({"hostname": "  " + case["hostname"].upper() + "  "})
assert device.hostname == case["hostname"]
print(device.hostname)
````

## Pattern 008: Static validation utility

**Core concept:** A static method has no implicit self or cls.

**Network-engineering purpose:** A VLAN predicate can be used before an interface object is created.

## Pattern 009: Avoid shared mutable default arguments

**Core concept:** Allocate owned mutable containers per instance.

**Network-engineering purpose:** Adding an interface to one device must not silently add it to another.

### Read the implementation in this order
1. Accept None instead of a mutable default.
2. Allocate a new list or copy the incoming one.
3. Mutate the first device only.
4. Assert both the second object and caller's input stay unchanged.

**Common mistake / boundary:** Using interfaces=[] in the signature reuses one list; list(...) is only a shallow copy.

**Practice after the ten worked scenarios:** Replace the strings with nested dictionaries and explain why a shallow copy is insufficient.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0081 - Avoid shared mutable default arguments: branch scenario

**Scenario 01/10; pattern 009; Beginner.** This is a self-contained offline exercise. Adding an interface to one device must not silently add it to another.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Accept None instead of a mutable default. Assert both the second object and caller's input stay unchanged. Follow the class-specific steps under Pattern 009 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['Ethernet1/1', 'Loopback0'] []
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Using interfaces=[] in the signature reuses one list; list(...) is only a shallow copy.

````python
# Example 0081: Avoid shared mutable default arguments - branch
case = {'interface': 'Ethernet1/1'}

class Device:
    def __init__(self, interfaces=None):
        self.interfaces = [] if interfaces is None else list(interfaces)

source = [case["interface"]]
first, second = Device(source), Device()
first.interfaces.append("Loopback0")
assert second.interfaces == []
assert source == [case["interface"]]
print(first.interfaces, second.interfaces)
````

# Chapter 03: Encapsulation and invariants

**Level:** Beginner to intermediate

**Learning goal:** Keep invalid state out, expose properties deliberately, and understand Python's privacy conventions.

**Encapsulation is a boundary, not secrecy.** Clients should use a documented public interface
rather than depend on internal storage. `_name` communicates an internal implementation detail.
`__name` triggers name mangling, mainly to reduce accidental collisions in subclasses. Neither
convention protects secrets against someone with access to the Python process.

**Validation belongs at the boundary.** A constructor should establish valid state, and every
mutating operation should preserve it. Validate before assignment so a failing update leaves
the previous value intact. Decide whether your API rejects or converts input; do not casually
mix the two. In these strict integer examples `type(value) is int` intentionally rejects
booleans. This strictness is a local teaching policy, not a universal Python API requirement.

**Properties** let callers use attribute syntax while a getter computes or a setter validates.
Inside a setter, assign `_mtu`, not `mtu`, or the setter will call itself recursively.
A getter-only property has no supported assignment path, although the backing attribute is
still accessible by convention. Expose a tuple or defensive copy when clients should not be
able to modify the model's internal collection accidentally.

**Useful exception categories.** Raise `TypeError` for an unsupported input type and `ValueError`
for an unacceptable value of an otherwise supported type. Catch only the errors you intend to
handle. An expected rejection is shown explicitly in the examples and its message is checked.

**Model policy versus platform policy.** A teaching MTU interval or VLAN interval is not a
vendor specification. Actual support depends on interface type, encapsulation, platform,
firmware, and operational requirements. Keep such policy visible and configurable when a
real system needs it. Never infer device readiness from a successful constructor alone.

**Pause and explain:** Explain why validation must occur before mutation and why underscores are not security controls.

## Pattern 011: Validate a constructor before storing state

**Core concept:** A constructor should establish an invariant or raise an explicit error.

**Network-engineering purpose:** An invalid retry budget must not enter a collection of otherwise usable policies.

### Read the implementation in this order
1. Check the type first.
2. Check the allowed range.
3. Store only a valid budget.
4. Catch ValueError for a known bad input and prove that rejection occurred.

**Common mistake / boundary:** Assertions are used for exercises, not for production input validation; Python -O can remove them.

**Practice after the ten worked scenarios:** Add a separate TypeError test for a string retry count.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0101 - Validate a constructor before storing state: branch scenario

**Scenario 01/10; pattern 011; Beginner to intermediate.** This is a self-contained offline exercise. An invalid retry budget must not enter a collection of otherwise usable policies.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `attempts` | `1` |

**Logic to trace:** Check the type first. Catch ValueError for a known bad input and prove that rejection occurred. Follow the class-specific steps under Pattern 011 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
attempts: 1
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Assertions are used for exercises, not for production input validation; Python -O can remove them.

````python
# Example 0101: Validate a constructor before storing state - branch
case = {'attempts': 1}

class RetryBudget:
    def __init__(self, attempts):
        if type(attempts) is not int:
            raise TypeError("attempts must be an integer")
        if attempts < 1:
            raise ValueError("attempts must be positive")
        self.attempts = attempts

budget = RetryBudget(case["attempts"])
try:
    RetryBudget(0)
except ValueError as error:
    assert str(error) == "attempts must be positive"
else:
    raise AssertionError("invalid budget accepted")
print("attempts:", budget.attempts)
````

## Pattern 012: Validated property setter

**Core concept:** A property preserves an invariant on both construction and later updates.

**Network-engineering purpose:** A model can reject an unsupported MTU before changing its desired configuration.

### Read the implementation in this order
1. Delegate initialization to the setter.
2. Read through the property getter.
3. Reject values outside this lab's 576..9216 policy before assignment.
4. Attempt a bad update and prove the previous value survived.

**Common mistake / boundary:** Assigning self.mtu within the mtu setter causes recursion; use a backing field.

**Practice after the ten worked scenarios:** Check the type explicitly and add a test rejecting True.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0111 - Validated property setter: branch scenario

**Scenario 01/10; pattern 012; Beginner to intermediate.** This is a self-contained offline exercise. A model can reject an unsupported MTU before changing its desired configuration.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Delegate initialization to the setter. Attempt a bad update and prove the previous value survived. Follow the class-specific steps under Pattern 012 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
1500
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Assigning self.mtu within the mtu setter causes recursion; use a backing field.

````python
# Example 0111: Validated property setter - branch
case = {'mtu': 1500}

class Interface:
    def __init__(self, mtu):
        self.mtu = mtu

    @property
    def mtu(self):
        return self._mtu

    @mtu.setter
    def mtu(self, value):
        if type(value) is not int or not 576 <= value <= 9216:
            raise ValueError("MTU outside lab policy")
        self._mtu = value

interface = Interface(case["mtu"])
try:
    interface.mtu = 0
except ValueError:
    assert interface.mtu == case["mtu"]
else:
    raise AssertionError("invalid MTU accepted")
print(interface.mtu)
````

````python
# Example 0120: Validated property setter - backup
case = {'mtu': 8192}

class Interface:
    def __init__(self, mtu):
        self.mtu = mtu

    @property
    def mtu(self):
        return self._mtu

    @mtu.setter
    def mtu(self, value):
        if type(value) is not int or not 576 <= value <= 9216:
            raise ValueError("MTU outside lab policy")
        self._mtu = value

interface = Interface(case["mtu"])
try:
    interface.mtu = 0
except ValueError:
    assert interface.mtu == case["mtu"]
else:
    raise AssertionError("invalid MTU accepted")
print(interface.mtu)
````

## Pattern 015: Name mangling avoids a subclass collision

**Core concept:** Double-leading underscores alter attribute names, not access permissions.

**Network-engineering purpose:** A base adapter and a child adapter can keep separate internal labels.

### Read the implementation in this order
1. Give the base and child classes their own __label.
2. Initialize the base with super().
3. Read each label through its defining class's method.
4. Check that both values coexist.

**Common mistake / boundary:** Never treat a mangled attribute as encrypted or access-controlled storage.

**Practice after the ten worked scenarios:** Inspect vars(adapter) and explain the generated attribute names without using them in application code.

# Chapter 04: Composition, aggregation, and dependencies

**Level:** Intermediate

**Learning goal:** Model has-a relationships and inject offline collaborators instead of hard-coding I/O.

**Composition** builds a larger behavior from smaller objects: a Device has Interfaces, and a
Collector has a Transport. In ownership terminology, composition often implies that the owner
controls a component's lifecycle; aggregation often describes a reference to an independently
existing component. Python references do not enforce those lifecycle distinctions for you.

**Prefer clear responsibility boundaries.** An interface can validate its own desired MTU.
A transport can send or simulate commands. A service coordinates them. An inventory repository
stores records. A policy decides whether the observed state is acceptable. Combining all of
this in one giant Device class makes testing, reuse, and error reporting harder.

**Dependency injection** means providing a collaborator from outside, often through __init__.
It is not a special library. Passing a fake transport makes tests deterministic and keeps these
notebooks offline. A real adapter would be responsible for authentication, timeouts, host-key
verification, cleanup, and platform-specific behavior. The service should not create a hidden
global SSH connection just because someone instantiates it.

**Ownership matters with mutable inputs.** If the owner promises a snapshot, copy the container.
If it intentionally shares a live collection, document that shared state and its synchronization
rules. Defining ownership is more useful than mechanically copying everything.

**Design question.** Use inheritance for a genuine substitutable is-a relationship. Use
composition for has-a, uses-a, and policies that may vary independently. A router with a
transport and a policy is usually easier to evolve than separate subclasses for every
combination of router platform, transport, region, and validation policy.

**Pause and explain:** Describe which object owns each dependency and how you would replace transport in a unit test.

## Pattern 016: A device owns interface objects

**Core concept:** Composition groups related objects behind one useful query.

**Network-engineering purpose:** A device-level report can summarize interface state without duplicating interface logic.

### Read the implementation in this order
1. Create Interface objects with names and admin states.
2. Give Device its own list container.
3. Filter interfaces using their public fields.
4. Check the resulting list of enabled names.

**Common mistake / boundary:** Copying the container does not clone its Interface objects.

**Practice after the ten worked scenarios:** Add a lookup that raises KeyError for an unknown interface name.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0151 - A device owns interface objects: branch scenario

**Scenario 01/10; pattern 016; Intermediate.** This is a self-contained offline exercise. A device-level report can summarize interface state without duplicating interface logic.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Create Interface objects with names and admin states. Check the resulting list of enabled names. Follow the class-specific steps under Pattern 016 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['Ethernet1/1']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Copying the container does not clone its Interface objects.

````python
# Example 0151: A device owns interface objects - branch
case = {'interface': 'Ethernet1/1'}

class Interface:
    def __init__(self, name, admin_up):
        self.name, self.admin_up = name, admin_up

class Device:
    def __init__(self, interfaces):
        self.interfaces = list(interfaces)

    def enabled_names(self):
        return [item.name for item in self.interfaces if item.admin_up]

device = Device([Interface(case["interface"], True), Interface("Loopback9", False)])
assert device.enabled_names() == [case["interface"]]
print(device.enabled_names())
````

## Pattern 017: Inject a fake transport

**Core concept:** A service delegates command execution to an injected collaborator.

**Network-engineering purpose:** Collection logic can be exercised without credentials, sockets, or a real device.

### Read the implementation in this order
1. Make a fake whose send method records the request.
2. Inject it through the service constructor.
3. Ask the service to collect a status.
4. Assert both returned data and the exact recorded command.

**Common mistake / boundary:** A fake proves local behavior only; it cannot establish compatibility with a real network OS.

**Practice after the ten worked scenarios:** Make the fake raise TimeoutError and design a caller that reports it explicitly.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0161 - Inject a fake transport: branch scenario

**Scenario 01/10; pattern 017; Intermediate.** This is a self-contained offline exercise. Collection logic can be exercised without credentials, sockets, or a real device.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Make a fake whose send method records the request. Assert both returned data and the exact recorded command. Follow the class-specific steps under Pattern 017 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
up ['show interface Ethernet1/1']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A fake proves local behavior only; it cannot establish compatibility with a real network OS.

````python
# Example 0161: Inject a fake transport - branch
case = {'interface': 'Ethernet1/1'}

class FakeTransport:
    def __init__(self):
        self.calls = []

    def send(self, command):
        self.calls.append(command)
        return "up"

class Collector:
    def __init__(self, transport):
        self.transport = transport

    def status(self, interface):
        return self.transport.send(f"show interface {interface}")

transport = FakeTransport()
result = Collector(transport).status(case["interface"])
assert result == "up"
assert transport.calls == [f"show interface {case['interface']}"]
print(result, transport.calls)
````

## Pattern 018: Aggregate independently owned devices

**Core concept:** Aggregation keeps references to objects that can outlive the group.

**Network-engineering purpose:** An inventory view can group a device by site without duplicating that device.

### Read the implementation in this order
1. Create a device independently.
2. Put its reference into a Site.
3. Update the device through its original reference.
4. Observe the same object through the site's collection.

**Common mistake / boundary:** A shared reference means mutations are visible to every owner of that reference.

**Practice after the ten worked scenarios:** Build a snapshot view instead and contrast its semantics with the live view.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0171 - Aggregate independently owned devices: branch scenario

**Scenario 01/10; pattern 018; Intermediate.** This is a self-contained offline exercise. An inventory view can group a device by site without duplicating that device.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Create a device independently. Observe the same object through the site's collection. Follow the class-specific steps under Pattern 018 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01-maint
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A shared reference means mutations are visible to every owner of that reference.

````python
# Example 0171: Aggregate independently owned devices - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

class Site:
    def __init__(self, devices):
        self.devices = list(devices)

device = Device(case["hostname"])
site = Site([device])
device.hostname += "-maint"
assert site.devices[0] is device
print(site.devices[0].hostname)
````

## Pattern 019: Inject a callable policy

**Core concept:** A dependency can be a function rather than a class hierarchy.

**Network-engineering purpose:** Different environments can choose different capacity thresholds with the same evaluator.

### Read the implementation in this order
1. Accept a predicate in the evaluator constructor.
2. Pass a computed value into that predicate.
3. Use a lambda that captures this scenario's limit.
4. Compare evaluation with the direct condition.

**Common mistake / boundary:** A lambda is appropriate for one short expression, not a place to hide a large workflow.

**Practice after the ten worked scenarios:** Replace the lambda with a named function that documents the threshold policy.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0181 - Inject a callable policy: branch scenario

**Scenario 01/10; pattern 019; Intermediate.** This is a self-contained offline exercise. Different environments can choose different capacity thresholds with the same evaluator.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `load` | `20` |
| `threshold` | `80` |

**Logic to trace:** Accept a predicate in the evaluator constructor. Compare evaluation with the direct condition. Follow the class-specific steps under Pattern 019 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
acceptable: True
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A lambda is appropriate for one short expression, not a place to hide a large workflow.

````python
# Example 0181: Inject a callable policy - branch
case = {'load': 20, 'threshold': 80}

class Evaluator:
    def __init__(self, policy):
        self.policy = policy

    def acceptable(self, value):
        return self.policy(value)

evaluator = Evaluator(lambda value: value <= case["threshold"])
result = evaluator.acceptable(case["load"])
assert result == (case["load"] <= case["threshold"])
print("acceptable:", result)
````

## Pattern 020: Repository separates storage from lookup

**Core concept:** A repository exposes a domain-oriented access operation.

**Network-engineering purpose:** Consumers can request a hostname without depending on the backing dictionary's organization.

### Read the implementation in this order
1. Copy the incoming record mapping.
2. Expose a get method with explicit missing-key behavior.
3. Retrieve a known record.
4. Try an unknown hostname and verify KeyError is not swallowed.

**Common mistake / boundary:** Returning a mutable stored record can expose internal state; this example stores immutable strings.

**Practice after the ten worked scenarios:** Add list_by_site without exposing the internal dictionary.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0191 - Repository separates storage from lookup: branch scenario

**Scenario 01/10; pattern 020; Intermediate.** This is a self-contained offline exercise. Consumers can request a hostname without depending on the backing dictionary's organization.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |
| `platform` | `'offline-compact'` |

**Logic to trace:** Copy the incoming record mapping. Try an unknown hostname and verify KeyError is not swallowed. Follow the class-specific steps under Pattern 020 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
missing record reported
offline-compact
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Returning a mutable stored record can expose internal state; this example stores immutable strings.

````python
# Example 0191: Repository separates storage from lookup - branch
case = {'hostname': 'branch-edge-01', 'platform': 'offline-compact'}

class InventoryRepository:
    def __init__(self, records):
        self._records = dict(records)

    def get(self, hostname):
        return self._records[hostname]

repository = InventoryRepository({case["hostname"]: case["platform"]})
assert repository.get(case["hostname"]) == case["platform"]
try:
    repository.get("missing-lab-device")
except KeyError:
    print("missing record reported")
else:
    raise AssertionError("missing record was hidden")
print(repository.get(case["hostname"]))
````

# Chapter 05: Inheritance, overriding, super, and MRO

**Level:** Intermediate

**Learning goal:** Use substitutable subclasses and trace cooperative multiple inheritance.

**Inheritance** creates a derived type from an existing base type. Single inheritance has one
direct base. Multilevel inheritance extends a derived class again. Hierarchical inheritance
has several children of one base. Multiple inheritance has more than one direct base; hybrid
hierarchies combine those shapes. The shape is less important than a coherent public contract.

**Overriding** replaces a base method with a method of the same name in a subclass. Python
dispatches at runtime using the actual object. A list of different device adapters can therefore
answer the same method call differently. This is subtype polymorphism. Duck typing can achieve
interchangeable behavior without an inherited base.

**super() is not simply 'call my parent'.** It creates a proxy that continues lookup after the
current class in the method resolution order (MRO) for the actual instance. In cooperative
multiple inheritance, every participating method calls super consistently and accepts
compatible arguments. Inspect `SomeClass.__mro__` when reasoning about a diamond.

**Mixins** provide a small reusable capability, not a second unrelated domain identity.
Keep their dependencies explicit and avoid conflicting constructor conventions. Prefer
composition if capabilities need complex ownership or independent configuration.

**Substitutability.** A subclass should honor the promises of its base. A read method that
suddenly deletes configuration, or a child that rejects inputs the base promised to support,
violates that expectation even if the code runs. Do not use inheritance merely to reuse a
few lines. Python does not overload methods by parameter signature: writing a second method
with the same name replaces the first definition. Use defaults, keyword-only parameters,
separate names, or explicit dispatch instead.

**Pause and explain:** Trace a diamond MRO; explain why runtime overriding is not signature-based overloading.

## Pattern 021: Specialize a base device

**Core concept:** A derived class inherits existing methods and adds a focused capability.

**Network-engineering purpose:** A router can reuse a common identity operation while adding a routing-related label.

### Read the implementation in this order
1. Define the shared constructor and identity method on Device.
2. Inherit Router from Device.
3. Add a capability without duplicating identity.
4. Check both inherited and specialized methods.

**Common mistake / boundary:** Adding a subclass is useful only if the distinction changes meaningful behavior.

**Practice after the ten worked scenarios:** Create a Switch subclass and process both through the shared identity() contract.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0201 - Specialize a base device: branch scenario

**Scenario 01/10; pattern 021; Intermediate.** This is a self-contained offline exercise. A router can reuse a common identity operation while adding a routing-related label.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Define the shared constructor and identity method on Device. Check both inherited and specialized methods. Follow the class-specific steps under Pattern 021 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01 routing
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Adding a subclass is useful only if the distinction changes meaningful behavior.

````python
# Example 0201: Specialize a base device - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    def identity(self):
        return self.hostname

class Router(Device):
    def role(self):
        return "routing"

router = Router(case["hostname"])
assert router.identity() == case["hostname"]
assert isinstance(router, Device)
print(router.identity(), router.role())
````

## Pattern 025: Cooperative diamond inheritance

**Core concept:** super follows the instance MRO and avoids manually calling both parents.

**Network-engineering purpose:** A pipeline can combine two tiny behaviors while visiting a shared base once.

### Read the implementation in this order
1. Define a terminal implementation on Root.
2. Have each branch prepend its label then call super.
3. Inherit Combined from Left and Right in that order.
4. Check the exact visit order and inspect __mro__.

**Common mistake / boundary:** Calling Root directly in both branches breaks cooperative traversal and may duplicate work.

**Practice after the ten worked scenarios:** Reverse the two bases and predict the new visit order before running it.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0241 - Cooperative diamond inheritance: branch scenario

**Scenario 01/10; pattern 025; Intermediate.** This is a self-contained offline exercise. A pipeline can combine two tiny behaviors while visiting a shared base once.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Define a terminal implementation on Root. Check the exact visit order and inspect __mro__. Follow the class-specific steps under Pattern 025 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01 ['left', 'right', 'root']
['Combined', 'Left', 'Right', 'Root', 'object']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Calling Root directly in both branches breaks cooperative traversal and may duplicate work.

````python
# Example 0241: Cooperative diamond inheritance - branch
case = {'hostname': 'branch-edge-01'}

class Root:
    def trace(self):
        return ["root"]

class Left(Root):
    def trace(self):
        return ["left"] + super().trace()

class Right(Root):
    def trace(self):
        return ["right"] + super().trace()

class Combined(Left, Right):
    pass

assert Combined().trace() == ["left", "right", "root"]
print(case["hostname"], Combined().trace())
print([cls.__name__ for cls in Combined.__mro__])
````

# Chapter 06: Abstraction, duck typing, protocols, and generics

**Level:** Intermediate

**Learning goal:** Express contracts without confusing type hints with runtime validation.

**Abstraction** presents the operations a caller needs while hiding irrelevant implementation.
A collector should need `send(command) -> str`, not every detail of an SSH or HTTP client.
Abstract base classes (ABCs) make a nominal contract explicit. A class with unimplemented
abstract methods cannot normally be instantiated. Concrete subclasses still need sensible
semantics: the ABC machinery cannot prove their output is correct.

**Duck typing** uses behavior directly: if an object supports the required operation, it can
be used. A `typing.Protocol` describes that shape to a static type checker without requiring
explicit inheritance. `@runtime_checkable` permits limited runtime presence checks, but does
not validate signatures, return values, or semantic correctness. The defining protocol in
these examples is explicitly decorated rather than relying on inherited runtime checkability.

**Type hints** document intent and help editors and static analyzers. Python does not generally
enforce them during an ordinary call. Validate untrusted input separately. `list[Device]`
describes a collection of devices; `Device | None` expresses an optional value. A type variable
can preserve the relationship between a container's input and output type. This course uses
the traditional `TypeVar` and `Generic` form for readability across supported Python versions.

**Dependency inversion** puts the service's required abstraction ahead of a particular adapter.
Inject a fake during a test and a reviewed real implementation at the application boundary.
Do not add an ABC, Protocol, and registry for every two-line helper: choose the smallest
contract that actually helps callers and tests.

## Pattern 026: Enforce an abstract operation

**Core concept:** ABC and abstractmethod prevent incomplete concrete construction.

**Network-engineering purpose:** A collector adapter must implement the operation the service depends on.

### Read the implementation in this order
1. Declare an abstract read method.
2. Implement it in an offline subclass.
3. Try instantiating the incomplete base.
4. Check the concrete implementation's result.

**Common mistake / boundary:** An abstract method only requires implementation; it does not validate that implementation's semantics.

**Practice after the ten worked scenarios:** Add a second abstract operation and observe the effect on the existing subclass.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0251 - Enforce an abstract operation: branch scenario

**Scenario 01/10; pattern 026; Intermediate.** This is a self-contained offline exercise. A collector adapter must implement the operation the service depends on.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Declare an abstract read method. Check the concrete implementation's result. Follow the class-specific steps under Pattern 026 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
abstract reader rejected
branch-edge-01
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** An abstract method only requires implementation; it does not validate that implementation's semantics.

````python
# Example 0251: Enforce an abstract operation - branch
case = {'hostname': 'branch-edge-01'}

from abc import ABC, abstractmethod

class Reader(ABC):
    @abstractmethod
    def read(self):
        raise NotImplementedError

class OfflineReader(Reader):
    def read(self):
        return case["hostname"]

try:
    Reader()
except TypeError:
    print("abstract reader rejected")
else:
    raise AssertionError("abstract class instantiated")
assert OfflineReader().read() == case["hostname"]
print(OfflineReader().read())
````

## Pattern 027: Duck-typed collaborators

**Core concept:** Compatible behavior can be used without a common explicit base.

**Network-engineering purpose:** A fake probe and a replayed probe can both feed a health service.

### Read the implementation in this order
1. Give unrelated classes the same check method.
2. Write a client that calls that method.
3. Pass both classes through the client.
4. Assert their distinct results are preserved.

**Common mistake / boundary:** Missing behavior will fail at runtime; annotations and tests can catch mistakes earlier.

**Practice after the ten worked scenarios:** Create a bad collaborator and identify the exact failure boundary.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0261 - Duck-typed collaborators: branch scenario

**Scenario 01/10; pattern 027; Intermediate.** This is a self-contained offline exercise. A fake probe and a replayed probe can both feed a health service.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `healthy` | `True` |

**Logic to trace:** Give unrelated classes the same check method. Assert their distinct results are preserved. Follow the class-specific steps under Pattern 027 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
[True, True]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Missing behavior will fail at runtime; annotations and tests can catch mistakes earlier.

````python
# Example 0261: Duck-typed collaborators - branch
case = {'healthy': True}

class GoodProbe:
    def check(self):
        return True

class ReplayProbe:
    def check(self):
        return case["healthy"]

def health(probe):
    return probe.check()

result = [health(GoodProbe()), health(ReplayProbe())]
assert result == [True, case["healthy"]]
print(result)
````

## Pattern 028: Structural contract with Protocol

**Core concept:** A Protocol describes the operations a collaborator must expose.

**Network-engineering purpose:** A service can type against a transport shape rather than a specific library.

### Read the implementation in this order
1. Declare the send signature on a Protocol.
2. Decorate that Protocol for a limited runtime check.
3. Implement send without inheriting from the protocol.
4. Use the object through an annotated function.

**Common mistake / boundary:** isinstance against a runtime protocol checks presence, not argument types or result correctness.

**Practice after the ten worked scenarios:** Run a static type checker on a variant whose send method returns an integer.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0271 - Structural contract with Protocol: branch scenario

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Declare the send signature on a Protocol. Use the object through an annotated function. Follow the class-specific steps under Pattern 028 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
replay:Ethernet1/1
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. **Remember:** isinstance against a runtime protocol checks presence, not argument types or result correctness.

````python
# Example 0271: Structural contract with Protocol - branch
case = {'interface': 'Ethernet1/1'}

from typing import Protocol, runtime_checkable

@runtime_checkable
class Transport(Protocol):
    def send(self, command: str) -> str: ...

class ReplayTransport:
    def send(self, command: str) -> str:
        return "replay:" + command

def collect(transport: Transport, command: str) -> str:
    return transport.send(command)

transport = ReplayTransport()
assert isinstance(transport, Transport)
result = collect(transport, case["interface"])
assert result == "replay:" + case["interface"]
print(result)
````

## Pattern 029: Generic result container

**Core concept:** A type variable links a wrapper's stored value to its return type.

**Network-engineering purpose:** An inventory result wrapper can preserve the type of its payload for static analysis.

### Read the implementation in this order
1. Define a TypeVar.
2. Parameterize Box with Generic.
3. Return the stored T from unwrap.
4. Construct a string box and check that its payload is unchanged.

**Common mistake / boundary:** Generic annotations are not automatic runtime type checks.

**Practice after the ten worked scenarios:** Create a Box[int] and explain the inferred return type of unwrap().

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0281 - Generic result container: branch scenario

**Scenario 01/10; pattern 029; Intermediate.** This is a self-contained offline exercise. An inventory result wrapper can preserve the type of its payload for static analysis.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Define a TypeVar. Construct a string box and check that its payload is unchanged. Follow the class-specific steps under Pattern 029 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Generic annotations are not automatic runtime type checks.

````python
# Example 0281: Generic result container - branch
case = {'hostname': 'branch-edge-01'}

from typing import Generic, TypeVar

T = TypeVar("T")

class Box(Generic[T]):
    def __init__(self, value: T):
        self.value = value

    def unwrap(self) -> T:
        return self.value

result = Box[str](case["hostname"])
assert result.unwrap() == case["hostname"]
print(result.unwrap())
````

## Pattern 030: Separate a service from policy implementation

**Core concept:** A service depends on a small abstract contract.

**Network-engineering purpose:** Compliance rules can change without rewriting the report orchestration.

### Read the implementation in this order
1. Declare the accepts operation on Policy.
2. Implement a scenario-specific upper bound.
3. Inject the policy into the service.
4. Compare service output with the direct policy condition.

**Common mistake / boundary:** Do not place transport or formatting responsibilities inside the policy.

**Practice after the ten worked scenarios:** Add a range policy with a lower and upper bound using the same service.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0291 - Separate a service from policy implementation: branch scenario

**Scenario 01/10; pattern 030; Intermediate.** This is a self-contained offline exercise. Compliance rules can change without rewriting the report orchestration.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `load` | `20` |
| `threshold` | `80` |

**Logic to trace:** Declare the accepts operation on Policy. Compare service output with the direct policy condition. Follow the class-specific steps under Pattern 030 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
pass
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not place transport or formatting responsibilities inside the policy.

````python
# Example 0291: Separate a service from policy implementation - branch
case = {'load': 20, 'threshold': 80}

from abc import ABC, abstractmethod

class Policy(ABC):
    @abstractmethod
    def accepts(self, value):
        raise NotImplementedError

class MaximumPolicy(Policy):
    def __init__(self, maximum):
        self.maximum = maximum

    def accepts(self, value):
        return value <= self.maximum

class ComplianceService:
    def __init__(self, policy):
        self.policy = policy

    def check(self, value):
        return "pass" if self.policy.accepts(value) else "fail"

result = ComplianceService(MaximumPolicy(case["threshold"])).check(case["load"])
assert result == ("pass" if case["load"] <= case["threshold"] else "fail")
print(result)
````

# Chapter 07: Dataclasses, value objects, and copying

**Level:** Intermediate

**Learning goal:** Choose value semantics, safe factories, immutable fields, and deliberate copy depth.

**Dataclasses** generate common methods from annotated fields. A normal dataclass usually gets
initialization, representation, and field-based equality. This reduces boilerplate for records,
but it does not validate types, parse addresses, or make every domain model correct. Use
`__post_init__` for additional validation after generated initialization.

**Identity versus value.** Two separately created route records may compare equal because their
fields match while `first is second` remains false. Choose equality based on domain meaning.
A live session is often an identity-bearing resource; a route key is often a value.

**Mutable fields** should use `field(default_factory=list)` or the appropriate factory, which
creates a fresh container per instance. `frozen=True` blocks normal field reassignment but is
not deep immutability: a list stored in a frozen object can still change. Prefer immutable
nested values when you need a dependable dictionary key.

**Ordering** is opt-in and follows the dataclass field order for fields included in comparison.
Put the intended sort key first or define an explicit key function. Avoid accidental ordering
by a hostname when your intent was route preference.

**Copy choices.** Assignment shares the same object. A shallow copy creates a new outer object
while nested references remain shared. A deep copy recursively copies suitable nested values,
subject to customization and memoization. Neither approach makes a meaningful duplicate of a
live network session, lock, or external resource. For a frozen configuration record,
`dataclasses.replace` expresses a changed value clearly without mutating the original.

**Pause and explain:** Explain frozen versus deep immutability and predict shallow-copy behavior for a nested list.

## Pattern 031: Dataclass value equality

**Core concept:** Generated equality compares fields rather than requiring identical objects.

**Network-engineering purpose:** Two independently parsed route keys can represent the same logical destination.

### Read the implementation in this order
1. Declare annotated fields on a frozen dataclass.
2. Construct two equal records independently.
3. Compare them with == and check their hashes.
4. Check that identity remains different and a set deduplicates equal values.

**Common mistake / boundary:** Field equality may be wrong for resources with identity or lifecycle; choose semantics deliberately.

**Practice after the ten worked scenarios:** Add a non-key field with compare=False and reason about its effect on equality.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

| Input | Value |

| `prefix` | `'192.0.2.0/28'` |

**Logic to trace:** Declare annotated fields on a frozen dataclass. Check that identity remains different and a set deduplicates equal values. Follow the class-specific steps under Pattern 031 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**

````python
# Example 0301: Dataclass value equality - branch
case = {'site': 'branch', 'prefix': '192.0.2.0/28'}

from dataclasses import dataclass

@dataclass(frozen=True)
class RouteKey:
    prefix: str
    vrf: str

first = RouteKey(case["prefix"], case["site"])
second = RouteKey(case["prefix"], case["site"])
assert first == second
assert first is not second
assert hash(first) == hash(second)
assert len({first, second}) == 1
print(first)
````

## Pattern 032: Fresh dataclass collection fields

**Core concept:** default_factory builds a separate mutable value for each instance.

**Network-engineering purpose:** Each device's change log must remain independent.

### Read the implementation in this order
1. Use field(default_factory=list).
2. Create two records without supplying the log.
3. Append a change to the first.
4. Assert the second remains empty and the lists have different identities.

**Common mistake / boundary:** A factory is called for each needed default; passing one existing list explicitly can still share it.

**Practice after the ten worked scenarios:** Give the class a method that appends a timestamp-free structured event.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0311 - Fresh dataclass collection fields: branch scenario

**Scenario 01/10; pattern 032; Intermediate.** This is a self-contained offline exercise. Each device's change log must remain independent.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Use field(default_factory=list). Assert the second remains empty and the lists have different identities. Follow the class-specific steps under Pattern 032 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['reviewed'] []
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A factory is called for each needed default; passing one existing list explicitly can still share it.

````python
# Example 0311: Fresh dataclass collection fields - branch
case = {'hostname': 'branch-edge-01'}

from dataclasses import dataclass, field

@dataclass
class Device:
    hostname: str
    changes: list[str] = field(default_factory=list)

first = Device(case["hostname"])
second = Device("spare")
first.changes.append("reviewed")
assert second.changes == []
assert first.changes is not second.changes
print(first.changes, second.changes)
````

## Pattern 033: Frozen configuration and replace

**Core concept:** Immutable field assignment encourages explicit new desired-state values.

**Network-engineering purpose:** A proposed MTU can be represented without overwriting the approved baseline.

### Read the implementation in this order
1. Create a frozen dataclass.
2. Use replace to derive a changed record.
3. Check both old and new values.
4. Attempt direct mutation and verify the specific frozen-instance error.

**Common mistake / boundary:** Frozen is shallow; do not put mutable children in a value object you intend to hash.

**Practice after the ten worked scenarios:** Add a tuple of allowed VLANs and construct an updated value with replace().

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

**Scenario 01/10; pattern 033; Intermediate.** This is a self-contained offline exercise. A proposed MTU can be represented without overwriting the approved baseline.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Create a frozen dataclass. Attempt direct mutation and verify the specific frozen-instance error. Follow the class-specific steps under Pattern 033 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
baseline protected

```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. **Remember:** Frozen is shallow; do not put mutable children in a value object you intend to hash.

````python
# Example 0321: Frozen configuration and replace - branch
case = {'mtu': 1500}

from dataclasses import FrozenInstanceError, dataclass, replace

@dataclass(frozen=True)
class Config:
    mtu: int

baseline = Config(1500)
proposed = replace(baseline, mtu=case["mtu"])
assert baseline.mtu == 1500 and proposed.mtu == case["mtu"]
try:
    baseline.mtu = 9000
except FrozenInstanceError:
    print("baseline protected")
else:
    raise AssertionError("frozen field changed")
print(proposed)
````

## Pattern 034: Explicit dataclass ordering

## Pattern 035: Shallow versus deep copy

**Core concept:** Copy depth determines which nested objects remain shared.

**Network-engineering purpose:** A candidate inventory snapshot should not accidentally mutate a baseline interface list.

### Read the implementation in this order
1. Build a dataclass with a mutable child list.
2. Make a shallow and a deep copy.
3. Mutate the shallow copy's child.
4. Check the baseline changed but the deep snapshot did not.

**Common mistake / boundary:** Deep-copy configuration data, not live connection handles.

**Practice after the ten worked scenarios:** Replace the list with a tuple and compare the need for defensive copying.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

# Chapter 08: Python's object protocols and special methods

**Level:** Intermediate

**Learning goal:** Make domain objects work naturally with repr, bool, len, iteration, lookup, and with.

**Special methods** connect objects to Python syntax. `repr(obj)` uses `__repr__`; `str(obj)`
uses `__str__` when supplied. A useful repr is unambiguous and helps debugging, but should not
disclose credentials or tokens. A string representation is not a serialization contract.

**Truth and size.** `bool(obj)` uses `__bool__`, or falls back to `__len__` when appropriate.
Return an actual boolean from __bool__ and a nonnegative integer from __len__. A domain
choice such as 'healthy results are truthy' must be documented; otherwise write an explicit
`result.healthy` query so readers do not have to guess.

**Iteration** normally works by __iter__ returning an iterator. A collection should usually
create a fresh iterator each time, rather than store one exhausted iterator on the object.
**Subscription** uses __getitem__. A mapping-like API should preserve KeyError for missing
keys rather than silently return a plausible empty record.

**Context managers** model resource lifetime. `with` calls __enter__, runs the block, and calls
__exit__ even when the body raises. __exit__ receives exception information. Returning false
(or None) allows the exception to propagate; true suppresses it. Do not suppress an operational
failure merely to make a notebook cell look successful.

**Equality and hashing.** If objects compare equal, their hashes must agree. A value used as a
set member or dictionary key must not change in a way that changes equality/hash semantics.
For unsupported operand types, an equality method should normally return NotImplemented.
Not every possible operator belongs on a network object: prefer a clear named method when
overloaded syntax would surprise an engineer reading the automation.

## Pattern 036: Readable representations without sensitive fields

**Core concept:** str is for friendly display; repr is for useful diagnostics.

**Network-engineering purpose:** A log should identify the device without exposing unrelated session internals.

### Read the implementation in this order
1. Store only the identity needed for this model.
2. Provide __str__ and __repr__.
3. Call the built-ins instead of the special methods directly.
4. Assert both exact formats.

**Common mistake / boundary:** Do not include passwords, tokens, or full credential objects in repr.

**Practice after the ten worked scenarios:** Add a site field and decide whether it belongs in each representation.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0351 - Readable representations without sensitive fields: branch scenario

**Scenario 01/10; pattern 036; Intermediate.** This is a self-contained offline exercise. A log should identify the device without exposing unrelated session internals.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Store only the identity needed for this model. Assert both exact formats. Follow the class-specific steps under Pattern 036 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01 Device(hostname='branch-edge-01')
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not include passwords, tokens, or full credential objects in repr.

````python
# Example 0351: Readable representations without sensitive fields - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    def __str__(self):
        return self.hostname

    def __repr__(self):
        return f"Device(hostname={self.hostname!r})"

device = Device(case["hostname"])
assert str(device) == case["hostname"]
assert repr(device) == f"Device(hostname={case['hostname']!r})"
print(str(device), repr(device))
````

## Pattern 037: Truthiness and collection length

**Core concept:** A domain object can participate in bool and len with a clear contract.

**Network-engineering purpose:** A lab report is truthy exactly when it has no failed checks, while len counts all checks.

### Read the implementation in this order
1. Store the check outcomes.
2. Implement len as the number of checks.
3. Implement bool as all checks passing.
4. Check size and truth independently.

**Common mistake / boundary:** all([]) is True; choose and document whether an empty report should count as healthy.

**Practice after the ten worked scenarios:** Change the contract so an empty report is not healthy, then add the empty-case assertion.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0361 - Truthiness and collection length: branch scenario

**Scenario 01/10; pattern 037; Intermediate.** This is a self-contained offline exercise. A lab report is truthy exactly when it has no failed checks, while len counts all checks.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `healthy` | `True` |

**Logic to trace:** Store the check outcomes. Check size and truth independently. Follow the class-specific steps under Pattern 037 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
checks: 2 healthy: True
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** all([]) is True; choose and document whether an empty report should count as healthy.

````python
# Example 0361: Truthiness and collection length - branch
case = {'healthy': True}

class Report:
    def __init__(self, checks):
        self.checks = tuple(checks)

    def __len__(self):
        return len(self.checks)

    def __bool__(self):
        return all(self.checks)

report = Report([True, case["healthy"]])
assert len(report) == 2
assert bool(report) == case["healthy"]
print("checks:", len(report), "healthy:", bool(report))
````

## Pattern 038: A repeatably iterable inventory

**Core concept:** Return a fresh iterator over the owned collection.

**Network-engineering purpose:** Multiple report consumers can traverse the same inventory independently.

### Read the implementation in this order
1. Store a tuple snapshot.
2. Return iter(self._names) in __iter__.
3. Convert the inventory to a list twice.
4. Verify that the second traversal is not exhausted.

**Common mistake / boundary:** Returning one stored iterator makes later traversals unexpectedly empty.

**Practice after the ten worked scenarios:** Use a generator expression to yield only hostnames matching a site prefix.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0371 - A repeatably iterable inventory: branch scenario

**Scenario 01/10; pattern 038; Intermediate.** This is a self-contained offline exercise. Multiple report consumers can traverse the same inventory independently.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Store a tuple snapshot. Verify that the second traversal is not exhausted. Follow the class-specific steps under Pattern 038 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['branch-edge-01', 'lab-spare'] ['branch-edge-01', 'lab-spare']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Returning one stored iterator makes later traversals unexpectedly empty.

````python
# Example 0371: A repeatably iterable inventory - branch
case = {'hostname': 'branch-edge-01'}

class Inventory:
    def __init__(self, names):
        self._names = tuple(names)

    def __iter__(self):
        return iter(self._names)

inventory = Inventory([case["hostname"], "lab-spare"])
first, second = list(inventory), list(inventory)
assert first == second == [case["hostname"], "lab-spare"]
print(first, second)
````

## Pattern 039: Mapping-style device lookup

**Core concept:** Subscription delegates to __getitem__ and preserves missing-key errors.

**Network-engineering purpose:** A small inventory index can feel natural without pretending a missing device exists.

### Read the implementation in this order
1. Copy a dictionary into the index.
2. Implement __getitem__ by delegating lookup.
3. Use index[hostname].
4. Exercise the missing-key branch explicitly.

**Common mistake / boundary:** Returning None for every missing item can move a clear lookup error far downstream.

**Practice after the ten worked scenarios:** Implement __contains__ and discuss whether inheritance from Mapping would help.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0381 - Mapping-style device lookup: branch scenario

**Scenario 01/10; pattern 039; Intermediate.** This is a self-contained offline exercise. A small inventory index can feel natural without pretending a missing device exists.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `site` | `'branch'` |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Copy a dictionary into the index. Exercise the missing-key branch explicitly. Follow the class-specific steps under Pattern 039 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
unknown host rejected
branch
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Returning None for every missing item can move a clear lookup error far downstream.

````python
# Example 0381: Mapping-style device lookup - branch
case = {'site': 'branch', 'hostname': 'branch-edge-01'}

class DeviceIndex:
    def __init__(self, values):
        self._values = dict(values)

    def __getitem__(self, hostname):
        return self._values[hostname]

index = DeviceIndex({case["hostname"]: case["site"]})
assert index[case["hostname"]] == case["site"]
try:
    index["absent"]
except KeyError:
    print("unknown host rejected")
else:
    raise AssertionError("missing key hidden")
print(index[case["hostname"]])
````

## Pattern 040: Context-managed simulated session

**Core concept:** A context manager provides deterministic cleanup even when a block fails.

**Network-engineering purpose:** Session-like resources need explicit lifetime management rather than reliance on __del__.

### Read the implementation in this order
1. Open the simulated state in __enter__.
2. Return self for the with target.
3. Close state in __exit__ and return False.

# Chapter 09: Network value models and parsing boundaries

**Level:** Intermediate

**Learning goal:** Represent addresses, routes, VLAN sets, and simplified observations with explicit assumptions.

**A model is not a network protocol implementation.** The examples use Python's ipaddress
objects for address arithmetic rather than splitting dotted strings manually. Membership
answers 'is this address in this network?' and not 'is this host reachable?'. Likewise, a
selected route here is a teaching model, not a full vendor RIB/FIB calculation.

**IPv4 and IPv6.** Use ip_address for either address family, ip_network for a prefix, and
ip_interface for an address with its prefix length. Strict network construction rejects
host bits; strict=False intentionally normalizes to a network boundary. Do not silently mix
address families. IPv6 does not use IPv4 broadcast semantics. Avoid enumerating every address
of a large prefix; a /64 is not a list you should materialize.

**Parsing is a trust boundary.** A parser should accept one documented grammar, normalize
deliberately, and report malformed data rather than convert it into an apparently healthy
status. These lesson parsers accept tiny synthetic formats. They are not robust vendor CLI
parsers and make no claim to handle localized, truncated, paginated, or version-changing output.

**Useful collections.** Sets model allowed VLAN membership and difference; dictionaries model
exact lookup; a sequence plus an explicit key can implement longest-prefix selection for a
small lab. A production routing table may require a prefix tree and more detailed policies.
Keep numeric values numeric when comparing costs or counters; lexical string order is not
numeric order.

**Operational interpretation.** An admin-up interface can still be operationally down. A
matching ACL prefix in this simplified model is not a complete ACL evaluation: protocols,
ports, order, direction, VRF, and platform behavior are intentionally outside that exercise.

**Pause and explain:** Distinguish address membership, route selection, and actual reachability.

## Pattern 041: Prefix containment as a domain operation

**Core concept:** Wrap an ipaddress network behind a meaningful method.

**Network-engineering purpose:** A policy can ask whether a management address belongs to an approved lab subnet.

### Read the implementation in this order
1. Normalize the configured prefix with ip_network.
2. Parse the candidate with ip_address.
3. Use the membership operation.
4. Check one member and one explicitly unrelated address.

**Common mistake / boundary:** Containment does not test ping, routing, ACL permission, or service availability.

**Practice after the ten worked scenarios:** Reject an address-family mismatch explicitly rather than treating it as an ordinary miss.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0401 - Prefix containment as a domain operation: branch scenario

**Scenario 01/10; pattern 041; Intermediate.** This is a self-contained offline exercise. A policy can ask whether a management address belongs to an approved lab subnet.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `prefix` | `'192.0.2.0/28'` |
| `ip` | `'192.0.2.3'` |

**Logic to trace:** Normalize the configured prefix with ip_network. Check one member and one explicitly unrelated address. Follow the class-specific steps under Pattern 041 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
192.0.2.3 True
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Containment does not test ping, routing, ACL permission, or service availability.

````python
# Example 0401: Prefix containment as a domain operation - branch
case = {'prefix': '192.0.2.0/28', 'ip': '192.0.2.3'}

from ipaddress import ip_address, ip_network

class PrefixPolicy:
    def __init__(self, prefix):
        self.network = ip_network(prefix)

    def contains(self, address):
        return ip_address(address) in self.network

policy = PrefixPolicy(case["prefix"])
assert policy.contains(case["ip"])
assert not policy.contains("203.0.113.250")
print(case["ip"], policy.contains(case["ip"]))
````

## Pattern 042: Longest-prefix route lookup

**Core concept:** A route table can select the matching destination with the greatest prefix length.

**Network-engineering purpose:** An engineer can reason about a specific route winning over a default in a small offline model.

### Read the implementation in this order
1. Parse route prefixes once in the constructor.
2. Filter routes containing the destination.
3. Choose max prefixlen or raise when none match.
4. Check the scenario's specific route beats default.

**Common mistake / boundary:** This model omits route installation, administrative preference, metric, ECMP, and forwarding verification.

**Practice after the ten worked scenarios:** Add an equal-prefix tie policy and document the result.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0411 - Longest-prefix route lookup: branch scenario

**Scenario 01/10; pattern 042; Intermediate.** This is a self-contained offline exercise. An engineer can reason about a specific route winning over a default in a small offline model.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `prefix` | `'192.0.2.0/28'` |
| `ip` | `'192.0.2.3'` |

**Logic to trace:** Parse route prefixes once in the constructor. Check the scenario's specific route beats default. Follow the class-specific steps under Pattern 042 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
specific
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This model omits route installation, administrative preference, metric, ECMP, and forwarding verification.

````python
# Example 0411: Longest-prefix route lookup - branch
case = {'prefix': '192.0.2.0/28', 'ip': '192.0.2.3'}

from ipaddress import ip_address, ip_network

class RouteTable:
    def __init__(self, records):
        self.routes = [(ip_network(prefix), hop) for prefix, hop in records]

    def lookup(self, address):
        destination = ip_address(address)
        matches = [item for item in self.routes if destination in item[0]]
        if not matches:
            raise LookupError("no matching route")
        return max(matches, key=lambda item: item[0].prefixlen)[1]

table = RouteTable([("0.0.0.0/0", "default"), (case["prefix"], "specific")])
assert table.lookup(case["ip"]) == "specific"
print(table.lookup(case["ip"]))
````

## Pattern 043: VLAN set reconciliation

**Core concept:** Set difference expresses desired additions and removals without duplicate commands.

**Network-engineering purpose:** A trunk intent model can compare unordered memberships rather than raw CLI text.

### Read the implementation in this order
1. Store allowed VLANs as a set.
2. Compute desired minus observed for additions.
3. Compute observed minus desired for removals.
4. Sort the report for deterministic output.

**Common mistake / boundary:** A set ignores ordering; use it only where ordering has no semantic meaning.

**Practice after the ten worked scenarios:** Add policy validation before producing a plan and reject reserved IDs for your chosen platform.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0421 - VLAN set reconciliation: branch scenario

**Scenario 01/10; pattern 043; Intermediate.** This is a self-contained offline exercise. A trunk intent model can compare unordered memberships rather than raw CLI text.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `vlan` | `1` |

**Logic to trace:** Store allowed VLANs as a set. Sort the report for deterministic output. Follow the class-specific steps under Pattern 043 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
add: [2] remove: [4094]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A set ignores ordering; use it only where ordering has no semantic meaning.

````python
# Example 0421: VLAN set reconciliation - branch
case = {'vlan': 1}

class VlanIntent:
    def __init__(self, allowed):
        self.allowed = set(allowed)

    def diff(self, observed):
        current = set(observed)
        return sorted(self.allowed - current), sorted(current - self.allowed)

intent = VlanIntent([case["vlan"], case["vlan"] + 1])
added, removed = intent.diff([case["vlan"], 4094])
if case["vlan"] + 1 == 4094:
    assert added == [] and removed == []
else:
    assert added == [case["vlan"] + 1] and removed == [4094]
print("add:", added, "remove:", removed)
````

## Pattern 044: Parse a tiny interface observation

**Core concept:** A parser converts one declared external grammar into a validated object.

**Network-engineering purpose:** Telemetry consumers should receive structured status instead of reparsing strings everywhere.

### Read the implementation in this order
1. Require exactly a name and one status token.
2. Accept only up or down.
3. Build the object through a named constructor.
4. Check valid parsing and explicit rejection of malformed state.

**Common mistake / boundary:** Do not use this two-token grammar on arbitrary real-device output.

**Practice after the ten worked scenarios:** Add a test for missing tokens and a separate test for extra tokens.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0431 - Parse a tiny interface observation: branch scenario

**Scenario 01/10; pattern 044; Intermediate.** This is a self-contained offline exercise. Telemetry consumers should receive structured status instead of reparsing strings everywhere.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |
| `healthy` | `True` |

**Logic to trace:** Require exactly a name and one status token. Check valid parsing and explicit rejection of malformed state. Follow the class-specific steps under Pattern 044 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
malformed observation rejected
Observation(interface='Ethernet1/1', up=True)
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not use this two-token grammar on arbitrary real-device output.

````python
# Example 0431: Parse a tiny interface observation - branch
case = {'interface': 'Ethernet1/1', 'healthy': True}

from dataclasses import dataclass

@dataclass
class Observation:
    interface: str
    up: bool

    @classmethod
    def parse(cls, text):
        fields = text.split()
        if len(fields) != 2 or fields[1] not in {"up", "down"}:
            raise ValueError("expected: interface up|down")
        return cls(fields[0], fields[1] == "up")

word = "up" if case["healthy"] else "down"
item = Observation.parse(f"{case['interface']} {word}")
assert item.up == case["healthy"]
try:
    Observation.parse("port unknown")
except ValueError:
    print("malformed observation rejected")
else:
    raise AssertionError("malformed observation accepted")
print(item)
````

# Chapter 10: Exceptions and bounded resilience

**Level:** Intermediate

**Learning goal:** Report domain failures, preserve causes, and model retries without hiding errors.

**Failure is part of the interface.** A read can time out, a parser can reject a payload, and
a policy can reject an intent. Decide what each caller may catch. Custom exceptions communicate
domain meaning; exception chaining (`raise DomainError(...) from error`) preserves the original
cause so diagnostic detail is not lost.

**Do not catch everything.** Catching Exception and returning an empty dictionary can transform
a failed read into an apparently empty configuration. Catch a narrow, expected error near a
boundary and either recover deliberately or re-raise with context. A validation failure should
usually not be retried; a transient read timeout sometimes can be.

**Retries need budgets.** Limit attempts, use real timeouts in real adapters, and distinguish
transient failures from permanent errors. Backoff and jitter reduce synchronized retry load;
the teaching examples use deterministic schedules and simulated outcomes rather than sleeps.
Retrying a write is unsafe unless the operation is idempotent or you can establish whether
the first attempt took effect.

**Circuit breakers** reject work after repeated failures and later permit a controlled trial.
A two-state sketch is not a complete production breaker. A production design needs a clock,
half-open behavior, concurrency rules, metrics, and careful reset semantics.

**Cleanup and failure reporting.** Use context managers or finally to release resources.
Expected failures in this course are caught only so their exact behavior can be asserted.
Unexpected failures must stop execution, not produce a success-shaped fallback.

**Pause and explain:** Explain which errors are retryable and why blindly retrying a configuration write is dangerous.

## Pattern 046: Custom domain exception

**Core concept:** A dedicated exception lets a caller handle one meaningful category of failure.

**Network-engineering purpose:** A missing interface can be distinguished from an unexpected programming error.

### Read the implementation in this order
1. Define InterfaceNotFound from LookupError.
2. Raise it at the domain lookup boundary.
3. Catch only that exception.
4. Assert the error includes the missing name.

**Common mistake / boundary:** Do not catch every exception around an entire fleet and mark the whole operation successful.

**Practice after the ten worked scenarios:** Attach a hostname field to the exception while keeping the human message concise.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0451 - Custom domain exception: branch scenario

**Scenario 01/10; pattern 046; Intermediate.** This is a self-contained offline exercise. A missing interface can be distinguished from an unexpected programming error.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Define InterfaceNotFound from LookupError. Assert the error includes the missing name. Follow the class-specific steps under Pattern 046 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
unknown interface: missing-port
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not catch every exception around an entire fleet and mark the whole operation successful.

````python
# Example 0451: Custom domain exception - branch
case = {'interface': 'Ethernet1/1'}

class InterfaceNotFound(LookupError):
    pass

class Device:
    def __init__(self, interfaces):
        self.interfaces = set(interfaces)

    def require(self, name):
        if name not in self.interfaces:
            raise InterfaceNotFound(f"unknown interface: {name}")
        return name

device = Device([case["interface"]])
assert device.require(case["interface"]) == case["interface"]
try:
    device.require("missing-port")
except InterfaceNotFound as error:
    assert "missing-port" in str(error)
    print(str(error))
else:
    raise AssertionError("missing interface accepted")
````

## Pattern 047: Preserve a low-level cause

**Core concept:** Exception chaining adds domain context without discarding the original error.

**Network-engineering purpose:** A telemetry conversion failure should identify the field and retain the original ValueError.

### Read the implementation in this order
1. Attempt one narrow conversion.
2. Catch ValueError only.
3. Raise ParseError using from error.
4. Check both domain message and __cause__.

**Common mistake / boundary:** The final error should not dump secrets or entire sensitive payloads.

**Practice after the ten worked scenarios:** Add a supported numeric value path and a separate TypeError policy for None.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0461 - Preserve a low-level cause: branch scenario

**Scenario 01/10; pattern 047; Intermediate.** This is a self-contained offline exercise. A telemetry conversion failure should identify the field and retain the original ValueError.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `counter` | `1000` |

**Logic to trace:** Attempt one narrow conversion. Check both domain message and __cause__. Follow the class-specific steps under Pattern 047 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
invalid counter field ValueError
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** The final error should not dump secrets or entire sensitive payloads.

````python
# Example 0461: Preserve a low-level cause - branch
case = {'counter': 1000}

class ParseError(ValueError):
    pass

class CounterParser:
    def parse(self, text):
        try:
            return int(text)
        except ValueError as error:
            raise ParseError("invalid counter field") from error

parser = CounterParser()
assert parser.parse(str(case["counter"])) == case["counter"]
try:
    parser.parse("not-a-counter")
except ParseError as error:
    assert isinstance(error.__cause__, ValueError)
    print(str(error), type(error.__cause__).__name__)
else:
    raise AssertionError("parse failure hidden")
````

## Pattern 048: Bounded retry of a simulated read

**Core concept:** Retry only a declared transient error and stop at a fixed budget.

**Network-engineering purpose:** A read collector can recover from temporary lab timeouts without an infinite loop.

### Read the implementation in this order
1. Inject an operation with deterministic outcomes.
2. Attempt at most the configured budget.
3. Catch only TimeoutError and re-raise the final one.
4. Check successful retry count against the simulated failures.

**Common mistake / boundary:** This is a read-only simulation; applying the same policy blindly to writes can duplicate changes.

**Practice after the ten worked scenarios:** Add an exhausted-budget case and prove the final TimeoutError propagates.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0471 - Bounded retry of a simulated read: branch scenario

**Scenario 01/10; pattern 048; Intermediate.** This is a self-contained offline exercise. A read collector can recover from temporary lab timeouts without an infinite loop.

**Local inputs** (all are defined again in the code cell):

| Input | Value |

## Pattern 049: Deterministic exponential backoff schedule

**Core concept:** A policy object computes a capped delay rather than sleeping in a unit example.

**Network-engineering purpose:** Retry timing can be tested separately from transport behavior.

### Read the implementation in this order
1. Validate a nonnegative attempt index.
2. Compute base times two to the index.
3. Use __call__ to expose the same computation through policy(index).
4. Check monotonicity and the maximum across a small schedule.

**Common mistake / boundary:** A production schedule usually needs jitter and a total deadline, not only exponential delays.

**Practice after the ten worked scenarios:** Inject a deterministic random source and add bounded jitter without changing the transport.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0481 - Deterministic exponential backoff schedule: branch scenario

**Scenario 01/10; pattern 049; Intermediate.** This is a self-contained offline exercise. Retry timing can be tested separately from transport behavior.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `base_delay` | `1` |

**Logic to trace:** Validate a nonnegative attempt index. Check monotonicity and the maximum across a small schedule. Follow the class-specific steps under Pattern 049 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
[1, 2, 4, 8, 8, 8]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A production schedule usually needs jitter and a total deadline, not only exponential delays.

````python
# Example 0481: Deterministic exponential backoff schedule - branch
case = {'base_delay': 1}

class Backoff:
    def __init__(self, base, cap):
        if base <= 0 or cap < base:
            raise ValueError("invalid backoff bounds")
        self.base, self.cap = base, cap

    def delay(self, attempt):
        if type(attempt) is not int or attempt < 0:
            raise ValueError("attempt index must be nonnegative")
        return min(self.cap, self.base * 2 ** attempt)

    def __call__(self, attempt):
        return self.delay(attempt)

policy = Backoff(case["base_delay"], 8)
delays = [policy(index) for index in range(6)]
assert delays == sorted(delays) and max(delays) <= 8
assert delays[0] == case["base_delay"]
assert policy(0) == policy.delay(0)
print(delays)
````

# Chapter 11: Behavioral and structural design patterns

**Level:** Intermediate

**Learning goal:** Apply strategy, factory, adapter, facade, and decorator only where they simplify a real boundary.

**Patterns are names for recurring design choices, not rules requiring more classes.**
A strategy supplies interchangeable decision logic; a callable is enough for many strategies.
A factory centralizes construction when callers should not select concrete classes themselves.
An adapter translates one interface into another. A facade presents a simple operation over
several collaborators. A decorator object wraps another object to add behavior while keeping
the relevant public contract.

**A networking example.** An inventory parser produces a platform key. A factory picks an
offline adapter. A transport adapter translates a library's run method into the service's send
method. A strategy chooses a compliance threshold. A facade coordinates collection and
evaluation. A metrics decorator records how many reads occurred. Each component has one reason
to change, so tests can replace exactly the dependency they need.

**Object decorator versus @decorator syntax.** Both can wrap behavior, but the syntax and object
lifecycle differ. A function decorator transforms a callable at definition time; preserve
function metadata with functools.wraps when writing one. A wrapper object can instead implement
send and delegate to another transport.

**Avoid pattern inflation.** If a dictionary of callables solves construction, do not add a
metaclass-driven service locator. A factory must reject unknown keys explicitly. An adapter
should translate behavior without lying about error semantics. A facade should not conceal
operational failure behind an empty success report. Composition remains the common mechanism.

**Pause and explain:** Identify which pattern changes an interface and which one preserves it while adding behavior.

## Pattern 051: Strategy object for a threshold policy

**Core concept:** Swap an algorithm through an injected strategy object.

**Network-engineering purpose:** A health evaluator can use stricter or looser policies without branching on environment names.

### Read the implementation in this order
1. Define a policy with evaluate.
2. Store the limit in the policy instance.
3. Inject it into Monitor.
4. Check behavior for the scenario's load.

**Common mistake / boundary:** A strategy should encapsulate one decision, not become a second orchestration service.

**Practice after the ten worked scenarios:** Add a policy that checks a minimum headroom instead of maximum load.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0501 - Strategy object for a threshold policy: branch scenario

**Scenario 01/10; pattern 051; Intermediate.** This is a self-contained offline exercise. A health evaluator can use stricter or looser policies without branching on environment names.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `load` | `20` |
| `threshold` | `80` |

**Logic to trace:** Define a policy with evaluate. Check behavior for the scenario's load. Follow the class-specific steps under Pattern 051 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
True
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

## Pattern 052: Factory with an explicit supported set

**Core concept:** Construction can be selected by a validated external key.

**Network-engineering purpose:** A platform key can select one of a few known offline command renderers.

### Read the implementation in this order
1. Declare concrete renderers.
2. Store their constructors in a registry.
3. Look up the requested key.
4. Convert an unsupported key to an explicit domain-facing ValueError.

**Common mistake / boundary:** Do not import or execute arbitrary names supplied by an untrusted inventory file.

**Practice after the ten worked scenarios:** Inject the registry into the factory to make supported plugins application-controlled.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0511 - Factory with an explicit supported set: branch scenario

**Scenario 01/10; pattern 052; Intermediate.** This is a self-contained offline exercise. A platform key can select one of a few known offline command renderers.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `style` | `'compact'` |

**Logic to trace:** Declare concrete renderers. Convert an unsupported key to an explicit domain-facing ValueError. Follow the class-specific steps under Pattern 052 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
status brief
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not import or execute arbitrary names supplied by an untrusted inventory file.

````python
# Example 0511: Factory with an explicit supported set - branch
case = {'style': 'compact'}

class Compact:
    def render(self):
        return "status brief"

class Verbose:
    def render(self):
        return "status detail"

class Factory:
    types = {"compact": Compact, "verbose": Verbose}

    @classmethod
    def create(cls, key):
        try:
            constructor = cls.types[key]
        except KeyError as error:
            raise ValueError(f"unsupported style: {key}") from error
        return constructor()

key = case["style"]
result = Factory.create(key).render()
assert result == ("status brief" if key == "compact" else "status detail")
print(result)
````

## Pattern 054: Facade coordinates a small report

**Core concept:** A facade offers a clear operation over several collaborators.

**Network-engineering purpose:** A caller can request one report without knowing collection and formatting details.

### Read the implementation in this order
1. Separate the source from the formatter.
2. Inject both into ReportFacade.
3. Collect once and format once.
4. Verify the final report.

**Common mistake / boundary:** A facade should simplify access, not erase error information or own every responsibility.

**Practice after the ten worked scenarios:** Make the source fail and preserve that failure through report().

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0531 - Facade coordinates a small report: branch scenario

**Scenario 01/10; pattern 054; Intermediate.** This is a self-contained offline exercise. A caller can request one report without knowing collection and formatting details.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |
| `healthy` | `True` |

**Logic to trace:** Separate the source from the formatter. Verify the final report. Follow the class-specific steps under Pattern 054 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
branch-edge-01: up
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A facade should simplify access, not erase error information or own every responsibility.

````python
# Example 0531: Facade coordinates a small report - branch
case = {'hostname': 'branch-edge-01', 'healthy': True}

class Source:
    def read(self):
        return case["healthy"]

class Formatter:
    def format(self, hostname, healthy):
        return f"{hostname}: {'up' if healthy else 'down'}"

class ReportFacade:
    def __init__(self, source, formatter):
        self.source, self.formatter = source, formatter

    def report(self, hostname):
        return self.formatter.format(hostname, self.source.read())

result = ReportFacade(Source(), Formatter()).report(case["hostname"])
assert result.endswith("up" if case["healthy"] else "down")
print(result)
````

## Pattern 055: Decorator object counts transport calls

**Core concept:** A wrapper preserves send while adding instrumentation.

**Network-engineering purpose:** A service can observe read volume without changing its replay transport.

### Read the implementation in this order
1. Create a minimal transport.
2. Wrap it with a counting transport.
3. Increment then delegate each call.
4. Check the count and returned value.

**Common mistake / boundary:** Decide whether the counter tracks attempts or successful calls; this one tracks attempts.

**Practice after the ten worked scenarios:** Track successes and failures separately while re-raising transport errors.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0541 - Decorator object counts transport calls: branch scenario

**Scenario 01/10; pattern 055; Intermediate.** This is a self-contained offline exercise. A service can observe read volume without changing its replay transport.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Create a minimal transport. Check the count and returned value. Follow the class-specific steps under Pattern 055 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
ETHERNET1/1 calls: 2
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Decide whether the counter tracks attempts or successful calls; this one tracks attempts.

````python
# Example 0541: Decorator object counts transport calls - branch
case = {'interface': 'Ethernet1/1'}

class ReplayTransport:
    def send(self, command):
        return command.upper()

class CountingTransport:
    def __init__(self, wrapped):
        self.wrapped, self.calls = wrapped, 0

    def send(self, command):
        self.calls += 1
        return self.wrapped.send(command)

transport = CountingTransport(ReplayTransport())
result = transport.send(case["interface"])
transport.send("status")
assert transport.calls == 2
assert result == case["interface"].upper()
print(result, "calls:", transport.calls)
````

# Chapter 12: Events, commands, templates, registries, and composites

**Level:** Intermediate to advanced

**Learning goal:** Use small coordinated objects to express workflows and extension points.

**Observer** delivers an event to registered listeners. An in-memory teaching bus does not
provide durable delivery, retries, ordering across processes, or isolation from slow listeners.
Decide what happens when one listener fails; this notebook's simple bus propagates failures.

**Command objects** represent an action as data plus behavior. This helps review, queuing,
audit, and simulation. An undo action is not automatically safe on real devices: state may
have changed since execution, and external side effects may be irreversible.

**Template method** keeps an overall algorithm in a base class while subclasses supply a
designated step. Use it when that fixed sequence is genuinely shared. If every subclass
needs to alter every step, an injected pipeline is clearer.

**Registries** map stable names to known implementations. Enforce uniqueness and make
registration explicit unless the application genuinely benefits from class-creation hooks.
Unknown entries and duplicates are errors, not cues to load arbitrary external code.

**Composite** lets a group of objects and an individual object expose the same operation.
A site group can sum device counts without every caller special-casing nesting. Real
hierarchies must prevent cycles or define how repeated references are handled.

Keep these implementations small enough that the mechanism is visible. The exercises ask you
to reason about ownership, failure behavior, and idempotency instead of memorizing pattern names.

**Pause and explain:** Explain why an in-memory observer is not a durable queue and why undo is not guaranteed rollback.

## Pattern 056: Observer notifies registered listeners

**Core concept:** Publish an event through explicit listener registration.

**Network-engineering purpose:** A simulated interface event can update an audit collector without hard-coding it into the publisher.

### Read the implementation in this order
1. Store listeners in instance state.
2. Subscribe a callable.
3. Publish to a snapshot of the listener list.
4. Check that the listener received exactly the event sent.

**Common mistake / boundary:** This synchronous bus propagates listener exceptions and has no persistence.

**Practice after the ten worked scenarios:** Add unsubscribe and decide how duplicate subscriptions should behave.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0551 - Observer notifies registered listeners: branch scenario

**Scenario 01/10; pattern 056; Intermediate to advanced.** This is a self-contained offline exercise. A simulated interface event can update an audit collector without hard-coding it into the publisher.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |
| `healthy` | `True` |

**Logic to trace:** Store listeners in instance state. Check that the listener received exactly the event sent. Follow the class-specific steps under Pattern 056 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
[{'interface': 'Ethernet1/1', 'up': True}]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This synchronous bus propagates listener exceptions and has no persistence.

````python
# Example 0551: Observer notifies registered listeners - branch
case = {'interface': 'Ethernet1/1', 'healthy': True}

class EventBus:
    def __init__(self):
        self.listeners = []

    def subscribe(self, listener):
        self.listeners.append(listener)

    def publish(self, event):
        for listener in tuple(self.listeners):
            listener(event)

events = []
bus = EventBus()
bus.subscribe(events.append)
event = {"interface": case["interface"], "up": case["healthy"]}
bus.publish(event)
assert events == [event]
print(events)
````

## Pattern 057: Command object with simulated undo

**Core concept:** An action can retain enough prior state for a local reversal.

**Network-engineering purpose:** A reviewable desired-MTU command can demonstrate execution and rollback on an in-memory model.

### Read the implementation in this order
1. Keep the receiver object separate from the command.
2. Record the old value at execution.
3. Set the proposed value.
4. Undo and verify the original state returns.

**Common mistake / boundary:** Real-device undo needs concurrency and drift checks; this only reverses local assignment.

**Practice after the ten worked scenarios:** Reject undo before execute and reject a second execute until the first is resolved.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0561 - Command object with simulated undo: branch scenario

**Scenario 01/10; pattern 057; Intermediate to advanced.** This is a self-contained offline exercise. A reviewable desired-MTU command can demonstrate execution and rollback on an in-memory model.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Keep the receiver object separate from the command. Undo and verify the original state returns. Follow the class-specific steps under Pattern 057 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
restored MTU: 1500
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Real-device undo needs concurrency and drift checks; this only reverses local assignment.

````python
# Example 0561: Command object with simulated undo - branch
case = {'mtu': 1500}

from dataclasses import dataclass

@dataclass
class Interface:
    mtu: int

class SetMtu:
    def __init__(self, interface, proposed):
        self.interface, self.proposed, self.previous = interface, proposed, None

    def execute(self):
        self.previous = self.interface.mtu
        self.interface.mtu = self.proposed

    def undo(self):
        if self.previous is None:
            raise RuntimeError("command not executed")
        self.interface.mtu = self.previous

interface = Interface(1500)
command = SetMtu(interface, case["mtu"])
command.execute()
assert interface.mtu == case["mtu"]
command.undo()
assert interface.mtu == 1500
print("restored MTU:", interface.mtu)
````

## Pattern 059: Explicit registry rejects duplicates

**Network-engineering purpose:** Known checks can be looked up by name without importing untrusted modules.

### Read the implementation in this order
1. Give each registry its own mapping.
2. Reject a duplicate before storing it.

**Common mistake / boundary:** A module-global mutable registry can leak state between tests and notebook reruns.

````python
# Example 0581: Explicit registry rejects duplicates - branch
case = {'mtu': 1500}

class CheckRegistry:
    def __init__(self):
        self._checks = {}

    def register(self, name, check):
        if name in self._checks:
            raise ValueError(f"duplicate check: {name}")
        self._checks[name] = check

    def run(self, name, value):
        return self._checks[name](value)

registry = CheckRegistry()
registry.register("mtu", lambda value: 576 <= value <= 9216)
assert registry.run("mtu", case["mtu"])
try:
    registry.register("mtu", lambda value: True)
except ValueError:
    print("duplicate registration rejected")
else:
    raise AssertionError("duplicate overwritten")
````

## Pattern 060: Composite inventory groups

**Core concept:** A leaf and a group share the same counting operation.

**Network-engineering purpose:** Nested site inventory can be summarized without flattening every object manually.

### Read the implementation in this order
1. Give a leaf count() returning one.
2. Give a group count() summing child counts.
3. Nest one group within another.
4. Check the recursive result.

**Common mistake / boundary:** This example assumes an acyclic tree with unique ownership; repeated references would count twice.

**Practice after the ten worked scenarios:** Add cycle prevention or explicitly document a DAG counting policy.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

````python
# Example 0591: Composite inventory groups - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    def __init__(self, hostname):
        self.hostname = hostname

    def count(self):
        return 1

class Group:
    def __init__(self, children):
        self.children = tuple(children)

    def count(self):
        return sum(child.count() for child in self.children)

site = Group([Device(case["hostname"]), Group([Device("spare-a"), Device("spare-b")])])
assert site.count() == 3
print("devices:", site.count())
````

# Chapter 13: Desired state and safe change workflows

**Level:** Advanced

**Learning goal:** Plan differences, preserve idempotency, simulate rollback, and keep approvals visible.

**Intent, observation, and plan are separate things.** Intent describes what should be true.
Observation describes what was read at a particular time. A plan describes the proposed
transition. A successful command response is not proof that intent is now satisfied; collect
fresh observations and verify postconditions.

**Idempotency** means repeating an operation has the same final intended effect as performing
it once. A set-style assignment can be idempotent; incrementing a value usually is not.
Reconciliation should produce no changes when observed and desired state already match.
Normalize semantically equivalent data before comparing it.

**Dry run** must not execute writes. A plan should be reviewable before execution and bound to
the exact target, intended change, and baseline. Approval is distinct from validation. These
lessons use boolean approvals only to teach control flow, not as real authorization.

**Transactions and rollback.** A local dictionary can be copied and restored atomically from
one process's point of view, but a multi-device rollout is not a single Python transaction.
Devices may have candidate configurations or commit-confirmed features with their own rules.
A rollback can also fail. Production needs prechecks, backups, staged rollout, postchecks,
drift/concurrency controls, durable audit records, and explicit partial-failure reporting.

**Audit records** should say what was planned, what was attempted, what succeeded, and what
failed, while excluding credentials and unnecessary sensitive configuration. A list of
in-memory events here illustrates semantics but is neither durable nor tamper-evident.

**Pause and explain:** State the difference between a dry-run plan, an approved change, and verified desired state.

## Pattern 061: Desired versus observed diff

**Core concept:** Produce a deterministic change map only for differing fields.

**Network-engineering purpose:** A configuration object can report exactly which intended fields need reconciliation.

### Read the implementation in this order
1. Copy desired fields into the intent.
2. Compare only fields managed by this intent.
3. Record old and new values for changes.
4. Check equality produces an empty plan.

**Common mistake / boundary:** observed.get treats absent and None alike; this toy model only manages non-None MTU values.

**Practice after the ten worked scenarios:** Introduce an explicit missing sentinel for a model where None is a legitimate desired value.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0601 - Desired versus observed diff: branch scenario

**Scenario 01/10; pattern 061; Advanced.** This is a self-contained offline exercise. A configuration object can report exactly which intended fields need reconciliation.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Copy desired fields into the intent. Check equality produces an empty plan. Follow the class-specific steps under Pattern 061 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
{}
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** observed.get treats absent and None alike; this toy model only manages non-None MTU values.

````python
# Example 0601: Desired versus observed diff - branch
case = {'mtu': 1500}

class Intent:
    def __init__(self, desired):
        self.desired = dict(desired)

    def diff(self, observed):
        return {key: (observed.get(key), value)
                for key, value in self.desired.items() if observed.get(key) != value}

intent = Intent({"mtu": case["mtu"]})
plan = intent.diff({"mtu": 1500})
expected = {} if case["mtu"] == 1500 else {"mtu": (1500, case["mtu"])}
assert plan == expected
assert intent.diff({"mtu": case["mtu"]}) == {}
print(plan)
````

## Pattern 062: Idempotent local reconciliation

**Core concept:** Only write when the desired value differs from the current value.

**Network-engineering purpose:** Rerunning the same notebook cell should not count an unnecessary second change.

### Read the implementation in this order
1. Keep simulated state and a write counter.
2. Compare before assignment.
3. Apply the desired value twice.
4. Assert the second call reports no change and adds no write.

**Common mistake / boundary:** Local idempotency does not resolve a real transport timeout with an unknown write outcome.

**Practice after the ten worked scenarios:** Add a fresh observation after a simulated write and verify the resulting state.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0611 - Idempotent local reconciliation: branch scenario

**Scenario 01/10; pattern 062; Advanced.** This is a self-contained offline exercise. Rerunning the same notebook cell should not count an unnecessary second change.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Keep simulated state and a write counter. Assert the second call reports no change and adds no write. Follow the class-specific steps under Pattern 062 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
writes: 0
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Local idempotency does not resolve a real transport timeout with an unknown write outcome.

````python
# Example 0611: Idempotent local reconciliation - branch
case = {'mtu': 1500}

class Reconciler:
    def __init__(self, current):
        self.current, self.writes = current, 0

    def ensure(self, desired):
        if self.current == desired:
            return False
        self.current = desired
        self.writes += 1
        return True

reconciler = Reconciler(1500)
changed = reconciler.ensure(case["mtu"])
assert changed == (case["mtu"] != 1500)
assert reconciler.ensure(case["mtu"]) is False
assert reconciler.writes == int(case["mtu"] != 1500)
print("writes:", reconciler.writes)
````

## Pattern 064: Rollback a failed local transaction

**Core concept:** Snapshot state and restore it if a known postcondition fails.

**Network-engineering purpose:** A simulated rollout can illustrate why applying is not the final success criterion.

### Read the implementation in this order
1. Copy the baseline before mutation.
2. Apply proposed state.
3. Run an explicit postcheck.
4. Restore on the known verification error and propagate the error.

**Common mistake / boundary:** This restores one dictionary only; it does not provide network-wide atomicity.

**Practice after the ten worked scenarios:** Model rollback failure separately and report both the original and rollback errors.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0631 - Rollback a failed local transaction: branch scenario

**Scenario 01/10; pattern 064; Advanced.** This is a self-contained offline exercise. A simulated rollout can illustrate why applying is not the final success criterion.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Copy the baseline before mutation. Restore on the known verification error and propagate the error. Follow the class-specific steps under Pattern 064 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
postcheck failure reported; local state restored
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This restores one dictionary only; it does not provide network-wide atomicity.

````python
# Example 0631: Rollback a failed local transaction - branch
case = {'mtu': 1500}

class VerificationError(RuntimeError):
    pass

class LocalTransaction:
    def __init__(self):
        self.state = {"mtu": 1500}

    def apply(self, mtu, verify):
        before = self.state.copy()
        self.state["mtu"] = mtu
        try:
            if not verify():
                raise VerificationError("postcheck failed")
        except VerificationError:
            self.state = before
            raise

transaction = LocalTransaction()
try:
    transaction.apply(case["mtu"], lambda: False)
except VerificationError:
    assert transaction.state == {"mtu": 1500}
    print("postcheck failure reported; local state restored")
else:
    raise AssertionError("postcheck failure ignored")
````

## Pattern 065: Approval gate and explicit audit outcome

**Core concept:** Validation, approval, and audit are separate responsibilities.

**Network-engineering purpose:** A simulated change should record a denial instead of reporting that it applied successfully.

### Read the implementation in this order
1. Track the current value and audit events.
2. Require approval before mutation.
3. Record denied or applied explicitly.
4. Check the resulting state and outcome agree.

**Common mistake / boundary:** A boolean in a notebook is not a production identity, authorization, or approval system.

**Practice after the ten worked scenarios:** Bind approval to a digest of a specific reviewed plan and reject a changed plan.

### Example 0641 - Approval gate and explicit audit outcome: branch scenario

**Scenario 01/10; pattern 065; Advanced.** This is a self-contained offline exercise. A simulated change should record a denial instead of reporting that it applied successfully.

| Input | Value |

| `mtu` | `1500` |
| `approved` | `True` |

**Logic to trace:** Track the current value and audit events. Check the resulting state and outcome agree. Follow the class-specific steps under Pattern 065 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['applied'] 1500
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints.
````python
# Example 0641: Approval gate and explicit audit outcome - branch
case = {'mtu': 1500, 'approved': True}

class ApprovedChange:
    def __init__(self):
        self.mtu, self.audit = 1500, []

    def apply(self, mtu, *, approved):
        if not approved:
            self.audit.append("denied")
            raise PermissionError("approval required")
        self.mtu = mtu
        self.audit.append("applied")

change = ApprovedChange()
try:
    change.apply(case["mtu"], approved=case["approved"])
except PermissionError:
    assert not case["approved"] and change.mtu == 1500
else:
    assert case["approved"] and change.mtu == case["mtu"]
assert change.audit == (["applied"] if case["approved"] else ["denied"])
print(change.audit, change.mtu)
````

# Chapter 14: Testing object behavior

**Level:** Intermediate to advanced

**Learning goal:** Use fakes, mocks, injected time, rejection checks, and behavioral invariants.

**Test the promise, not the spelling of the implementation.** A good test demonstrates that a
change is rejected, a result is correct, or a dependency is called with required arguments.
Tests that merely check a constant or repeat the implementation's exact expression can miss
the real requirement. Use small known examples with independently understandable answers.

**A fake** is a usable simplified implementation, such as an in-memory transport with a call
log. **A mock** records interactions and can assert required calls. Autospeccing constrains a
mock to a known API, reducing accidental acceptance of misspelled methods. Neither replaces
integration testing against a reviewed lab environment.

**Inject time.** Expiry, retries, and cache tests should not sleep for real seconds. Pass a
clock callable and advance a fake clock deliberately. This makes boundary conditions such as
exact expiry deterministic and fast.

**Negative tests** must prove rejection happened. In a try/except test, add an else branch
that fails if the expected exception did not occur. Check state after the failure so an
invalid update cannot partially mutate the object. In unittest use assertRaises for the same
intent. Catch a specific expected exception, not all errors.

**Property-style thinking** asks what should always hold: applying intent twice should produce
no second change; snapshots should not mutate when the source changes; normalization should
be stable on repeated application. These bounded examples use loops and assertions, not an
external property-testing dependency. Broader generated tests can complement, not replace,
explicit edge-case examples.

**Pause and explain:** Name one interaction worth mocking and one outcome better checked through a real fake.

## Pattern 066: Test a fake's interaction log

**Core concept:** A deterministic fake can verify both results and required calls.

**Network-engineering purpose:** A status reader should issue exactly one read command for the selected interface.

### Read the implementation in this order
1. Record commands in a fake.
2. Inject it into a reader.
3. Run one read.
4. Assert exact command order and the returned response.

**Common mistake / boundary:** A fake that behaves unlike the real adapter can give false confidence.

**Practice after the ten worked scenarios:** Add an adapter contract test suite that can be reused for fake and lab implementations.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0651 - Test a fake's interaction log: branch scenario

**Scenario 01/10; pattern 066; Intermediate to advanced.** This is a self-contained offline exercise. A status reader should issue exactly one read command for the selected interface.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Record commands in a fake. Assert exact command order and the returned response. Follow the class-specific steps under Pattern 066 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['read Ethernet1/1']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A fake that behaves unlike the real adapter can give false confidence.

````python
# Example 0651: Test a fake's interaction log - branch
case = {'interface': 'Ethernet1/1'}

class Fake:
    def __init__(self):
        self.commands = []

    def send(self, command):
        self.commands.append(command)
        return "sample"

class Reader:
    def __init__(self, transport):
        self.transport = transport

    def read(self, name):
        return self.transport.send("read " + name)

fake = Fake()
assert Reader(fake).read(case["interface"]) == "sample"
assert fake.commands == ["read " + case["interface"]]
print(fake.commands)
````

## Pattern 067: Autospecced mock for a collaborator

**Core concept:** An autospecced mock checks interaction against a real API shape.

**Network-engineering purpose:** A transport call can be verified without enabling actual networking.

### Read the implementation in this order
1. Define the intended Transport API.
2. Create an instance mock using create_autospec.
3. Set its deterministic return value.
4. Assert the exact call and returned result.

**Common mistake / boundary:** Interaction checks alone do not prove that a parser or domain policy is correct.

**Practice after the ten worked scenarios:** Make a service accept this mock and assert it calls send once rather than testing the mock directly.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0661 - Autospecced mock for a collaborator: branch scenario

**Scenario 01/10; pattern 067; Intermediate to advanced.** This is a self-contained offline exercise. A transport call can be verified without enabling actual networking.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Define the intended Transport API. Assert the exact call and returned result. Follow the class-specific steps under Pattern 067 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
expected transport interaction verified
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Interaction checks alone do not prove that a parser or domain policy is correct.

````python
# Example 0661: Autospecced mock for a collaborator - branch
case = {'interface': 'Ethernet1/1'}

from unittest.mock import create_autospec

class Transport:
    def send(self, command: str) -> str:
        raise NotImplementedError

class Reader:
    def __init__(self, transport):
        self.transport = transport

    def read(self, interface):
        return self.transport.send(f"read {interface}")

transport = create_autospec(Transport, instance=True, spec_set=True)
transport.send.return_value = "up"
result = Reader(transport).read(case["interface"])
transport.send.assert_called_once_with(f"read {case['interface']}")
assert result == "up"
print("expected transport interaction verified")
````

## Pattern 068: Injected clock tests exact expiry

**Core concept:** Time-based state can be deterministic when time is a dependency.

**Network-engineering purpose:** An observation cache should reject a value at its documented expiry boundary.

### Read the implementation in this order
1. Use a callable clock with mutable test time.
2. Store an absolute expiry.
3. Read successfully before expiry.
4. Advance to exact expiry and verify a stale-data error.

**Common mistake / boundary:** Use a monotonic clock for elapsed-time logic in a real process; wall clocks can jump.

**Practice after the ten worked scenarios:** Test one instant just before expiry and one just after without any sleeps.

````python
# Example 0671: Injected clock tests exact expiry - branch
case = {'hostname': 'branch-edge-01', 'ttl': 5}

class Clock:
    def __init__(self):
        self.now = 0

    def __call__(self):
        return self.now

class Cache:
    def __init__(self, clock, ttl, value):
        self.clock, self.expires, self.value = clock, clock() + ttl, value

    def get(self):
        if self.clock() >= self.expires:
            raise LookupError("observation expired")
        return self.value

clock = Clock()
cache = Cache(clock, case["ttl"], case["hostname"])
assert cache.get() == case["hostname"]
clock.now = case["ttl"]
try:
    cache.get()
except LookupError:
    print("exact expiry rejected")
else:
    raise AssertionError("stale observation returned")
````

## Pattern 069: Unit test a rejected state transition

**Core concept:** A negative test checks both the exception and unchanged state.

**Network-engineering purpose:** A bad VLAN proposal must not overwrite the previous valid value.

### Read the implementation in this order
1. Define explicit runtime validation.
2. Use unittest.TestCase for a focused test.
3. Assert ValueError for the bad update.
4. Verify the existing VLAN is unchanged.

**Common mistake / boundary:** Do not test only that some exception occurred; a programming error might satisfy that weak test.

**Practice after the ten worked scenarios:** Add subTest cases for 0, 4095, True, and a string.

### Example 0681 - Unit test a rejected state transition: branch scenario

**Scenario 01/10; pattern 069; Intermediate to advanced.** This is a self-contained offline exercise. A bad VLAN proposal must not overwrite the previous valid value.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `vlan` | `1` |

**Logic to trace:** Define explicit runtime validation. Verify the existing VLAN is unchanged. Follow the class-specific steps under Pattern 069 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
rejection and unchanged state verified
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not test only that some exception occurred; a programming error might satisfy that weak test.

````python
# Example 0681: Unit test a rejected state transition - branch
case = {'vlan': 1}

import unittest

class AccessPort:
    def __init__(self, vlan):
        self.set_vlan(vlan)

    def set_vlan(self, vlan):
        if type(vlan) is not int or not 1 <= vlan <= 4094:
            raise ValueError("invalid VLAN")
        self.vlan = vlan

class PortTests(unittest.TestCase):
    def runTest(self):
        port = AccessPort(case["vlan"])
        with self.assertRaisesRegex(ValueError, "invalid VLAN"):
            port.set_vlan(0)
        self.assertEqual(port.vlan, case["vlan"])

result = unittest.TestResult()
PortTests().run(result)
assert result.wasSuccessful(), (result.errors, result.failures)
assert result.testsRun == 1
print("rejection and unchanged state verified")
````

## Pattern 070: Check an idempotency invariant across values

**Core concept:** Behavioral properties can be checked over a bounded set of meaningful inputs.

**Network-engineering purpose:** A reconciliation helper should do no extra work when the same intent is applied again.

### Read the implementation in this order
1. Create a reconciler with a write counter.
2. Try normal and boundary MTU values.
3. Apply each value once, then again.
4. Assert the second call leaves the write count unchanged.

**Common mistake / boundary:** A bounded loop is useful evidence, not a mathematical proof over every possible input.

**Practice after the ten worked scenarios:** Add a property about final state and a rejection property for out-of-policy values.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0691 - Check an idempotency invariant across values: branch scenario

**Scenario 01/10; pattern 070; Intermediate to advanced.** This is a self-contained offline exercise. A reconciliation helper should do no extra work when the same intent is applied again.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Create a reconciler with a write counter. Assert the second call leaves the write count unchanged. Follow the class-specific steps under Pattern 070 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
idempotency holds for normal and boundary inputs
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A bounded loop is useful evidence, not a mathematical proof over every possible input.

````python
# Example 0691: Check an idempotency invariant across values - branch
case = {'mtu': 1500}

class Reconciler:
    def __init__(self):
        self.value, self.writes = None, 0

    def ensure(self, value):
        if self.value != value:
            self.value = value
            self.writes += 1

reconciler = Reconciler()
for desired in [576, 1500, case["mtu"], 9216]:
    reconciler.ensure(desired)
    writes = reconciler.writes
    reconciler.ensure(desired)
    assert reconciler.writes == writes
    assert reconciler.value == desired
print("idempotency holds for normal and boundary inputs")
````

# Chapter 15: Serialization, schemas, and explicit data boundaries

**Level:** Intermediate to advanced

**Learning goal:** Serialize domain values deliberately and reject unsupported or malformed records.

**Serialization** converts a model to a transportable representation, not to executable code.
JSON supports a limited set of primitive shapes. An ipaddress value or Enum needs an explicit
representation. Avoid dumping arbitrary __dict__ contents: internal fields, caches, and
credentials can leak, and the resulting schema changes when implementation details change.

**Deserialization is validation.** Require the fields and types your schema promises, reject
unknown schema versions, and normalize only intentionally. JSON decoding alone says nothing
about whether a VLAN, hostname, or MTU is acceptable. bool is a subclass of int in Python, so
strict numeric schemas may need a deliberate check.

**Version your records.** A schema_version field creates an explicit compatibility boundary.
Migrate old formats intentionally; do not silently treat an unsupported version as today's
format. Decide whether unknown fields are rejected, ignored, or preserved. This chapter's
small schemas prefer explicit rejection for clarity.

**Enums** give named states with a defined allowed set. Store the stable value rather than
relying on a debug repr. A redacted report should be constructed from an allowlist of public
fields. Simply deleting a key called password does not guarantee all sensitive data is gone.

**Snapshots** should express ownership. Returning a copied mapping prevents some accidental
mutation; immutable nested values or deeper copies may be needed for nested structures.
These exercises use strings, integers, tuples, and JSON records, never real secrets.

**Pause and explain:** Explain why JSON decoding is not schema validation and why dumping __dict__ is risky.

## Pattern 071: JSON round trip through explicit fields

**Core concept:** A domain model controls its public serialized schema.

## Pattern 072: Versioned schema rejects unsupported records

**Core concept:** A version field prevents accidental interpretation of incompatible data.

**Network-engineering purpose:** A saved intent should not be applied using assumptions from a different record format.

### Read the implementation in this order
1. Require exactly the declared keys.
2. Check version and strict MTU type.
3. Validate the model's MTU range.
4. Accept a valid record and reject a newer version.

**Common mistake / boundary:** Do not silently default an unknown version to a known schema.

**Practice after the ten worked scenarios:** Implement an explicit version-0-to-version-1 migration as a separate function.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0711 - Versioned schema rejects unsupported records: branch scenario

**Scenario 01/10; pattern 072; Intermediate to advanced.** This is a self-contained offline exercise. A saved intent should not be applied using assumptions from a different record format.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Require exactly the declared keys. Accept a valid record and reject a newer version. Follow the class-specific steps under Pattern 072 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
unsupported schema version
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not silently default an unknown version to a known schema.

````python
# Example 0711: Versioned schema rejects unsupported records - branch
case = {'mtu': 1500}

class IntentRecord:
    def __init__(self, mtu):
        self.mtu = mtu

    @classmethod
    def from_mapping(cls, record):
        if set(record) != {"schema_version", "mtu"}:
            raise ValueError("unexpected schema fields")
        if type(record["schema_version"]) is not int or record["schema_version"] != 1:
            raise ValueError("unsupported schema version")
        if type(record["mtu"]) is not int or not 576 <= record["mtu"] <= 9216:
            raise ValueError("invalid MTU")
        return cls(record["mtu"])

intent = IntentRecord.from_mapping({"schema_version": 1, "mtu": case["mtu"]})
assert intent.mtu == case["mtu"]
try:
    IntentRecord.from_mapping({"schema_version": 2, "mtu": case["mtu"]})
except ValueError as error:
    assert "version" in str(error)
    print(str(error))
else:
    raise AssertionError("unsupported version accepted")
````

## Pattern 074: Allowlist a public report

**Core concept:** Build the safe report from approved fields rather than copying every internal field.

**Network-engineering purpose:** An inventory object may contain internal notes that do not belong in a user-facing summary.

### Read the implementation in this order
1. Keep internal metadata separate.
2. Return only the approved hostname and site.
3. Verify the exact public key set.
4. Check internal-only data is not in the serialized report.

**Common mistake / boundary:** An allowlist limits accidental disclosure but does not replace a data-handling review.

**Practice after the ten worked scenarios:** Add a nested field and decide explicitly whether it is safe for the public report.

````python
# Example 0731: Allowlist a public report - branch
case = {'site': 'branch', 'hostname': 'branch-edge-01'}

import json

class InventoryItem:
    def __init__(self, hostname, site):
        self.hostname, self.site = hostname, site
        self._internal_note = "lab-only-maintenance-note"

    def public_report(self):
        return {"hostname": self.hostname, "site": self.site}

item = InventoryItem(case["hostname"], case["site"])
report = item.public_report()
assert set(report) == {"hostname", "site"}
assert item._internal_note not in json.dumps(report)
print(json.dumps(report, sort_keys=True))
````

## Pattern 075: Snapshot ownership at a boundary

**Core concept:** A method can return independent nested data rather than a live internal mapping.

**Network-engineering purpose:** A caller editing a report should not silently change stored interface intent.

### Read the implementation in this order
1. Store the model's nested state.
2. Deep-copy only serializable configuration data for a snapshot.
3. Modify the returned snapshot.
4. Check that internal state remains at its original value.

**Common mistake / boundary:** Do not deep-copy connection objects, locks, or other external resources.

**Practice after the ten worked scenarios:** Use frozen dataclasses and tuples to design an immutable snapshot instead.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0741 - Snapshot ownership at a boundary: branch scenario

**Scenario 01/10; pattern 075; Intermediate to advanced.** This is a self-contained offline exercise. A caller editing a report should not silently change stored interface intent.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |
| `mtu` | `1500` |

**Logic to trace:** Store the model's nested state. Check that internal state remains at its original value. Follow the class-specific steps under Pattern 075 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
{'Ethernet1/1': {'mtu': 1500}}
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Do not deep-copy connection objects, locks, or other external resources.

````python
# Example 0741: Snapshot ownership at a boundary - branch
case = {'interface': 'Ethernet1/1', 'mtu': 1500}

from copy import deepcopy

class Configuration:
    def __init__(self, interface, mtu):
        self._state = {interface: {"mtu": mtu}}

    def snapshot(self):
        return deepcopy(self._state)

configuration = Configuration(case["interface"], case["mtu"])
snapshot = configuration.snapshot()
snapshot[case["interface"]]["mtu"] = 1
assert configuration.snapshot()[case["interface"]]["mtu"] == case["mtu"]
print(configuration.snapshot())
````

# Chapter 16: Advanced object construction and attribute mechanics

**Level:** Advanced

**Learning goal:** Understand descriptors, slots, caching, __new__, and class-creation hooks without overusing them.

**Descriptors** define attribute behavior through methods such as __get__, __set__, and
__set_name__. Properties and ordinary bound methods rely on descriptor mechanisms. A
reusable validating descriptor can enforce the same invariant across several fields. Access
through the class should normally return the descriptor itself when instance is None.

**Attribute lookup** is richer than 'instance dictionary then class'. Data descriptors can
take precedence over instance dictionary entries; non-data descriptors such as ordinary
methods participate differently. __getattr__ is a fallback for missing attributes.
__getattribute__ intercepts ordinary attribute access and is easy to make recursively broken.
Prefer a property or explicit method unless interception is truly required.

**__slots__** declares a set of attribute storage names for instances and can remove their
ordinary __dict__ when the hierarchy permits it. It is not a security boundary, deep
immutability, or a universal performance fix. A base class with a dictionary can change the
result. Weak references and inheritance require deliberate design.

**Cached properties** compute once and store a result. They suit expensive derived values only
when invalidation semantics are clear. A cached value can become stale if inputs change.
Concurrent access may also compute a value more than once, so do not rely on exactly-once
side effects.

**Creation hooks.** __new__ creates instances and is useful when subclassing immutable built-ins.
__init_subclass__ customizes subclass creation without requiring a custom metaclass.
A metaclass controls class creation itself; `type(SomeClass)` is its metaclass. Most network
automation should prefer ordinary classes and explicit registries over metaclass machinery.

**Pause and explain:** Choose a property before a descriptor unless reuse justifies it; explain cache invalidation.

## Pattern 076: Reusable validating descriptor

**Core concept:** One descriptor can validate assignment to a declared model field.

**Network-engineering purpose:** Several numeric configuration fields can share range-checking machinery.

### Read the implementation in this order
1. Capture a backing name with __set_name__.
2. Return self for access through the class.
3. Validate before writing through instance storage.
4. Check valid construction and preserved state after rejection.

**Common mistake / boundary:** This descriptor uses __dict__; it would need a different storage design for dictionary-free slots.

**Practice after the ten worked scenarios:** Add a second descriptor field with a different range and test both independently.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0751 - Reusable validating descriptor: branch scenario

**Scenario 01/10; pattern 076; Advanced.** This is a self-contained offline exercise. Several numeric configuration fields can share range-checking machinery.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Capture a backing name with __set_name__. Check valid construction and preserved state after rejection. Follow the class-specific steps under Pattern 076 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
1500
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This descriptor uses __dict__; it would need a different storage design for dictionary-free slots.

````python
# Example 0751: Reusable validating descriptor - branch
case = {'mtu': 1500}

class BoundedInteger:
    def __init__(self, low, high):
        self.low, self.high = low, high

    def __set_name__(self, owner, name):
        self.storage = "_" + name

    def __get__(self, instance, owner=None):
        if instance is None:
            return self
        return instance.__dict__[self.storage]

    def __set__(self, instance, value):
        if type(value) is not int or not self.low <= value <= self.high:
            raise ValueError("outside declared range")
        instance.__dict__[self.storage] = value

class Interface:
    mtu = BoundedInteger(576, 9216)

    def __init__(self, mtu):
        self.mtu = mtu

interface = Interface(case["mtu"])
try:
    interface.mtu = 0
except ValueError:
    assert interface.mtu == case["mtu"]
else:
    raise AssertionError("descriptor accepted invalid value")
assert isinstance(Interface.mtu, BoundedInteger)
print(interface.mtu)
````

## Pattern 077: Slots catch accidental field names

**Core concept:** A dictionary-free slotted object restricts ordinary instance attribute names.

**Network-engineering purpose:** A misspelled hostname field can fail immediately instead of silently creating a new attribute.

### Read the implementation in this order
1. Declare slots on a direct object subclass.
2. Initialize the declared field.
3. Try writing a misspelled field.
4. Verify AttributeError and the original field.

**Common mistake / boundary:** Slots do not make values immutable and do not replace validation or access control.

**Practice after the ten worked scenarios:** Compare vars() behavior for a normal instance and a slotted instance.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

````python
# Example 0761: Slots catch accidental field names - branch
case = {'hostname': 'branch-edge-01'}

class Device:
    __slots__ = ("hostname",)

    def __init__(self, hostname):
        self.hostname = hostname

device = Device(case["hostname"])
try:
    device.hostnmae = "typo"
except AttributeError:
    print("misspelled field rejected")
else:
    raise AssertionError("unexpected attribute accepted")
assert device.hostname == case["hostname"]
assert not hasattr(device, "__dict__")
````

## Pattern 078: Cached derived value and invalidation

**Core concept:** cached_property stores a computed result until its cached attribute is deleted.

**Network-engineering purpose:** A normalized inventory summary can be cached, but changes to source data must invalidate it.

### Read the implementation in this order
1. Define the cached property.
2. Read it once to populate the cache.
3. Change the input and observe the old cached value.
4. Delete the cache entry and verify recomputation.

**Common mistake / boundary:** Caching mutable or externally changing observations without expiry can return dangerously stale data.

**Practice after the ten worked scenarios:** Replace manual deletion with an update method that invalidates automatically.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

````python
# Example 0771: Cached derived value and invalidation - branch
case = {'hostname': 'branch-edge-01'}

from functools import cached_property

class InventoryLabel:
    def __init__(self, hostname):
        self.hostname = hostname

    @cached_property
    def normalized(self):
        return self.hostname.lower()

label = InventoryLabel(case["hostname"])
assert label.normalized == case["hostname"]
label.hostname = "CHANGED"
assert label.normalized == case["hostname"]
del label.normalized
assert label.normalized == "changed"
print(label.normalized)
````

## Pattern 079: Validate an immutable value in __new__

**Core concept:** An immutable built-in subclass must choose its value during instance creation.

**Network-engineering purpose:** A small lab VLAN value can behave numerically while rejecting invalid construction.

### Read the implementation in this order
1. Override __new__ on an int subclass.
2. Validate before creating the immutable integer.
3. Delegate construction to super().__new__.
4. Check numeric behavior and invalid-value rejection.

**Common mistake / boundary:** Subclassing built-ins is advanced; a frozen wrapper dataclass is often clearer for a domain value.

**Practice after the ten worked scenarios:** Use composition instead and compare equality semantics with ordinary integers.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0781 - Validate an immutable value in __new__: branch scenario

**Scenario 01/10; pattern 079; Advanced.** This is a self-contained offline exercise. A small lab VLAN value can behave numerically while rejecting invalid construction.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `vlan` | `1` |

**Logic to trace:** Override __new__ on an int subclass. Check numeric behavior and invalid-value rejection. Follow the class-specific steps under Pattern 079 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
immutable value rejected invalid construction
1
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Subclassing built-ins is advanced; a frozen wrapper dataclass is often clearer for a domain value.

````python
# Example 0781: Validate an immutable value in __new__ - branch
case = {'vlan': 1}

class VlanId(int):
    def __new__(cls, value):
        if type(value) is not int or not 1 <= value <= 4094:
            raise ValueError("invalid lab VLAN")
        return super().__new__(cls, value)

vlan = VlanId(case["vlan"])
assert int(vlan) == case["vlan"]
assert vlan + 1 == case["vlan"] + 1
try:
    VlanId(0)
except ValueError:
    print("immutable value rejected invalid construction")
else:
    raise AssertionError("invalid VLAN created")
print(vlan)
````

## Pattern 080: Register subclasses with __init_subclass__

**Core concept:** A subclass creation hook can enforce an extension invariant.

**Network-engineering purpose:** A controlled plugin family can require unique short names at definition time.

### Read the implementation in this order
1. Define the registry on a local base class.
2. Accept a named class-creation keyword.
3. Call super's hook and reject duplicate names.
4. Check the concrete class was registered and inspect its metaclass.

**Common mistake / boundary:** Implicit import-time registration can hide dependencies; explicit registration is usually easier to audit.

**Practice after the ten worked scenarios:** Write the same registry using an explicit register call and compare discoverability.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0791 - Register subclasses with __init_subclass__: branch scenario

**Scenario 01/10; pattern 080; Advanced.** This is a self-contained offline exercise. A controlled plugin family can require unique short names at definition time.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |

**Logic to trace:** Define the registry on a local base class. Check the concrete class was registered and inspect its metaclass. Follow the class-specific steps under Pattern 080 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['mtu'] type
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Implicit import-time registration can hide dependencies; explicit registration is usually easier to audit.

````python
# Example 0791: Register subclasses with __init_subclass__ - branch
case = {'mtu': 1500}

class Check:
    registry = {}

    def __init_subclass__(cls, *, name, **kwargs):
        super().__init_subclass__(**kwargs)
        if name in Check.registry:
            raise ValueError("duplicate plugin name")
        Check.registry[name] = cls

class MtuCheck(Check, name="mtu"):
    def run(self, value):
        return 576 <= value <= 9216

assert Check.registry["mtu"] is MtuCheck
assert Check.registry["mtu"]().run(case["mtu"])
assert type(MtuCheck) is type
print(sorted(Check.registry), type(MtuCheck).__name__)
````

# Chapter 17: Concurrency and asynchronous object APIs

**Level:** Advanced

**Learning goal:** Bound concurrent work, preserve errors, and avoid blocking the notebook event loop.

**Concurrency** means tasks can make progress over overlapping periods. It is not the same as
CPU parallelism. asyncio is cooperative: a coroutine gives other tasks opportunities to run
at awaits. A normal blocking transport call inside async def still blocks the event loop.
Use an asynchronous adapter or move appropriate blocking I/O to a bounded thread pool.

**Notebook event loops.** Jupyter usually already has an event loop. These cells use top-level
await rather than asyncio.run inside the notebook. In a normal script, put equivalent work
in async main() and call asyncio.run(main()) once at the entry point. Each async cell defines
all its classes and imports itself.

**Structured concurrency.** A TaskGroup owns a collection of tasks; exiting its context waits
for them and handles task failures according to its contract. Do not create detached tasks
and forget their results. Cancellation is a control signal; cleanup belongs in finally or an
async context manager, and CancelledError should normally propagate after cleanup.

**Bound work.** A semaphore limits simultaneous operations, not total queue size. A large
inventory may also need a bounded producer/consumer queue. Timeouts and rate limits address
different problems. Avoid launching a task for every device in a very large fleet at once.

**Shared state.** Threads need synchronization around compound updates. A Lock can protect an
invariant, but keep critical sections short and do not hold a regular blocking lock across
an await. ThreadPoolExecutor is suitable for bounded blocking I/O; ordinary CPU-bound Python
work and interpreter build details require separate performance reasoning. These lessons
use tiny local workloads and no real network, real sleeps, or long-running background tasks.

**Pause and explain:** Explain top-level await, why blocking I/O still blocks inside async def, and what a semaphore limits.

## Pattern 082: Semaphore bounds in-flight operations

**Core concept:** A semaphore guards a concurrent section around each operation.

**Network-engineering purpose:** A simulated poller should never exceed its configured in-flight budget.

### Read the implementation in this order
1. Create one shared semaphore.
2. Track active and peak counts inside its async context.
3. Use finally to decrement active.
4. Gather a small workload and assert the measured peak is within the limit.

**Common mistake / boundary:** A semaphore alone does not bound memory used by a huge list of scheduled tasks.

**Practice after the ten worked scenarios:** Use a bounded queue and a fixed number of workers for a much larger input.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0811 - Semaphore bounds in-flight operations: branch scenario

**Scenario 01/10; pattern 082; Advanced.** This is a self-contained offline exercise. A simulated poller should never exceed its configured in-flight budget.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |
| `concurrency` | `1` |

**Logic to trace:** Create one shared semaphore. Gather a small workload and assert the measured peak is within the limit. Follow the class-specific steps under Pattern 082 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
peak: 1 limit: 1
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A semaphore alone does not bound memory used by a huge list of scheduled tasks.

````python
# Example 0811: Semaphore bounds in-flight operations - branch
case = {'hostname': 'branch-edge-01', 'concurrency': 1}

import asyncio

class BoundedPoller:
    def __init__(self, limit):
        self.semaphore = asyncio.Semaphore(limit)
        self.active, self.peak = 0, 0

    async def read(self, hostname):
        async with self.semaphore:
            self.active += 1
            self.peak = max(self.peak, self.active)
            try:
                await asyncio.sleep(0)
                return hostname
            finally:
                self.active -= 1

limit = case["concurrency"]
poller = BoundedPoller(limit)
names = [case["hostname"] + f"-{index}" for index in range(6)]
result = await asyncio.gather(*(poller.read(name) for name in names))
assert result == names
assert poller.peak == min(limit, len(names)) and poller.active == 0
print("peak:", poller.peak, "limit:", limit)
````

## Pattern 083: TaskGroup owns concurrent tasks

**Core concept:** A structured task scope keeps task lifetime tied to the calling operation.

**Network-engineering purpose:** A small fleet collection waits for every child before building its final report.

### Read the implementation in this order
1. Define an async collector method.
2. Create tasks inside TaskGroup.
3. Read results only after the group exits.
4. Check the report contains all expected values.

**Common mistake / boundary:** Production callers must handle grouped failures deliberately rather than assume all tasks succeed.

**Practice after the ten worked scenarios:** Make one task raise a domain error and inspect the resulting ExceptionGroup in a separate exercise.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0821 - TaskGroup owns concurrent tasks: branch scenario

**Scenario 01/10; pattern 083; Advanced.** This is a self-contained offline exercise. A small fleet collection waits for every child before building its final report.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Define an async collector method. Check the report contains all expected values. Follow the class-specific steps under Pattern 083 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
[{'hostname': 'branch-edge-01', 'healthy': True}, {'hostname': 'lab-secondary', 'healthy': True}]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Production callers must handle grouped failures deliberately rather than assume all tasks succeed.

````python
# Example 0821: TaskGroup owns concurrent tasks - branch
case = {'hostname': 'branch-edge-01'}

import asyncio

class Collector:
    async def read(self, name):
        await asyncio.sleep(0)
        return {"hostname": name, "healthy": True}

collector = Collector()
names = [case["hostname"], "lab-secondary"]
async with asyncio.TaskGroup() as group:
    tasks = [group.create_task(collector.read(name)) for name in names]
results = [task.result() for task in tasks]
assert [item["hostname"] for item in results] == names
assert all(item["healthy"] for item in results)
print(results)
````

## Pattern 084: Lock protects a shared threaded counter

**Core concept:** A lock protects the complete read-modify-write operation.

## Pattern 085: Bounded executor preserves per-item failures

**Core concept:** Future results must be inspected so worker errors reach the caller.

**Network-engineering purpose:** A fleet report can distinguish one failed simulated read from successful devices.

### Read the implementation in this order
1. Submit known targets to a small pool.
2. Associate each future with its target.
3. Read every result and catch only the expected TimeoutError.
4. Check success and failure counts and the failed hostname.

**Common mistake / boundary:** Fire-and-forget futures can hide failures; inspecting every result is part of completion.

**Practice after the ten worked scenarios:** Return a structured result dataclass instead of separate success and failure lists.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0841 - Bounded executor preserves per-item failures: branch scenario

**Scenario 01/10; pattern 085; Advanced.** This is a self-contained offline exercise. A fleet report can distinguish one failed simulated read from successful devices.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Submit known targets to a small pool. Check success and failure counts and the failed hostname. Follow the class-specific steps under Pattern 085 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
success: ['branch-edge-01:up'] failure: ['lab-failed']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Fire-and-forget futures can hide failures; inspecting every result is part of completion.

````python
# Example 0841: Bounded executor preserves per-item failures - branch
case = {'hostname': 'branch-edge-01'}

from concurrent.futures import ThreadPoolExecutor

class Poller:
    def read(self, hostname):
        if hostname == "lab-failed":
            raise TimeoutError("simulated read timeout")
        return hostname + ":up"

successes, failures = [], []
with ThreadPoolExecutor(max_workers=2) as executor:
    futures = {executor.submit(Poller().read, name): name
               for name in [case["hostname"], "lab-failed"]}
    for future, name in futures.items():
        try:
            successes.append(future.result())
        except TimeoutError:
            failures.append(name)
assert successes == [case["hostname"] + ":up"]
assert failures == ["lab-failed"]
print("success:", successes, "failure:", failures)
````

# Chapter 18: Telemetry, counters, and bounded state

**Level:** Advanced

**Learning goal:** Model units, counter deltas, explicit rollover assumptions, windows, and priority.

**Always name units.** Bytes, bits, seconds, milliseconds, percentages, and fractions are not
interchangeable. For a monotonically increasing byte counter, a basic rate is
`(new_bytes - old_bytes) * 8 / elapsed_seconds`. Reject nonpositive elapsed time. A negative
delta is not a negative traffic rate; it may mean reset, rollover, an out-of-order sample,
or a different interface identity.

**Rollover requires assumptions.** A modulo calculation works only when the counter width is
known, there has been no reset, and the observation interval permits at most the number of
wraps your model can distinguish. Two samples alone generally cannot prove those assumptions.
The rollover exercise states a one-wrap, no-reset lab assumption explicitly.

**Sliding windows** retain recent samples using bounded storage such as deque(maxlen=...).
An average smooths variation but can hide spikes and depends on sample intervals. These
examples use equally weighted samples; irregular telemetry may need time weighting.

**Health decisions** need clear boundaries: does a load equal to the limit pass? Is missing
data unknown or unhealthy? A label such as OK should be backed by explicit conditions.
Hysteresis can avoid flapping when values oscillate near a threshold.

**Priority queues** can schedule more important checks first. Include a stable sequence
tie-breaker if task objects themselves are not orderable. Priority ordering is not fairness,
rate limiting, or bounded total work; a production scheduler must address those separately.

**Pause and explain:** Compute units for a rate and explain why two samples cannot distinguish every reset from rollover.

## Pattern 086: Byte counters to bits per second

**Core concept:** A value object can make a unit conversion explicit and validated.

**Network-engineering purpose:** An interface rate should not silently treat bytes as bits or accept zero elapsed time.

### Read the implementation in this order
1. Validate positive elapsed seconds.
2. Reject a negative delta as reset or unordered input.
3. Convert byte delta to bits and divide by time.
4. Check a known arithmetic result.

**Common mistake / boundary:** This simple rate assumes both counters describe the same interface and no wrap occurred.

**Practice after the ten worked scenarios:** Return a structured unavailable result for resets rather than inventing a traffic rate.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0851 - Byte counters to bits per second: branch scenario

**Scenario 01/10; pattern 086; Advanced.** This is a self-contained offline exercise. An interface rate should not silently treat bytes as bits or accept zero elapsed time.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `counter` | `1000` |

**Logic to trace:** Validate positive elapsed seconds. Check a known arithmetic result. Follow the class-specific steps under Pattern 086 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
zero interval rejected
bps: 4000.0
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This simple rate assumes both counters describe the same interface and no wrap occurred.

````python
# Example 0851: Byte counters to bits per second - branch
case = {'counter': 1000}

class RateCalculator:
    def bits_per_second(self, previous, current, seconds):
        if seconds <= 0:
            raise ValueError("elapsed seconds must be positive")
        if current < previous:
            raise ValueError("counter reset or unordered sample")
        return (current - previous) * 8 / seconds

calculator = RateCalculator()
rate = calculator.bits_per_second(case["counter"], case["counter"] + 1000, 2)
assert rate == 4000
try:
    calculator.bits_per_second(10, 20, 0)
except ValueError:
    print("zero interval rejected")
else:
    raise AssertionError("zero interval accepted")
print("bps:", rate)
````

## Pattern 087: Counter rollover under an explicit assumption

**Core concept:** Modulo arithmetic can model one known-width wrap in a controlled lab.

**Network-engineering purpose:** A small synthetic counter makes rollover reasoning easy to inspect.

### Read the implementation in this order
1. Store the modulus from the bit width.
2. Validate both samples fit the width.
3. Compute the modular difference.
4. Test a deliberate wrap near the maximum.

**Common mistake / boundary:** A reset or multiple wraps cannot be identified from this result alone.

**Practice after the ten worked scenarios:** Add a maximum plausible delta and report ambiguity instead of claiming a reliable rate.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0861 - Counter rollover under an explicit assumption: branch scenario

**Scenario 01/10; pattern 087; Advanced.** This is a self-contained offline exercise. A small synthetic counter makes rollover reasoning easy to inspect.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `small_counter` | `0` |

**Logic to trace:** Store the modulus from the bit width. Test a deliberate wrap near the maximum. Follow the class-specific steps under Pattern 087 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
one-wrap/no-reset lab delta: 6
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A reset or multiple wraps cannot be identified from this result alone.

````python
# Example 0861: Counter rollover under an explicit assumption - branch
case = {'small_counter': 0}

class WrappingCounter:
    def __init__(self, bits):
        if type(bits) is not int or bits < 1:
            raise ValueError("invalid width")
        self.modulus = 1 << bits

    def delta_assuming_no_reset(self, previous, current):
        if not 0 <= previous < self.modulus or not 0 <= current < self.modulus:
            raise ValueError("sample outside width")
        return (current - previous) % self.modulus

counter = WrappingCounter(8)
previous = 250
current = case["small_counter"]
delta = counter.delta_assuming_no_reset(previous, current)
assert delta == 6 + current
print("one-wrap/no-reset lab delta:", delta)
````

## Pattern 088: Bounded sliding sample window

**Core concept:** A deque limits retained history while the object exposes a useful query.

**Network-engineering purpose:** A telemetry collector can keep a fixed number of recent observations without unbounded growth.

### Read the implementation in this order
1. Initialize deque with a positive maxlen.
2. Append more samples than the window size.
3. Compute a mean only when data exists.
4. Check only the most recent samples remain.

**Common mistake / boundary:** This is a sample-weighted mean, not a time-weighted mean for irregular polling intervals.

**Practice after the ten worked scenarios:** Expose minimum and maximum without returning the internal deque.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0871 - Bounded sliding sample window: branch scenario

**Scenario 01/10; pattern 088; Advanced.** This is a self-contained offline exercise. A telemetry collector can keep a fixed number of recent observations without unbounded growth.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `load` | `20` |

**Logic to trace:** Initialize deque with a positive maxlen. Check only the most recent samples remain. Follow the class-specific steps under Pattern 088 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
recent mean: 20.0
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This is a sample-weighted mean, not a time-weighted mean for irregular polling intervals.

````python
# Example 0871: Bounded sliding sample window - branch
case = {'load': 20}

from collections import deque

class Window:
    def __init__(self, size):
        if size < 1:
            raise ValueError("window must be positive")
        self.samples = deque(maxlen=size)

    def add(self, value):
        self.samples.append(value)

    def mean(self):
        if not self.samples:
            raise LookupError("no samples")
        return sum(self.samples) / len(self.samples)

window = Window(3)
for value in [case["load"], 10, 20, 30]:
    window.add(value)
assert list(window.samples) == [10, 20, 30]
assert window.mean() == 20
print("recent mean:", window.mean())
````

````python
# Example 0880: Bounded sliding sample window - backup
case = {'load': 10}

from collections import deque

class Window:
    def __init__(self, size):
        if size < 1:
            raise ValueError("window must be positive")
        self.samples = deque(maxlen=size)

    def add(self, value):
        self.samples.append(value)

    def mean(self):
        if not self.samples:
            raise LookupError("no samples")
        return sum(self.samples) / len(self.samples)

window = Window(3)
for value in [case["load"], 10, 20, 30]:
    window.add(value)
assert list(window.samples) == [10, 20, 30]
assert window.mean() == 20
print("recent mean:", window.mean())
````

## Pattern 089: Hysteresis avoids threshold flapping

**Core concept:** Different enter and clear thresholds make a stateful alarm deliberate.

**Network-engineering purpose:** A utilization alarm can remain active until load falls far enough below the alarm boundary.

### Read the implementation in this order

## Pattern 090: Priority queue with a stable tie-breaker

**Core concept:** An explicit sequence number avoids comparing non-orderable task objects.

**Network-engineering purpose:** Equal-priority health checks should keep a predictable submission order.

### Read the implementation in this order
1. Store priority, sequence, and payload in each heap item.
2. Increment sequence per submission.
3. Pop through the queue API.
4. Check priority order and stability for a tie.

**Common mistake / boundary:** A priority queue can starve low-priority work; fairness is a separate scheduling requirement.

**Practice after the ten worked scenarios:** Add an aging policy that eventually promotes waiting checks.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0891 - Priority queue with a stable tie-breaker: branch scenario

**Scenario 01/10; pattern 090; Advanced.** This is a self-contained offline exercise. Equal-priority health checks should keep a predictable submission order.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Store priority, sequence, and payload in each heap item. Check priority order and stability for a tie. Follow the class-specific steps under Pattern 090 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
['branch-edge-01', 'second-critical', 'routine']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** A priority queue can starve low-priority work; fairness is a separate scheduling requirement.

````python
# Example 0891: Priority queue with a stable tie-breaker - branch
case = {'hostname': 'branch-edge-01'}

import heapq
from itertools import count

class Scheduler:
    def __init__(self):
        self.queue, self.sequence = [], count()

    def add(self, priority, payload):
        heapq.heappush(self.queue, (priority, next(self.sequence), payload))

    def pop(self):
        return heapq.heappop(self.queue)[2]

scheduler = Scheduler()
scheduler.add(5, "routine")
scheduler.add(1, case["hostname"])
scheduler.add(1, "second-critical")
result = [scheduler.pop() for _ in range(3)]
assert result == [case["hostname"], "second-critical", "routine"]
print(result)
````

# Chapter 19: Topology and route-analysis objects

**Level:** Advanced

**Learning goal:** Encapsulate small graph algorithms and state the limits of a topology model.

**A topology graph** represents nodes and edges, not the whole operational network.
Decide whether links are directed, weighted, parallel, or associated with interfaces. The
simple undirected graph here stores neighbor sets and assumes one logical link between any
pair of different nodes. A missing path in this graph means the model has no path, not proof
that a real device is isolated.

**Breadth-first search (BFS)** finds a path with the fewest edges in an unweighted graph.
Use deque, record visited nodes, and avoid revisiting them. Sorted neighbors make tiny lab
examples deterministic. Weighted cost requires a different algorithm such as Dijkstra when
weights are nonnegative.

**Cycle detection** must respect graph type. For an undirected graph, the edge back to the
immediate parent is not itself a cycle. A directed graph needs a different treatment of the
active recursion path. A graph cycle does not automatically imply an Ethernet forwarding loop:
protocols and forwarding state determine actual behavior.

**Redundancy analysis** can remove one modeled link and check reachability again. It is only
a simplified single-failure exercise. Real resilience includes shared-risk groups, power,
control-plane convergence, capacity, fate-sharing, and failure detection.

**Address planning** benefits from value objects and overlap checks. Overlapping prefixes are
not always an error: nested routes are normal, while unintended overlap in the same address
allocation scope can be a problem. Scope and policy determine interpretation.

**Pause and explain:** Explain the difference between modeled connectivity and actual forwarding reachability.

## Pattern 091: Encapsulated undirected adjacency

**Core concept:** A graph object maintains reciprocal adjacency through one operation.

**Network-engineering purpose:** Topology callers should not have to remember to add both directions themselves.

### Read the implementation in this order
1. Initialize an adjacency mapping.
2. Add both directions when linking two different nodes.
3. Expose an immutable sorted neighbor view.
4. Check reciprocity and deduplication.

**Common mistake / boundary:** This simple graph intentionally omits parallel links and interface-level metadata.

**Practice after the ten worked scenarios:** Represent each link as an immutable value with two endpoint interfaces.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0901 - Encapsulated undirected adjacency: branch scenario

**Scenario 01/10; pattern 091; Advanced.** This is a self-contained offline exercise. Topology callers should not have to remember to add both directions themselves.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Initialize an adjacency mapping. Check reciprocity and deduplication. Follow the class-specific steps under Pattern 091 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
('branch-edge-01',)
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This simple graph intentionally omits parallel links and interface-level metadata.

````python
# Example 0901: Encapsulated undirected adjacency - branch
case = {'hostname': 'branch-edge-01'}

class Topology:
    def __init__(self):
        self.adjacency = {}

    def link(self, first, second):
        if first == second:
            raise ValueError("self links not supported")
        self.adjacency.setdefault(first, set()).add(second)
        self.adjacency.setdefault(second, set()).add(first)

    def neighbors(self, node):
        return tuple(sorted(self.adjacency[node]))

topology = Topology()
topology.link(case["hostname"], "core")
topology.link(case["hostname"], "core")
assert topology.neighbors(case["hostname"]) == ("core",)
assert topology.neighbors("core") == (case["hostname"],)
print(topology.neighbors("core"))
````

## Pattern 092: Breadth-first shortest hop path

**Core concept:** BFS explores increasing hop distance and remembers visited nodes.

**Network-engineering purpose:** An offline topology model can explain a shortest unweighted path.

### Read the implementation in this order
1. Put the source path in a deque.
2. Pop paths from the front.
3. Visit sorted unvisited neighbors.
4. Return the first destination path or raise when unreachable.

**Common mistake / boundary:** Fewest hops is not lowest metric, highest bandwidth, or a routing protocol decision.

**Practice after the ten worked scenarios:** Track predecessors instead of copying full paths for better scaling.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0911 - Breadth-first shortest hop path: branch scenario

**Scenario 01/10; pattern 092; Advanced.** This is a self-contained offline exercise. An offline topology model can explain a shortest unweighted path.

**Local inputs** (all are defined again in the code cell):

| Input | Value |

## Pattern 093: Undirected cycle detection

**Core concept:** A DFS distinguishes a parent edge from a genuine previously visited neighbor.

**Network-engineering purpose:** A lab topology checker can flag a structural cycle without claiming a real forwarding loop.

### Read the implementation in this order
1. Keep one visited set across graph components.
2. Track the parent in each recursive visit.
3. Skip only the immediate parent edge.
4. Compare a chain with a triangle.

**Common mistake / boundary:** This algorithm assumes a simple undirected graph; use an appropriate algorithm for directed or multigraph models.

**Practice after the ten worked scenarios:** Add disconnected components and verify a cycle is found outside the first component.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0921 - Undirected cycle detection: branch scenario

**Scenario 01/10; pattern 093; Advanced.** This is a self-contained offline exercise. A lab topology checker can flag a structural cycle without claiming a real forwarding loop.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Keep one visited set across graph components. Compare a chain with a triangle. Follow the class-specific steps under Pattern 093 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
chain: False triangle: True
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This algorithm assumes a simple undirected graph; use an appropriate algorithm for directed or multigraph models.

````python
# Example 0921: Undirected cycle detection - branch
case = {'hostname': 'branch-edge-01'}

class Topology:
    def __init__(self, adjacency):
        self.adjacency = adjacency

    def has_cycle(self):
        visited = set()

        def visit(node, parent):
            visited.add(node)
            for neighbor in sorted(self.adjacency[node]):
                if neighbor == parent:
                    continue
                if neighbor in visited or visit(neighbor, node):
                    return True
            return False

        return any(visit(node, None) for node in self.adjacency if node not in visited)

host = case["hostname"]
chain = Topology({host: {"b"}, "b": {host, "c"}, "c": {"b"}})
triangle = Topology({host: {"b", "c"}, "b": {host, "c"}, "c": {host, "b"}})
assert chain.has_cycle() is False and triangle.has_cycle() is True
print("chain:", chain.has_cycle(), "triangle:", triangle.has_cycle())
````

## Pattern 094: Single-link failure reachability

**Core concept:** Analyze a copied model so a what-if scenario does not mutate the baseline.

**Network-engineering purpose:** A redundant lab triangle can retain connectivity after one modeled link is removed.

### Read the implementation in this order
1. Copy all adjacency sets.
2. Remove both directions of the selected link.
3. Search from source to target.
4. Assert reachability and prove baseline links remain present.

**Common mistake / boundary:** This checks graph reachability only, not capacity, convergence time, or shared-risk failures.

**Practice after the ten worked scenarios:** Enumerate every link once and report which failures partition the topology.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0931 - Single-link failure reachability: branch scenario

**Scenario 01/10; pattern 094; Advanced.** This is a self-contained offline exercise. A redundant lab triangle can retain connectivity after one modeled link is removed.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Copy all adjacency sets. Assert reachability and prove baseline links remain present. Follow the class-specific steps under Pattern 094 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
modeled path survives one link failure
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This checks graph reachability only, not capacity, convergence time, or shared-risk failures.

````python
# Example 0931: Single-link failure reachability - branch
case = {'hostname': 'branch-edge-01'}

class Topology:
    def __init__(self, adjacency):
        self.adjacency = {node: set(neighbors) for node, neighbors in adjacency.items()}

    def reachable_without(self, source, target, failed):
        graph = {node: set(neighbors) for node, neighbors in self.adjacency.items()}
        first, second = failed
        graph[first].discard(second)
        graph[second].discard(first)
        pending, visited = [source], set()
        while pending:
            node = pending.pop()
            if node == target:
                return True
            if node not in visited:
                visited.add(node)
                pending.extend(graph[node] - visited)
        return False

host = case["hostname"]
topology = Topology({host: {"b", "c"}, "b": {host, "c"}, "c": {host, "b"}})
assert topology.reachable_without(host, "b", (host, "b"))
assert "b" in topology.adjacency[host]
print("modeled path survives one link failure")
````

**Remember:** This checks graph reachability only, not capacity, convergence time, or shared-risk failures.

````python
# Example 0940: Single-link failure reachability - backup
case = {'hostname': 'backup-edge-10'}

class Topology:
    def __init__(self, adjacency):
        self.adjacency = {node: set(neighbors) for node, neighbors in adjacency.items()}

    def reachable_without(self, source, target, failed):
        graph = {node: set(neighbors) for node, neighbors in self.adjacency.items()}
        first, second = failed
        graph[first].discard(second)
        graph[second].discard(first)
        pending, visited = [source], set()
        while pending:
            node = pending.pop()
            if node == target:
                return True
            if node not in visited:
                visited.add(node)
                pending.extend(graph[node] - visited)
        return False

host = case["hostname"]
topology = Topology({host: {"b", "c"}, "b": {host, "c"}, "c": {host, "b"}})
assert topology.reachable_without(host, "b", (host, "b"))
assert "b" in topology.adjacency[host]
print("modeled path survives one link failure")
````

## Pattern 096: Capstone: reviewed idempotent MTU deployment

**Core concept:** Combine an immutable intent, a plan, approval, reconciliation, and postcheck.

**Network-engineering purpose:** An engineer can trace exactly when the simulated state changes and why a second apply is a no-op.

### Read the implementation in this order
1. Represent desired MTU with a frozen dataclass.
2. Compute whether baseline differs.
3. Require approval before any mutation.
4. Verify the postcondition and check a repeated approved apply is unchanged.

**Common mistake / boundary:** This local state transition has no real-device atomicity, authorization, or durable audit guarantee.

**Practice after the ten worked scenarios:** Bind approval to baseline plus intent and reject drift between plan and apply.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0951 - Capstone: reviewed idempotent MTU deployment: branch scenario

**Scenario 01/10; pattern 096; Advanced.** This is a self-contained offline exercise. An engineer can trace exactly when the simulated state changes and why a second apply is a no-op.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |
| `approved` | `True` |

**Logic to trace:** Represent desired MTU with a frozen dataclass. Verify the postcondition and check a repeated approved apply is unchanged. Follow the class-specific steps under Pattern 096 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
observed: 1500 audit: ['unchanged', 'unchanged']
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** This local state transition has no real-device atomicity, authorization, or durable audit guarantee.

````python
# Example 0951: Capstone: reviewed idempotent MTU deployment - branch
case = {'mtu': 1500, 'approved': True}

from dataclasses import dataclass

@dataclass(frozen=True)
class Intent:
    mtu: int

    def __post_init__(self):
        if type(self.mtu) is not int or not 576 <= self.mtu <= 9216:
            raise ValueError("invalid MTU")

class Deployment:
    def __init__(self):
        self.observed, self.audit = 1500, []

    def plan(self, intent):
        return None if self.observed == intent.mtu else (self.observed, intent.mtu)

    def apply(self, intent, *, approved):
        if not approved:
            self.audit.append("denied")
            raise PermissionError("approval required")
        plan = self.plan(intent)
        if plan is not None:
            self.observed = intent.mtu
        if self.observed != intent.mtu:
            raise RuntimeError("postcheck failed")
        self.audit.append("changed" if plan else "unchanged")
        return plan is not None

deployment, intent = Deployment(), Intent(case["mtu"])
before = deployment.observed
try:
    deployment.apply(intent, approved=case["approved"])
except PermissionError:
    assert not case["approved"] and deployment.observed == before
else:
    assert deployment.observed == intent.mtu
    assert deployment.apply(intent, approved=True) is False
print("observed:", deployment.observed, "audit:", deployment.audit)
````

## Pattern 097: Capstone: fleet collection with structured outcomes

**Core concept:** A result object preserves target identity and explicit transport failures.

**Network-engineering purpose:** Fleet reporting should not hide one failed read behind an aggregate success count.

### Read the implementation in this order
1. Use a frozen outcome record.
2. Inject a replay transport.
3. Catch only the expected TimeoutError per target.
4. Assert the successful and failed result shapes separately.

**Common mistake / boundary:** Unexpected exceptions should still fail loudly rather than be mislabeled as ordinary device failure.

**Practice after the ten worked scenarios:** Add bounded concurrency while preserving a stable input-to-output mapping.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0961 - Capstone: fleet collection with structured outcomes: branch scenario

**Scenario 01/10; pattern 097; Advanced.** This is a self-contained offline exercise. Fleet reporting should not hide one failed read behind an aggregate success count.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `hostname` | `'branch-edge-01'` |

**Logic to trace:** Use a frozen outcome record. Assert the successful and failed result shapes separately. Follow the class-specific steps under Pattern 097 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
[Outcome(hostname='branch-edge-01', state='up', error=None), Outcome(hostname='lab-unavailable', state=None, error='read timeout')]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Unexpected exceptions should still fail loudly rather than be mislabeled as ordinary device failure.

````python
# Example 0961: Capstone: fleet collection with structured outcomes - branch
case = {'hostname': 'branch-edge-01'}

from dataclasses import dataclass

@dataclass(frozen=True)
class Outcome:
    hostname: str
    state: str | None
    error: str | None

class Replay:
    def read(self, hostname):
        if hostname == "lab-unavailable":
            raise TimeoutError("read timeout")
        return "up"

class FleetCollector:
    def __init__(self, transport):
        self.transport = transport

    def collect(self, names):
        results = []
        for name in names:
            try:
                state = self.transport.read(name)
            except TimeoutError as error:
                results.append(Outcome(name, None, str(error)))
            else:
                results.append(Outcome(name, state, None))
        return results

results = FleetCollector(Replay()).collect([case["hostname"], "lab-unavailable"])
assert results[0] == Outcome(case["hostname"], "up", None)
assert results[1] == Outcome("lab-unavailable", None, "read timeout")
assert sum(item.error is None for item in results) == 1
print(results)
````

## Pattern 098: Capstone: rate calculation and health policy

**Core concept:** Separate immutable samples, unit-aware calculation, and threshold evaluation.

**Network-engineering purpose:** A telemetry report can show both the measured bps value and the policy result.

### Read the implementation in this order
1. Represent each timestamped counter sample.
2. Reject non-increasing time and counter resets.
3. Calculate bits per second.
4. Evaluate against an injected policy limit and assert both data and classification.

**Common mistake / boundary:** These synthetic samples assume no wrap, no reset, and the same interface identity.

**Practice after the ten worked scenarios:** Add a result state for unavailable evidence without converting it to a healthy zero rate.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0971 - Capstone: rate calculation and health policy: branch scenario

**Scenario 01/10; pattern 098; Advanced.** This is a self-contained offline exercise. A telemetry report can show both the measured bps value and the policy result.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `counter` | `1000` |
| `traffic_bytes` | `0` |

**Logic to trace:** Represent each timestamped counter sample. Evaluate against an injected policy limit and assert both data and classification. Follow the class-specific steps under Pattern 098 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
bps: 0.0 health: ok
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** These synthetic samples assume no wrap, no reset, and the same interface identity.

````python
# Example 0971: Capstone: rate calculation and health policy - branch
case = {'counter': 1000, 'traffic_bytes': 0}

from dataclasses import dataclass

@dataclass(frozen=True)
class Sample:
    seconds: int
    byte_count: int

class Rate:
    def calculate(self, old, new):
        elapsed, delta = new.seconds - old.seconds, new.byte_count - old.byte_count
        if elapsed <= 0 or delta < 0:
            raise ValueError("incomparable samples")
        return delta * 8 / elapsed

class HealthPolicy:
    def __init__(self, limit):
        self.limit = limit

    def evaluate(self, bps):
        return "ok" if bps <= self.limit else "high"

old = Sample(0, case["counter"])
new = Sample(2, case["counter"] + case["traffic_bytes"])
bps = Rate().calculate(old, new)
result = HealthPolicy(4000).evaluate(bps)
assert bps == case["traffic_bytes"] * 4
assert result == ("ok" if case["traffic_bytes"] <= 1000 else "high")
print("bps:", bps, "health:", result)
````

## Pattern 099: Capstone: composable compliance evidence

**Core concept:** Individual checks return explicit evidence and an engine aggregates it.

**Network-engineering purpose:** An interface can fail one requirement without losing the evidence for checks that passed.

### Read the implementation in this order
1. Define a frozen check result.
2. Implement separate MTU and admin-state checks.
3. Inject checks into an engine.
4. Verify each result and the overall all-pass condition.

**Common mistake / boundary:** Missing evidence needs an explicit unknown/error state in production, not an automatic pass.

**Practice after the ten worked scenarios:** Extend the result with PASS/FAIL/UNKNOWN and define aggregation semantics.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0981 - Capstone: composable compliance evidence: branch scenario

**Scenario 01/10; pattern 099; Advanced.** This is a self-contained offline exercise. An interface can fail one requirement without losing the evidence for checks that passed.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `mtu` | `1500` |
| `healthy` | `True` |

**Logic to trace:** Define a frozen check result. Verify each result and the overall all-pass condition. Follow the class-specific steps under Pattern 099 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
[Evidence(check='mtu', passed=True, observed=1500), Evidence(check='admin', passed=True, observed=True)]
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** Missing evidence needs an explicit unknown/error state in production, not an automatic pass.

````python
# Example 0981: Capstone: composable compliance evidence - branch
case = {'mtu': 1500, 'healthy': True}

from dataclasses import dataclass

@dataclass(frozen=True)
class Evidence:
    check: str
    passed: bool
    observed: object

class MtuCheck:
    def run(self, record):
        return Evidence("mtu", record["mtu"] >= 1500, record["mtu"])

class AdminCheck:
    def run(self, record):
        return Evidence("admin", record["admin_up"] is True, record["admin_up"])

class Engine:
    def __init__(self, checks):
        self.checks = tuple(checks)

    def evaluate(self, record):
        return [check.run(record) for check in self.checks]

record = {"mtu": case["mtu"], "admin_up": case["healthy"]}
results = Engine([MtuCheck(), AdminCheck()]).evaluate(record)
assert results[0].passed == (case["mtu"] >= 1500)
assert results[1].passed == case["healthy"]
assert all(item.passed for item in results) == (case["mtu"] >= 1500 and case["healthy"])
print(results)
````

### Checkpoints
**Beginner:** Explain every line of a small Device class, build two independent instances, write an
instance method, and correct a mutable-default bug without looking at the answer.

**Intermediate:** Design a validated interface value, inject a fake transport, compare ABC and
Protocol, justify composition versus inheritance, serialize explicit fields, and write a negative test
that proves state remains unchanged after failure.

**Advanced:** Design an idempotent approved plan, report per-device failures honestly, bound async
work, reason about resource lifetime and cancellation, state counter-reset assumptions, and keep
domain models independent of vendor libraries.

### Five independent projects to try after the worked examples
1. Build a versioned offline inventory importer with duplicate detection, IPv4/IPv6 validation, and
   explicit error reports per source record.
2. Build a simulated interface-intent planner with immutable plans, baseline drift detection,
   approval binding, rollback outcomes, and idempotent reruns.
3. Build a bounded fleet collector with a fake clock, timeout injection, per-target results, and
   deterministic tests; no real devices are needed.
4. Build a telemetry evaluator that distinguishes valid rate, reset, wrap ambiguity, stale data,
   and missing data without inventing zero-valued success results.
5. Build a scoped topology/IPAM report with graph assumptions, link-failure analysis, VRF-aware
   overlap rules, and reproducible public JSON output.

These projects are extension ideas, not part of the counted 1,000 worked examples.
