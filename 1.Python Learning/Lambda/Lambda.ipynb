A **lambda in Python** is a small function written in one line. It’s useful when you need a simple function briefly, without defining it using `def`.

The syntax is:

```python
lambda arguments: expression
```

Python evaluates the expression and automatically returns its result.

For example:

```python
add = lambda a, b: a + b

print(add(3, 5))  # 8
```

This works like:

```python
def add(a, b):
    return a + b
```

When you call `add(3, 5)`, Python sets `a = 3` and `b = 5`, evaluates `a + b`, and returns `8`.

**Common use cases**

1. **Sorting by a specific value**

Sort a list of tuples by age:

```python
people = [("Alice", 25), ("Bob", 19), ("Charlie", 30)]

by_age = sorted(people, key=lambda person: person[1])

print(by_age)
# [('Bob', 19), ('Alice', 25), ('Charlie', 30)]
```

`sorted()` calls the lambda for each person. The lambda returns their age, which becomes the value used for sorting.

You can also sort dictionaries:

```python
products = [
    {"name": "Laptop", "price": 900},
    {"name": "Mouse", "price": 25},
    {"name": "Keyboard", "price": 70},
]

cheapest_first = sorted(products, key=lambda product: product["price"])
```

2. **Transforming values with `map()`**

```python
numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x ** 2, numbers))

print(squares)  # [1, 4, 9, 16]
```

`map()` applies the lambda to every item. A list comprehension is often clearer:

```python
squares = [x ** 2 for x in numbers]
```

3. **Selecting values with `filter()`**

```python
numbers = [1, 2, 3, 4, 5, 6]

evens = list(filter(lambda x: x % 2 == 0, numbers))

print(evens)  # [2, 4, 6]
```

`filter()` keeps each item for which the lambda returns `True`. Again, a comprehension is an alternative:

```python
evens = [x for x in numbers if x % 2 == 0]
```

4. **Passing a simple operation into another function**

```python
def apply_operation(value, operation):
    return operation(value)

print(apply_operation(10, lambda x: x * 2))    # 20
print(apply_operation(10, lambda x: x + 100))  # 110
```

Here, each lambda tells `apply_operation()` what to do with the value.

**Can a lambda contain a condition?** Yes, using a conditional expression:

```python
label = lambda age: "Adult" if age >= 18 else "Minor"

print(label(20))  # Adult
print(label(15))  # Minor
```

**When should you use `def` instead?**

A lambda is limited to **one expression**. Use `def` when the function needs several steps, statements such as `try` or `while`, documentation, or a meaningful reusable name.

As a practical rule, lambdas work especially well as short arguments such as `key=lambda item: item["price"]`. For named functions you’ll reuse, prefer `def`.

For a **network engineer**, Python lambdas are useful for sorting device inventories, filtering interface data, and processing monitoring results.

A lambda is a short function:

```python
lambda input: result
```

You’ll usually pass it to another function, such as `sorted()`.

**1. Sort interfaces by bandwidth usage**

Suppose your monitoring tool returns interface utilization:

```python
interfaces = [
    {"name": "Gi0/1", "utilization": 45},
    {"name": "Gi0/2", "utilization": 90},
    {"name": "Gi0/3", "utilization": 15},
]

busiest_first = sorted(
    interfaces,
    key=lambda interface: interface["utilization"],
    reverse=True,
)

for interface in busiest_first:
    print(interface["name"], interface["utilization"])
```

Output:

```text
Gi0/2 90
Gi0/1 45
Gi0/3 15
```

How it works:

- `sorted()` passes each interface dictionary to the lambda.
- The lambda returns its utilization.
- `sorted()` uses that value to order the interfaces.
- `reverse=True` puts the highest value first.

**2. Sort IP addresses correctly**

Sorting IP addresses as ordinary text can produce the wrong numerical order. Use Python’s built-in `ipaddress` module:

```python
import ipaddress

addresses = ["192.168.1.100", "192.168.1.2", "192.168.1.10"]

ordered = sorted(
    addresses,
    key=lambda ip: int(ipaddress.ip_address(ip)),
)

print(ordered)
```

Output:

```text
['192.168.1.2', '192.168.1.10', '192.168.1.100']
```

The lambda converts each IPv4 address into a number for comparison. The output still contains the original strings.

**3. Find interfaces that are down unexpectedly**

An interface that is administratively enabled but operationally down may need investigation:

```python
interfaces = [
    {"name": "Gi0/1", "admin": "up", "oper": "up"},
    {"name": "Gi0/2", "admin": "up", "oper": "down"},
    {"name": "Gi0/3", "admin": "down", "oper": "down"},
]

unexpected_down = list(filter(
    lambda interface: (
        interface["admin"] == "up"
        and interface["oper"] == "down"
    ),
    interfaces,
))

print(unexpected_down)
```

Output:

```text
[{'name': 'Gi0/2', 'admin': 'up', 'oper': 'down'}]
```

`filter()` keeps entries where the lambda returns `True`.

For everyday scripts, a list comprehension often reads more clearly:

```python
unexpected_down = [
    interface
    for interface in interfaces
    if interface["admin"] == "up" and interface["oper"] == "down"
]
```

**4. Sort routes by the most specific prefix**

Useful when inspecting overlapping routes:

```python
import ipaddress

routes = [
    "10.0.0.0/8",
    "10.1.0.0/16",
    "10.1.1.0/24",
]

most_specific_first = sorted(
    routes,
    key=lambda route: ipaddress.ip_network(route).prefixlen,
    reverse=True,
)

print(most_specific_first)
```

Output:

```text
['10.1.1.0/24', '10.1.0.0/16', '10.0.0.0/8']
```

This orders routes by prefix length. To find a route for a destination, you must also check whether that destination belongs to the network.

**When to use it in your work**

| Network task | Best choice |
|---|---|
| Sort devices by hostname, site, or uptime | `sorted()` with a lambda |
| Sort interfaces by errors or utilization | `sorted()` with a lambda |
| Filter alarms or interface states | Usually a list comprehension |
| Connect to switches and run commands | A regular `def` function |
| Push configuration with validation and error handling | A regular `def` function |

Start with this pattern—it’s likely the most useful one in your scripts:

```python
sorted(data, key=lambda item: item["field"])
```

Replace `"field"` with the value you want to sort by, such as `"hostname"`, `"errors"`, or `"utilization"`.

Lambda in Python (Anonymous Functions)

A lambda is a small, anonymous (nameless) function in Python that can be created in a single line.

Syntax
lambda arguments: expression


Equivalent to:

def function_name(arguments):
    return expression

Simple Example
Normal Function
def square(x):
    return x * x

print(square(5))


Output:

25

Lambda Function
square = lambda x: x * x

print(square(5))


Output:

25

How Lambda Works

When Python sees:

lambda x: x * x


It creates a function object that:

Accepts one argument (x)
Evaluates x * x
Returns the result automatically

No return keyword is needed.

Multiple Arguments
add = lambda a, b: a + b

print(add(10, 20))


Output:

30

Three Arguments
volume = lambda l, w, h: l * w * h

print(volume(2, 3, 4))


Output:

24

When NOT to Use Lambda

Bad example:

lambda x: (
    complicated_step1(x)
    + complicated_step2(x)
    - complicated_step3(x)
)


If the logic becomes complex:

✅ Use a normal function.

def calculate(x):
    step1 = complicated_step1(x)
    step2 = complicated_step2(x)
    step3 = complicated_step3(x)

    return step1 + step2 - step3


More readable and easier to debug.

Interview Questions
Q1: What is a lambda function?

A lambda is an anonymous one-line function used for short operations.

Q2: Can lambda contain multiple statements?

❌ No

lambda x: x + 1


Allowed.

lambda x:
    print(x)
    return x


Not allowed.

Q3: Why use lambda?
Shorter code
Useful with sorted(), map(), filter()
Avoid creating small temporary functions
Q4: Which is faster, lambda or def?

Practically, the performance difference is negligible. Choose based on readability, not speed.

 If you're writing Netmiko, Nornir, NAPALM, PyATS, REST API, Arista eAPI, Juniper PyEZ, or Azure DevOps automation, you'll most often see lambda used with:
