The **`tabulate`** library in Python is used to **format data (like lists, dictionaries, or tuples) into nicely structured tables** when printing output in the terminal or logs.

---

## 🔹 What `tabulate` does

It takes raw data and converts it into **readable tables** like:

* Grid tables
* Plain text tables
* Fancy formats (like GitHub markdown, HTML, etc.)

---

## 🔹 Why it is useful

Without `tabulate`, printing structured data looks messy:

```python
data = [["Yaswanth", 25], ["John", 30]]
for row in data:
    print(row)
```

Output:

```
['Yaswanth', 25]
['John', 30]
```

👉 Hard to read.

---

## 🔹 With `tabulate`

```python
from tabulate import tabulate

data = [["Yaswanth", 25], ["John", 30]]
headers = ["Name", "Age"]

print(tabulate(data, headers=headers))
```

Output:

```
Name       Age
--------  -----
Yaswanth    25
John        30
```

👉 Much cleaner and professional.

---

## 🔹 Common Uses (Very Important for You 👇 as Network Engineer)

### ✅ 1. Display Network Device Data

Example: Interface status

```python
interfaces = [
    ["Eth0", "up", "1Gbps"],
    ["Eth1", "down", "1Gbps"]
]

print(tabulate(interfaces, headers=["Interface", "Status", "Speed"]))
```

---

### ✅ 2. Show API / JSON Data Clearly

When you pull data from:

* REST APIs (Cisco, Arista, AWS)
* Network automation scripts

```python
devices = [
    {"name": "R1", "ip": "10.1.1.1"},
    {"name": "R2", "ip": "10.1.1.2"}
]

print(tabulate(devices, headers="keys"))
```

---

### ✅ 3. Logging / Debugging

Instead of messy logs, you can print structured tables:

* BGP neighbors
* Route tables
* Test results

---

### ✅ 4. Report Generation (CLI tools)

If you're building tools like:

* Network health checker
* Automation scripts

👉 `tabulate` makes CLI output look like a real tool.

---

### ✅ 5. Different Table Styles

You can change styles:

```python
print(tabulate(data, headers=headers, tablefmt="grid"))
```

Formats include:

* `"plain"`
* `"grid"`
* `"fancy_grid"`
* `"github"`
* `"html"`

````python
import pandas as pd
from tabulate import tabulate

df = pd.DataFrame({
    "Device": ["R1", "R2"],
    "Status": ["UP", "DOWN"]
})

print(tabulate(df, headers="keys", tablefmt="grid"))
````

````python
from tabulate import tabulate

data = [["Alice", 30, "Engineer"],
        ["Bob", 25, "Designer"],
        ["Charlie", 35, "Manager"]]

headers = ["Name", "Age", "Role"]

print(tabulate(data, headers=headers))
````

````python
tabulate(data, headers=headers, tablefmt="grid")
tabulate(data, headers=headers, tablefmt="pipe")       # Markdown
tabulate(data, headers=headers, tablefmt="html")
tabulate(data, headers=headers, tablefmt="pretty")
tabulate(data, headers=headers, tablefmt="rounded_grid")
````

````python
import pandas as pd
df = pd.DataFrame(data)
print(tabulate(df, headers="keys", tablefmt="grid", showindex=False))
````
