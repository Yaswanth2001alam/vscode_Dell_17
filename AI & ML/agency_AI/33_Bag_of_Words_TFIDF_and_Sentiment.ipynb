## Concept notes and mental model

Bag-of-words counts token occurrences in a fixed vocabulary. TF-IDF reweights terms using local
frequency and corpus document frequency so widespread words can contribute less than distinctive
ones. scikit-learn's default smoothing and normalization specify one particular variant.

Sparse matrices store mostly nonzero entries rather than a dense document-by-vocabulary table.
Fit the vocabulary and IDF on training text only. A unigram bag loses word order; bigrams can
capture short phrases such as 'not good', though they increase feature space.

Sentiment is a labeled classification task, not a universal emotional truth. Domain, sarcasm,
negation, and language shift can defeat a model trained on a different corpus. Here the tiny
operational message labels are synthetic positive/negative examples for mechanics, not a
representative evaluation.

## Mathematics or data contract

One smoothed IDF variant is log((1+n_documents)/(1+document_frequency))+1; TF-IDF may then normalize each row.

## Worked example 1: Build a sparse TF-IDF classification pipeline

**Purpose:** Train text representation and classifier together.

### Syntax and behavior to notice
- `fit(...)` learns model or preprocessing state from the supplied training data.
- `transform(...)` applies already-fitted preprocessing; it does not refit it.
- `predict(...)` returns the estimator's chosen output, not a guarantee of correctness.
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
toy sentiment labels: [1, 0]
sparse feature shape: (6, 24)
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

texts = ["good stable service", "excellent good connection", "happy reliable service",
         "bad unstable service", "terrible bad connection", "unhappy broken service"]
labels = [1, 1, 1, 0, 0, 0]
model = make_pipeline(TfidfVectorizer(ngram_range=(1, 2)), LogisticRegression(C=10))
model.fit(texts, labels)
predicted = model.predict(["good reliable connection", "bad broken connection"])
assert predicted.tolist() == [1, 0]
vectorizer = model.steps[0][1]
matrix = vectorizer.transform(texts)
assert matrix.shape[0] == 6 and "never-seen-token" not in vectorizer.vocabulary_
print("toy sentiment labels:", predicted.tolist())
print("sparse feature shape:", matrix.shape)
````
