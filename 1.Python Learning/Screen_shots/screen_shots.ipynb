Use an explicit wait for the page to load, then add a delay between the two screenshots:

```python
import time
from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait

driver = webdriver.Chrome()

try:
    driver.get("https://example.com")

    # Wait up to 20 seconds for the page to finish loading
    WebDriverWait(driver, 20).until(
        lambda d: d.execute_script("return document.readyState") == "complete"
    )

    # Optional extra time for images/animations to render
    time.sleep(2)

    driver.save_screenshot("example_1.png")

    # Delay before the second screenshot
    time.sleep(5)

    driver.save_screenshot("example_2.png")

finally:
    driver.quit()
```

Change `time.sleep(5)` to your desired delay in seconds. Note that the correct method name is `save_screenshot`, without a backslash.

You can use either time.sleep() (simple) or Selenium's WebDriverWait (better). Since you want the page to fully load before taking screenshots, I recommend WebDriverWait.

Option 1: Simple delay using time.sleep()
from selenium import webdriver
import time

driver = webdriver.Chrome()

driver.get("https://example.com")

# Wait 10 seconds for page to load
time.sleep(10)

# Screenshot 1
driver.save_screenshot("screenshot1.png")
print("Screenshot 1 captured")

# Wait another 5 seconds
time.sleep(5)

# Screenshot 2
driver.save_screenshot("screenshot2.png")
print("Screenshot 2 captured")

driver.quit()

Option 2: Better approach using WebDriverWait

This waits until the page is fully loaded instead of using a fixed timer.

from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
import time

driver = webdriver.Chrome()

driver.get("https://example.com")

# Wait until page is fully loaded
WebDriverWait(driver, 30).until(
    lambda d: d.execute_script(
        "return document.readyState"
    ) == "complete"
)

# Screenshot 1
driver.save_screenshot("screenshot1.png")
print("Screenshot 1 captured")

# Wait 5 seconds
time.sleep(5)

# Screenshot 2
driver.save_screenshot("screenshot2.png")
print("Screenshot 2 captured")

driver.quit()

Take screenshots every N seconds

For example, take 2 screenshots with a 10-second gap:

from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
import time

driver = webdriver.Chrome()

driver.get("https://example.com")

# Wait for page load
WebDriverWait(driver, 30).until(
    lambda d: d.execute_script(
        "return document.readyState"
    ) == "complete"
)

for i in range(2):
    filename = f"screenshot_{i+1}.png"
    driver.save_screenshot(filename)
    print(f"Saved {filename}")

    if i < 1:  # don't wait after last screenshot
        time.sleep(10)

driver.quit()

Add timestamps to screenshot names

Useful if you're monitoring a website:

from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
from datetime import datetime
import time

driver = webdriver.Chrome()

driver.get("https://example.com")

WebDriverWait(driver, 30).until(
    lambda d: d.execute_script(
        "return document.readyState"
    ) == "complete"
)

for _ in range(2):
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    driver.save_screenshot(f"screenshot_{timestamp}.png")
    print(f"Captured screenshot_{timestamp}.png")

    time.sleep(10)

driver.quit()


This will create files such as:

screenshot_20260901_224500.png
screenshot_20260901_224510.png


which makes it easier to track when each screenshot was taken.
