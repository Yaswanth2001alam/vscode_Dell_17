## Concept notes and mental model

Vectors represent features or parameters; dot products combine corresponding components.
A probability distribution has nonnegative entries summing to one. Expected value is a
probability-weighted average, not a promise about an individual trial.

A loss measures mismatch between predictions and targets. Squared error penalizes large
numeric errors strongly; cross-entropy penalizes assigning low probability to the true class.
Optimization adjusts parameters to reduce training loss. Generalization still requires a
separate evaluation dataset.

A derivative describes local sensitivity. Gradient descent moves against the gradient by a
step controlled by the learning rate. A large rate can diverge; a tiny rate can learn slowly.
Numerical stability matters: subtract a maximum before exponentiating logits in softmax,
and avoid log(0). Analytical gradients can be checked with finite differences on tiny problems.

## Worked example 1: Optimize a scalar and verify a probability vector

**Purpose:** Trace parameter updates with independently checkable arithmetic.

### Read the code in this order
1. Start away from the known optimum.

3. Compare initial and final loss.

### Syntax and behavior to notice
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
optimized weight: 3.0 loss decreased: True
probabilities: [0.09, 0.245, 0.665]
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import numpy as np

weight, rate = 0.0, 0.2
initial = (weight - 3) ** 2
for _ in range(40):
    gradient = 2 * (weight - 3)
    weight -= rate * gradient
final = (weight - 3) ** 2
assert final < initial and abs(weight - 3) < 1e-6
logits = np.array([1001., 1002., 1003.])
exp_values = np.exp(logits - logits.max())
probabilities = exp_values / exp_values.sum()
assert np.isclose(probabilities.sum(), 1) and (probabilities > 0).all()
print("optimized weight:", round(weight, 6), "loss decreased:", final < initial)
print("probabilities:", np.round(probabilities, 3).tolist())
````
