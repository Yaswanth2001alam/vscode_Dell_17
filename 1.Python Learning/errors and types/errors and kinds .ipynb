As a **Network Engineer**, Python errors are not just programming concepts—they are **real troubleshooting scenarios**. Most network automation failures happen because of:

* Wrong device data
* Missing keys in API/JSON responses
* Bad IP addresses
* SSH connection failures
* File issues
* Parsing CLI output
* Incorrect variable types

Understanding Python exceptions will make debugging automation scripts much easier.

# Handling Exceptions Properly

Always wrap network operations with `try`/`except`:

```python
try:
    device = {
        "hostname": "R1",
        "ip": "10.1.1.1"
    }

    print(device["vendor"])

except KeyError as e:
    print(f"Missing key: {e}")

except ValueError:
    print("Invalid value supplied")

except FileNotFoundError:
    print("Inventory file missing")

except ConnectionError:
    print("Device unreachable")

except Exception as e:
    print("Unexpected Error:", e)
```

# Errors You'll Encounter Most Often as a Network Engineer

| Error                   | How often? | Real-world example                                                       |
| ----------------------- | ---------- | ------------------------------------------------------------------------ |
| **KeyError**            | ⭐⭐⭐⭐⭐      | Missing field in JSON/API response (`serial`, `vendor`, `model`)         |
| **ValueError**          | ⭐⭐⭐⭐⭐      | Invalid VLAN, ASN, IP address, or user input                             |
| **ConnectionError**     | ⭐⭐⭐⭐⭐      | Device unreachable via SSH, RESTCONF, or NETCONF                         |
| **TimeoutError**        | ⭐⭐⭐⭐☆      | Slow device or API response timeout                                      |
| **FileNotFoundError**   | ⭐⭐⭐⭐☆      | Missing `inventory.yaml`, `devices.csv`, or configuration file           |
| **ModuleNotFoundError** | ⭐⭐⭐⭐☆      | Required library such as `netmiko`, `scrapli`, or `napalm` not installed |
| **IndexError**          | ⭐⭐⭐☆☆      | No neighbors, routes, or interfaces returned in a list                   |
| **AttributeError**      | ⭐⭐⭐☆☆      | Calling a string method on `None` or another unsupported object          |
| **TypeError**           | ⭐⭐⭐☆☆      | Mixing strings and integers (for example, port numbers or counters)      |
| **AssertionError**      | ⭐⭐⭐☆☆      | Automated tests (pytest) verifying protocol or interface state           |

---

## Practical advice for network automation

When writing Python for Cisco, Juniper, Arista, or cloud networking:

* Validate user input before using it (`ValueError` prevention).
* Use `dict.get()` instead of direct indexing whenever API fields may be optional (`KeyError` prevention).
* Check whether a list contains data before accessing an index (`IndexError` prevention).
* Wrap SSH, RESTCONF, NETCONF, and API calls in `try`/`except` blocks to handle `ConnectionError` and `TimeoutError`.
* Log the original exception (`except Exception as e`) so troubleshooting is easier without hiding the root cause.

These practices make your automation scripts much more reliable in production environments where devices, APIs, and network conditions are often unpredictable.
