## Concept notes and mental model

A prompt is input used to condition a model's output. Chat APIs often separate system,
developer, user, assistant, and tool messages, but supported roles and priority rules vary by
service. Roles are not a substitute for application authorization. Retrieved documents and
tool results are data that may contain untrusted instructions.

Good task specifications state the goal, constraints, output schema, evidence policy, and
failure behavior. Examples can clarify a format but can also bias answers. Test with both
normal and adversarial inputs rather than assuming a carefully worded prompt is sufficient.

A context window is finite working input, not a permanently updated knowledge base. Model
weights reflect training/post-training; current facts may require retrieval or tools. Reasoning
is a capability to perform multi-step computation or planning; fluent explanations are not
guaranteed faithful traces of internal computation or proof that a final answer is correct.

Open-weight means parameters are available under specified terms; it does not automatically
mean unrestricted open-source rights, complete training-data disclosure, or cheap inference.
Hosted services handle infrastructure but introduce changing contracts, cost, retention,
regional availability, and privacy requirements. Compare license, deployment needs, and your
own evaluations rather than choosing solely by a leaderboard.

## Worked example 1: Build an evidence-oriented request with explicit boundaries

**Purpose:** Separate instructions from untrusted source text.

### Read the code in this order
1. Define a system task contract.
2. Keep user intent and retrieved evidence in separate fields.
3. Set a finite output budget.
4. Verify structure without claiming the prompt enforces security by itself.

### Syntax and behavior to notice
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
request constructed; no model called
evidence IDs: ['lab-policy-1']
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
import json

request = {
    "messages": [
        {"role": "system", "content": "Answer from supplied evidence. If unsupported, say unknown. Return a concise answer and source IDs."},
        {"role": "user", "content": "What MTU does the lab policy require?"},
    ],
    "evidence": [{"id": "lab-policy-1", "text": "The simulated lab interface MTU is 1500 bytes."}],
    "max_output_tokens": 100,
}
assert request["messages"][0]["role"] == "system"
assert request["evidence"][0]["id"] == "lab-policy-1"
assert 0 < request["max_output_tokens"] <= 100
encoded = json.dumps(request, sort_keys=True)
assert json.loads(encoded) == request
print("request constructed; no model called")
print("evidence IDs:", [item["id"] for item in request["evidence"]])
````
