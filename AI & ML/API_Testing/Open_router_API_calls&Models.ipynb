````python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI

# -----------------------------
# Load API key
# -----------------------------

load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")

if not api_key:
    raise ValueError("OPENROUTER_API_KEY not found")


# -----------------------------
# Get models
# -----------------------------

url = "https://openrouter.ai/api/v1/models"

headers = {
    "Authorization": f"Bearer {api_key}"
}

response = requests.get(
    url,
    headers=headers,
    timeout=15
)

response.raise_for_status()

models = response.json()["data"]


# -----------------------------
# Search models
# -----------------------------

search = input(
    "Search model (openai, gemini, claude, llama, qwen, etc.): "
).lower()

matches = [
    model
    for model in models
    if search in model["id"].lower()
]

print(f"\nFound {len(matches)} models:\n")

for number, model in enumerate(matches, start=1):

    print(
        number,
        model["id"],
        "| Context:",
        model.get("context_length")
    )


# -----------------------------
# Select model
# -----------------------------

choice = int(
    input("\nSelect model number: ")
)

selected_model = matches[choice - 1]["id"]

print("\nSelected model:")
print(selected_model)


# -----------------------------
# Create OpenRouter client
# -----------------------------

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key
)


# -----------------------------
# Ask question
# -----------------------------

question = input(
    "\nAsk your networking question: "
)

response = client.chat.completions.create(
    model=selected_model,
    messages=[
        {
            "role": "system",
            "content": """
You are a senior network engineer specializing in
Cisco, Juniper, Arista, Azure Networking,
BGP, OSPF, ISIS, BFD, LACP, TCP/IP,
packet captures and network automation.
"""
        },
        {
            "role": "user",
            "content": question
        }
    ]
)


# -----------------------------
# Print answer
# -----------------------------

print("\n" + "=" * 70)
print("MODEL:", selected_model)
print("=" * 70)

print(response.choices[0].message.content)
````

# Project 15 — Automatically select a model

You already built a model-list API call.

You can make a simple interactive selector:

```python
import requests
import os

api_key = os.getenv("OPENROUTER_API_KEY")

headers = {
    "Authorization": f"Bearer {api_key}"
}

response = requests.get(
    "https://openrouter.ai/api/v1/models",
    headers=headers,
    timeout=20
)

response.raise_for_status()

models = response.json()["data"]

keyword = input(
    "Search models: "
).lower()

matches = [
    model
    for model in models
    if keyword in model["id"].lower()
]

for number, model in enumerate(
    matches,
    start=1
):
    print(
        number,
        model["id"]
    )

choice = int(
    input("\nSelect model: ")
)

MODEL = matches[choice - 1]["id"]

print("\nUsing:")
print(MODEL)
```

Then all of your projects can use:

```python
model=MODEL
```

This is much better than hardcoding:

```python
model="gemini-..."
```

everywhere.
