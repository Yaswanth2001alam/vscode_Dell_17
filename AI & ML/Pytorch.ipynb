### Code:

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Data
X = torch.randn(100, 3)
y = torch.randn(100, 1)

# Model
model = nn.Linear(3, 1)

loss_fn = nn.MSELoss()

    loss = loss_fn(y_pred, y)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
