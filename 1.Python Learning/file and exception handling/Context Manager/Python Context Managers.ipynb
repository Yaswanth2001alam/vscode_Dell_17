In the below code, the coe first open the books.txt file


After opening it, the titles is an variable that will read the context from books.txt file

Upon reading it, netx will sort the content from that. 

Now the sorted content will be written or updated to the new file called as context.txt

```python
with open("books.txt", "r") as input_file:
    with open("catalog.txt", "w") as output_file:
        titles = input_file.read().split("\n")
        titles.sort()
        output_file.write("\n".join(titles))
```

```python
with open("cities.txt" , "w") as file:
    file.write("Tokyo\n")
    file.write("Paris\n")
    file.write("New York\n")
    file.write("Sydney\n")

with open("cities.txt" , "r") as file:
    cities = file.readlines()

print(len(cities) , "Cities Saved")
```

```python
cities = ["Tokyo", "Paris", "New York", "Sydney"]

with open("cities.txt", "w") as file:
    for city in cities:
        file.write(city + "\n")

with open("cities.txt", "r") as file:
    lines = file.read().strip().split("\n")
    print(f"{len(lines)} cities saved.")
```

```python
with open("temperatures_c.txt", "r") as infile, open("temperatures_f.txt", "w") as outfile:
    for line in infile:
        celsius = float(line.strip())
        fahrenheit = celsius * 9 / 5 + 32
        outfile.write(str(fahrenheit) + "\n")
```
