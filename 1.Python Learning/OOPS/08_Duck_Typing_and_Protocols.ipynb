## Concept notes and mental model

**Level:** Intermediate

**Learning goal:** Express contracts without confusing type hints with runtime validation.

**Abstraction** presents the operations a caller needs while hiding irrelevant implementation.
A collector should need `send(command) -> str`, not every detail of an SSH or HTTP client.
Abstract base classes (ABCs) make a nominal contract explicit. A class with unimplemented
abstract methods cannot normally be instantiated. Concrete subclasses still need sensible
semantics: the ABC machinery cannot prove their output is correct.

**Duck typing** uses behavior directly: if an object supports the required operation, it can
be used. A `typing.Protocol` describes that shape to a static type checker without requiring
explicit inheritance. `@runtime_checkable` permits limited runtime presence checks, but does
not validate signatures, return values, or semantic correctness. The defining protocol in
these examples is explicitly decorated rather than relying on inherited runtime checkability.

**Type hints** document intent and help editors and static analyzers. Python does not generally
enforce them during an ordinary call. Validate untrusted input separately. `list[Device]`
describes a collection of devices; `Device | None` expresses an optional value. A type variable
can preserve the relationship between a container's input and output type. This course uses
the traditional `TypeVar` and `Generic` form for readability across supported Python versions.

**Dependency inversion** puts the service's required abstraction ahead of a particular adapter.
Inject a fake during a test and a reviewed real implementation at the application boundary.
Do not add an ABC, Protocol, and registry for every two-line helper: choose the smallest
contract that actually helps callers and tests.

**Pause and explain:** Compare an ABC with a Protocol and name one property that neither can verify automatically.

## Pattern 027: Duck-typed collaborators

**Core concept:** Compatible behavior can be used without a common explicit base.

**Network-engineering purpose:** A fake probe and a replayed probe can both feed a health service.

### Read the implementation in this order
1. Give unrelated classes the same check method.
2. Write a client that calls that method.
3. Pass both classes through the client.
4. Assert their distinct results are preserved.

**Common mistake / boundary:** Missing behavior will fail at runtime; annotations and tests can catch mistakes earlier.

**Practice after the ten worked scenarios:** Create a bad collaborator and identify the exact failure boundary.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

## Pattern 028: Structural contract with Protocol

**Core concept:** A Protocol describes the operations a collaborator must expose.

**Network-engineering purpose:** A service can type against a transport shape rather than a specific library.

### Read the implementation in this order
1. Declare the send signature on a Protocol.
2. Decorate that Protocol for a limited runtime check.
3. Implement send without inheriting from the protocol.
4. Use the object through an annotated function.

**Common mistake / boundary:** isinstance against a runtime protocol checks presence, not argument types or result correctness.

**Practice after the ten worked scenarios:** Run a static type checker on a variant whose send method returns an integer.

The following ten examples intentionally reuse this implementation with explicit local scenario data. The displayed output comes from running each full code cell; assertions also check behavior, including expected rejections where present.

### Example 0271 - Structural contract with Protocol: branch scenario

**Scenario 01/10; pattern 028; Intermediate.** This is a self-contained offline exercise. A service can type against a transport shape rather than a specific library.

**Local inputs** (all are defined again in the code cell):

| Input | Value |
| --- | --- |
| `interface` | `'Ethernet1/1'` |

**Logic to trace:** Declare the send signature on a Protocol. Use the object through an annotated function. Follow the class-specific steps under Pattern 028 for the intermediate operations. Predict the result from these inputs before running the cell.

**Expected printed output:**
```text
replay:Ethernet1/1
```

**What verifies it:** the assertions in the next cell check returned values, state, interactions, or explicit expected exceptions, not just whether the code prints. An assertion failure indicates an unexpected result. If you intentionally change input or behavior, update your prediction and relevant tests rather than deleting the checks.

**Remember:** isinstance against a runtime protocol checks presence, not argument types or result correctness.

````python
# Example 0271: Structural contract with Protocol - branch
case = {'interface': 'Ethernet1/1'}

from typing import Protocol, runtime_checkable

@runtime_checkable
class Transport(Protocol):
    def send(self, command: str) -> str: ...

class ReplayTransport:
    def send(self, command: str) -> str:
        return "replay:" + command

def collect(transport: Transport, command: str) -> str:
    return transport.send(command)

transport = ReplayTransport()
assert isinstance(transport, Transport)
result = collect(transport, case["interface"])
assert result == "replay:" + case["interface"]
print(result)
````
