```python
import psutil

processes = []

for process in psutil.process_iter(
    ['pid', 'name', 'memory_info']
):
    try:
        memory_gb = process.info['memory_info'].rss / (1024**3)

        processes.append(
            (
                process.info['pid'],
                process.info['name'],
                memory_gb
            )
        )

    except (psutil.NoSuchProcess, psutil.AccessDenied):
        pass


processes.sort(key=lambda x: x[2], reverse=True)

print(f"{'PID':<10} {'PROCESS':<30} {'MEMORY GB':>10}")
print("-" * 55)

for pid, name, memory in processes[:15]:
    print(f"{pid:<10} {name:<30} {memory:>10.2f}")
```

```python
import psutil
import time
import os

INTERVAL = 2

def gb(value):
    return value / (1024 ** 3)

def mb(value):
    return value / (1024 ** 2)

# Prime CPU counters
for p in psutil.process_iter():
    try:
        p.cpu_percent(None)
    except:
        pass

# Initial disk I/O counters
old_io = {}

for p in psutil.process_iter(['pid']):
    try:
        io = p.io_counters()
        old_io[p.pid] = io.read_bytes + io.write_bytes
    except:
        pass

# Initial network counters
old_net = psutil.net_io_counters()

while True:

    time.sleep(INTERVAL)

    os.system("cls" if os.name == "nt" else "clear")

    # ------------------------------------------------
    # SYSTEM SUMMARY
    # ------------------------------------------------

    cpu_total = psutil.cpu_percent(None)

    memory = psutil.virtual_memory()

    disk = psutil.disk_usage("C:\\")

    new_net = psutil.net_io_counters()

    sent_mbps = (
        (new_net.bytes_sent - old_net.bytes_sent) * 8
        / INTERVAL
        / 1_000_000
    )

    recv_mbps = (
        (new_net.bytes_recv - old_net.bytes_recv) * 8
        / INTERVAL
        / 1_000_000
    )

    old_net = new_net

    print("=" * 90)
    print("                    WINDOWS SYSTEM MONITOR")
    print("=" * 90)

    print(
        f"CPU: {cpu_total:5.1f}%   "
        f"Memory: {memory.percent:5.1f}%   "
        f"Disk C: {disk.percent:5.1f}%   "
        f"Network TX: {sent_mbps:7.2f} Mbps   "
        f"RX: {recv_mbps:7.2f} Mbps"
    )

    print("=" * 90)

    # ------------------------------------------------
    # PROCESS INFORMATION
    # ------------------------------------------------

    processes = []

    for p in psutil.process_iter(
        ['pid', 'name', 'memory_info']
    ):

        try:

            cpu = p.cpu_percent(None)

            memory_mb = (
                p.info['memory_info'].rss / (1024 ** 2)
            )

            try:
                io = p.io_counters()

                current_io = (
                    io.read_bytes +
                    io.write_bytes
                )

                previous_io = old_io.get(
                    p.pid,
                    current_io
                )

                disk_rate = (
                    current_io -
                    previous_io
                ) / INTERVAL / (1024 ** 2)

                old_io[p.pid] = current_io

            except:
                disk_rate = 0

            processes.append(
                {
                    "pid": p.pid,
                    "name": p.info['name'],
                    "cpu": cpu,
                    "memory": memory_mb,
                    "disk": disk_rate
                }
            )

        except (
            psutil.NoSuchProcess,
            psutil.AccessDenied
        ):
            continue

    # Sort by CPU usage
    processes.sort(
        key=lambda x: x["cpu"],
        reverse=True
    )

    print(
        f"{'PID':<8}"
        f"{'PROCESS':<35}"
        f"{'CPU %':>10}"
        f"{'MEMORY MB':>15}"
        f"{'DISK MB/s':>15}"
    )

    print("-" * 90)

    for p in processes[:20]:

        print(
            f"{p['pid']:<8}"
            f"{p['name'][:32]:<35}"
            f"{p['cpu']:>10.1f}"
            f"{p['memory']:>15.1f}"
            f"{p['disk']:>15.2f}"
```
````python
        )
````

```text
    |
    +--- TCP connection
    |
    +--- Application protocol
```

You can measure DNS and TCP separately:

```python
import socket
import time

hostname = "google.com"
port = 443

print("===== DNS =====")

start = time.perf_counter()

ip = socket.gethostbyname(hostname)

dns_time = (time.perf_counter() - start) * 1000

print("Hostname :", hostname)
print("IP       :", ip)
print(f"DNS Time : {dns_time:.2f} ms")


print("\n===== TCP =====")

start = time.perf_counter()

sock = socket.create_connection(
    (ip, port),
    timeout=3
)

tcp_time = (time.perf_counter() - start) * 1000

print("Port     :", port)
print("Status   : PASS")
print(f"TCP Time : {tcp_time:.2f} ms")

sock.close()
```

This helps distinguish:

```text
DNS slow?
      ↓
TCP connection slow?
      ↓
Application slow?
```

## 8. Test the actual HTTPS connection

TCP/443 succeeding only proves that the TCP endpoint accepted a connection. You can go one level higher and test TLS:

```python
import socket
import ssl
import time

hostname = "google.com"
port = 443

context = ssl.create_default_context()

start = time.perf_counter()

with socket.create_connection(
    (hostname, port),
    timeout=5
) as sock:

    tcp_time = (
        time.perf_counter() - start
    ) * 1000

    tls_start = time.perf_counter()

    with context.wrap_socket(
        sock,
        server_hostname=hostname
    ) as secure_socket:

        tls_time = (
            time.perf_counter() - tls_start
        ) * 1000

        print("TCP Connection : PASS")
        print(f"TCP Time       : {tcp_time:.2f} ms")

        print("TLS Handshake  : PASS")
        print(f"TLS Time       : {tls_time:.2f} ms")

        print(
            "TLS Version    :",
            secure_socket.version()
        )
```

Example:

```text
TCP Connection : PASS
TCP Time       : 18.31 ms
TLS Handshake  : PASS
TLS Time       : 32.48 ms
TLS Version    : TLSv1.3
```

This is a much better real-world troubleshooting sequence:

```text
PC
 │
 ├── DNS resolution
 │
 ├── IP reachability
 │
 ├── TCP SYN
 │      ↓
 │   SYN-ACK
 │      ↓
 │     ACK
 │
 ├── TLS handshake
 │
 └── HTTPS request
```

For network-engineering practice, I'd work through **ICMP → DNS → TCP connection → TCP latency → TLS handshake → HTTP response**. That mirrors how you would isolate whether a real application connectivity problem is at the network, transport, TLS, or application layer.
