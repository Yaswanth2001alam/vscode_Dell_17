````python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

# Load dataset (handwritten digits)
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize data (0–255 → 0–1)
x_train = x_train / 255.0
x_test = x_test / 255.0

# Build neural network model
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),   # Input layer
    keras.layers.Dense(128, activation='relu'),   # Hidden layer
    keras.layers.Dense(10, activation='softmax')  # Output layer
])

# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train model
print("Training model...")
model.fit(x_train, y_train, epochs=5)

# Evaluate model
print("\nTesting model...")
test_loss, test_acc = model.evaluate(x_test, y_test)

print(f"\nTest Accuracy: {test_acc}")

# Make predictions
predictions = model.predict(x_test)

# Show a sample prediction
plt.imshow(x_test[0], cmap='gray')
plt.title(f"Predicted: {np.argmax(predictions[0])}")
plt.show()
````

````python
"""
train.py - Load Iris data, train IrisNet, save model checkpoint
"""

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from model import IrisNet

# ── Reproducibility ────────────────────────────────────────────────────────────
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

# ── Hyperparameters ────────────────────────────────────────────────────────────
EPOCHS      = 100
BATCH_SIZE  = 16
LR          = 1e-3
WEIGHT_DECAY= 1e-4

# ── 1. Load & preprocess ───────────────────────────────────────────────────────
iris = load_iris()
X, y = iris.data.astype(np.float32), iris.target

scaler = StandardScaler()
X = scaler.fit_transform(X)

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=y
)

# Convert to tensors
X_train_t = torch.tensor(X_train)
y_train_t = torch.tensor(y_train, dtype=torch.long)
X_val_t   = torch.tensor(X_val)
y_val_t   = torch.tensor(y_val,   dtype=torch.long)

train_loader = DataLoader(
    TensorDataset(X_train_t, y_train_t),
    batch_size=BATCH_SIZE, shuffle=True
)

# ── 2. Model / loss / optimizer ────────────────────────────────────────────────
device    = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model     = IrisNet().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=30, gamma=0.5)

print(f"Training on: {device}")
print(model)
print(f"\nTrainable params: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}\n")

# ── 3. Training loop ───────────────────────────────────────────────────────────
train_losses, val_accuracies = [], []

for epoch in range(1, EPOCHS + 1):
    model.train()
    epoch_loss = 0.0

    for X_batch, y_batch in train_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)

        optimizer.zero_grad()
        logits = model(X_batch)
        loss   = criterion(logits, y_batch)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * len(X_batch)

    scheduler.step()
    avg_loss = epoch_loss / len(X_train_t)
    train_losses.append(avg_loss)

    # Validation accuracy
    model.eval()
    with torch.no_grad():
        val_logits = model(X_val_t.to(device))
        preds      = val_logits.argmax(dim=1).cpu()
        val_acc    = (preds == y_val_t).float().mean().item()
    val_accuracies.append(val_acc)

    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{EPOCHS}  loss={avg_loss:.4f}  val_acc={val_acc:.4f}")

# ── 4. Save checkpoint ─────────────────────────────────────────────────────────
checkpoint = {
    "epoch":        EPOCHS,
    "model_state":  model.state_dict(),
    "scaler_mean":  scaler.mean_,
    "scaler_scale": scaler.scale_,
}
torch.save(checkpoint, "iris_model.pth")
print("\nModel saved → iris_model.pth")

# ── 5. Plot training curves ────────────────────────────────────────────────────
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(train_losses, label="Train Loss", color="steelblue")
ax1.set(xlabel="Epoch", ylabel="Cross-Entropy Loss", title="Training Loss")
ax1.legend(); ax1.grid(True, alpha=0.3)

ax2.plot([v * 100 for v in val_accuracies], label="Val Accuracy", color="darkorange")
ax2.set(xlabel="Epoch", ylabel="Accuracy (%)", title="Validation Accuracy", ylim=[0, 105])
ax2.legend(); ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("training_curves.png", dpi=150)
print("Plot saved  → training_curves.png")
plt.show()
````

Here’s a **clear, structured explanation of Neural Networks (NN)** — this is the most advanced model among what you’ve learned so far 👇

***

# ✅ 1. What is a Neural Network?

A Neural Network is a **supervised machine learning model inspired by the human brain**.

👉 It is made of:

* **Neurons (nodes)**
* **Layers**
* **Connections (weights)**

👉 It can learn **very complex patterns** (more than all previous models).

***

# ✅ 2. Structure of Neural Network

```
Input Layer  →  Hidden Layers  →  Output Layer
```

### 🔹 Input Layer

* Receives raw data

### 🔹 Hidden Layers

* Perform calculations
* Learn patterns

### 🔹 Output Layer

* Final prediction

👉 Each layer transforms data step-by-step [\[https://us...s/original\]](https://us-prod.asyncgw.teams.microsoft.com/v1/objects/0-eus-d12-6c4f6126785cf38e5814a635b5259353/views/original)

***

# ✅ 3. Core Idea (Very Important)

👉 “Pass data through layers, apply weights, and produce output”

At each neuron:

```
z = (x1*w1 + x2*w2 + ... + xn*wn) + b
output = activation(z)
```

Where:

* `x` = input
* `w` = weights
* `b` = bias
* `activation()` = function like ReLU / sigmoid

👉 This is repeated layer by layer [\[geeksforgeeks.org\]](https://www.geeksforgeeks.org/deep-learning/what-is-forward-propagation-in-neural-networks/)

***

# ✅ 4. Forward Propagation (Prediction)

👉 Data flows like this:

```
Input → Hidden → Hidden → Output
```

Mathematically:

```
A1 = σ(W1X + b1)
A2 = σ(W2A1 + b2)
Y  = σ(W3A2 + b3)
```

👉 This is called **forward propagation** [\[https://us...s/original\]](https://us-prod.asyncgw.teams.microsoft.com/v1/objects/0-eus-d12-6c4f6126785cf38e5814a635b5259353/views/original)

***

# ✅ 5. Activation Functions (Key)

👉 Without activation → network becomes linear

Common activations:

| Function | Use                   |
| -------- | --------------------- |
| ReLU     | Most common           |
| Sigmoid  | Binary classification |
| Softmax  | Multi-class           |

👉 Activation adds **non-linearity** to learn complex data [\[linkedin.com\]](https://www.linkedin.com/pulse/dissecting-forward-propagation-neural-networks-saurav-prateek-ds7qc/)

***

Here’s a **clear, structured explanation of CNN (Convolutional Neural Networks)** — this is a specialized neural network mainly used for images/video 👇

***

# ✅ 1. What is CNN?

CNN (Convolutional Neural Network) is a type of **deep neural network** designed for:

* ✅ Image processing
* ✅ Video analysis
* ✅ Pattern recognition

👉 It is very powerful for:

* Object detection
* Face recognition
* Medical imaging

***

# ✅ 2. Core Idea (Very Important)

👉 “Instead of looking at the whole image, look at small parts step-by-step”

Traditional NN:

```
Uses full image → too complex
```

CNN:

```
Uses small filters → learns features (edges → shapes → objects)
```

***

# ✅ 3. CNN Architecture (Layers)

```
Input Image → Convolution → ReLU → Pooling → Fully Connected → Output
```

***

# ✅ 4. Key Layers Explained

***

## 🔹 1. Convolution Layer (Most Important)

👉 Applies **filters (kernels)** on image

Example:

* Detect edges
* Detect corners

👉 Operation:

```
Feature Map = Filter ⨂ Image
```

✅ Extracts features automatically

***

## 🔹 2. ReLU Activation

```
ReLU(x) = max(0, x)
```

👉 Removes negative values  
👉 Adds non-linearity

***

## 🔹 3. Pooling Layer

👉 Reduces size of image

Types:

* Max pooling ✅ (most used)
* Average pooling

Example:

```
4x4 → 2x2
```

👉 Keeps important features  
👉 Reduces computation

***

## 🔹 4. Fully Connected Layer

👉 Works like normal neural network

👉 Makes final prediction

***

# ✅ 5. How CNN Works (Step-by-Step)

1. Input image
2. Apply filters (feature detection)
3. Apply activation
4. Reduce size (pooling)
5. Repeat layers
6. Flatten output
7. Final classification

***

# ✅ 6. Simple Intuition

Example: Cat image

CNN learns step-by-step:

```
Layer 1 → edges
Layer 2 → shapes
Layer 3 → eyes, ears
Layer 4 → "cat"
```

👉 Learns features automatically ✅

***

# ✅ 7. Sample Python Code (Basic CNN)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# Step 1: Model
model = Sequential()

# Step 2: Convolution layer
model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))

# Step 3: Pooling
model.add(MaxPooling2D(pool_size=(2,2)))

# Step 4: Flatten
model.add(Flatten())

# Step 5: Fully connected
model.add(Dense(128, activation='relu'))

# Step 6: Output
model.add(Dense(10, activation='softmax'))

# Summary
model.summary()
```

***

# ✅ 9. Output Explanation

* Flatten → convert to vector
