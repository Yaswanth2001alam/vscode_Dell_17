5 Practical psutil python programs

1. 💻 CPU, RAM & Disk Usage
Monitor your system’s CPU, RAM, and disk usage in real time with Python. A simple way to understand your computer’s resource consumption using psutil.

2. 🔥 Top 5 Memory-Hungry Processes
Find which processes are consuming the most RAM on your system. Perfect for quickly identifying memory-heavy applications with Python.

3. 🌐 Network Upload & Download
Track how much data your computer has uploaded and downloaded. Use psutil to monitor network activity directly from Python.

4. 🔋 Battery Status
Check your laptop’s current battery percentage and charging status. A handy Python script for monitoring battery health at a glance.

5. ⚙️ Live System Information
Get essential system details such as OS, processor, CPU cores, and boot time. Build a quick system-information dashboard using Python.


````python
import psutil

# Show CPU usage
print(f"CPU Usage : {psutil.cpu_percent()}%")

# Show RAM usage
memory = psutil.virtual_memory()
print(f"RAM Usage : {memory.percent}%")

# Show Disk usage
disk = psutil.disk_usage("/")
print(f"Disk Usage: {disk.percent}%")
````

````python
import psutil

processes = []

for proc in psutil.process_iter(['pid', 'name', 'memory_percent']):
    try:
        processes.append(proc.info)
    except:
        pass

top = sorted(
    processes,
    key=lambda x: x['memory_percent'],
    reverse=True
)[:5]

for p in top:
    print(p)
````

````python
import psutil

net = psutil.net_io_counters()

print(f"Downloaded : {net.bytes_recv / (1024**2):.2f} MB")
print(f"Uploaded   : {net.bytes_sent / (1024**2):.2f} MB")
````

````python
import psutil

battery = psutil.sensors_battery()

if battery:
    print(f"Battery : {battery.percent}%")
    print(f"Charging: {battery.power_plugged}")
else:
    print("Battery information not available.")
````

````python
import platform
import psutil

print("OS :", platform.system())
print("Release :", platform.release())
print("Processor :", platform.processor())
print("Physical Cores :", psutil.cpu_count(logical=False))
print("Logical Cores :", psutil.cpu_count())
print("Boot Time :", psutil.boot_time())
````

````python
import psutil
import time

processes = []

for proc in psutil.process_iter(["pid", "name"]):
    try:
        proc.cpu_percent()
    except:
        pass

time.sleep(1)

for proc in psutil.process_iter(["pid", "name"]):
    try:
        processes.append({
            "pid": proc.pid,
            "name": proc.name(),
            "cpu": proc.cpu_percent()
        })
    except:
        pass

top = sorted(
    processes,
    key=lambda x: x["cpu"],
    reverse=True
)[:5]

for process in top:
    print(process)
````

````python
import psutil

cpu = psutil.cpu_percent(interval=1)
ram = psutil.virtual_memory().percent
disk = psutil.disk_usage("C:\\").percent

if cpu > 80:
    print("WARNING: High CPU usage!")

if ram > 80:
    print("WARNING: High RAM usage!")

if disk > 80:
    print("WARNING: High Disk usage!")

print(f"CPU  : {cpu}%")
print(f"RAM  : {ram}%")
print(f"Disk : {disk}%")
````

````python
import psutil
import time
import os

while True:

    os.system("cls" if os.name == "nt" else "clear")

    cpu = psutil.cpu_percent(interval=1)
    memory = psutil.virtual_memory()
    net = psutil.net_io_counters()

    print("=" * 40)
    print("       PYTHON SYSTEM MONITOR")
    print("=" * 40)

    print(f"CPU Usage      : {cpu}%")
    print(f"RAM Usage      : {memory.percent}%")

    print(
        f"Downloaded     : "
        f"{net.bytes_recv / (1024**2):.2f} MB"
    )

    print(
        f"Uploaded       : "
        f"{net.bytes_sent / (1024**2):.2f} MB"
    )

    print("=" * 40)

    time.sleep(2)
````

### 1. Check CPU usage continuously

```python
import psutil
import time

while True:
    cpu = psutil.cpu_percent(interval=1)
    print(f"CPU Usage: {cpu}%")
    time.sleep(2)
```

Stop it with `Ctrl + C`.

### 2. Check CPU usage for every core

```python
import psutil

cores = psutil.cpu_percent(interval=1, percpu=True)

for number, usage in enumerate(cores):
    print(f"CPU Core {number}: {usage}%")
```

Example:

```text
CPU Core 0: 12.5%
CPU Core 1: 4.2%
CPU Core 2: 18.7%
CPU Core 3: 2.1%
```

### 13. Build a simple live system monitor

This combines several of the examples:

```python
import psutil
import time
import os

while True:

    os.system("cls" if os.name == "nt" else "clear")

    cpu = psutil.cpu_percent(interval=1)
    memory = psutil.virtual_memory()
    net = psutil.net_io_counters()

    print("=" * 40)
    print("       PYTHON SYSTEM MONITOR")
    print("=" * 40)

    print(f"CPU Usage      : {cpu}%")
    print(f"RAM Usage      : {memory.percent}%")

    print(
        f"Downloaded     : "
        f"{net.bytes_recv / (1024**2):.2f} MB"
    )

    print(
        f"Uploaded       : "
        f"{net.bytes_sent / (1024**2):.2f} MB"
    )

    print("=" * 40)

    time.sleep(2)
```

You should see something like:

```text
========================================
       PYTHON SYSTEM MONITOR
========================================
CPU Usage      : 14.5%
RAM Usage      : 61.2%
Downloaded     : 4832.45 MB
Uploaded       : 742.28 MB
========================================
```
