## Concept notes and mental model

Artificial intelligence is a broad field concerned with systems that perform tasks associated
with perception, language, decision-making, or other intelligent behavior. AI includes engineered
rules and search as well as learned models. Machine learning learns patterns from data rather
than requiring every decision rule to be written explicitly.

Deep learning is machine learning using multilayer neural networks. Generative AI models a
distribution from which it can produce new text, images, audio, or other outputs. These categories
overlap: a transformer language model can be AI, ML, deep learning, and generative AI at once.
Not all ML is deep learning and not all AI is generative. A spam classifier is a discriminative
task; a model that generates a draft email is a generative task.

Narrow AI is designed/evaluated for bounded capabilities. General AI is a contested research
goal concerning broad adaptable capability; there is no single universally accepted operational
definition or agreed test. Fluent text alone does not establish consciousness or general competence.

A dataset is a collection of examples. Features are inputs available when predicting; labels are
target answers for supervised training. Parameters are values learned during training, while
hyperparameters, such as learning rate, configure the training procedure. Inference uses learned
parameters on an input. Online learning is an explicit update process: an ordinary inference
call does not automatically retrain the model.

For a network engineer, inputs might include utilization and packet loss. A target might be
next-minute latency (regression) or an outage flag (classification). Check that those inputs
really exist before the outcome you want to predict.

## Worked example 1: Fit a tiny numerical model, then use it

**Purpose:** Distinguish data, learned parameters, and a new prediction without needing an ML library.

### Read the code in this order
1. Separate observed input values and labels.
2. Estimate the slope and intercept from two consistent observations.
3. Predict for a previously unseen input.
4. Assert a known answer and inspect the learned parameters.

### Syntax and behavior to notice
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
learned parameters: 10.0 10.0
inference latency_ms: 15.0
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
utilization = [0.2, 0.4, 0.6]
latency_ms = [12.0, 14.0, 16.0]
weight = (latency_ms[-1] - latency_ms[0]) / (utilization[-1] - utilization[0])
bias = latency_ms[0] - weight * utilization[0]
prediction = weight * 0.5 + bias
assert abs(prediction - 15.0) < 1e-9
print("learned parameters:", round(weight, 2), round(bias, 2))
print("inference latency_ms:", round(prediction, 2))
````

## Worked example 2: Contrast classification, regression, and generated text

**Purpose:** Make output types explicit without confusing rule-based examples with learned models.

### Read the code in this order
1. Define three different output contracts.
2. Check a class label and a numeric prediction separately.
3. Generate a templated sentence and label it as templating, not generative learning.

### Syntax and behavior to notice
- `assert` checks the stated behavior; a failed check raises AssertionError.

### Expected output
This is the output captured from executing the worked example in the recorded environment. Package versions and floating-point rounding can change numeric details; the assertions express the required shapes, invariants, and thresholds.

```text
category: high numeric: 15.0
template, not a learned generator: edge-01 has high latency.
```

**Before running:** identify the input variables, which values are learned versus chosen, and what each assertion proves. A successful shape test does not by itself prove predictive quality. Training-only loss checks are labeled as such.

````python
def classify_latency(value):
    return "high" if value >= 20 else "normal"
def predict_linear_latency(utilization):
    return 5.0 + 0.2 * utilization
def render_template(hostname, label):
    return f"{hostname} has {label} latency."
label = classify_latency(25)
numeric = predict_linear_latency(50)
text = render_template("edge-01", label)
assert label == "high" and numeric == 15.0
assert text == "edge-01 has high latency."
print("category:", label, "numeric:", numeric)
print("template, not a learned generator:", text)
````
