Great question! A **testbed** is one of the most foundational concepts in network automation. Let me walk you through what it is, how it's configured, and what it does — especially in the context of your networking work.

## What is a testbed?

In network automation (particularly with frameworks like **pyATS/Genie** from Cisco, or similar tools), a **testbed** is a YAML file that acts as the **topology descriptor** — it tells your automation scripts *where* your devices are, *how* to connect to them, and *what* kind of devices they are.

Think of it as a phonebook + identity card for your entire lab or production network.

## What does a testbed file contain?

A typical testbed YAML has these sections: devices (the core — each device with its hostname, OS type, credentials, and connection method), topology (how devices are interconnected — links and interfaces), and optionally testbed-level credentials that devices inherit.

Here's a real-world example for an Arista + IOS-XR lab:

```yaml
testbed:
  name: my_lab
  credentials:
    default:
      username: admin
      password: admin123

devices:
  arista-sw1:
    os: eos
    type: switch
    platform: arista
    connections:
      defaults:
        class: unicon.Unicon
      cli:
        protocol: ssh
        ip: 192.168.1.10
        port: 22

  iosxr-rtr1:
    os: iosxr
    type: router
    connections:
      cli:
        protocol: ssh
        ip: 192.168.1.20

topology:
  arista-sw1:
    interfaces:
      Ethernet1:
        link: link-sw1-rtr1
        type: ethernet
  iosxr-rtr1:
    interfaces:
      GigabitEthernet0/0/0/0:
        link: link-sw1-rtr1
        type: ethernet
```

## What does a testbed *do* in network automation?

Let me diagram the full role of a testbed in an automation workflow:## Key functions of a testbed

**1. Device abstraction** — your script says `device = testbed.devices['arista-sw1']` and doesn't need to hardcode IPs or credentials anywhere in the code.

**2. Connection brokering** — testbed handles the SSH/Telnet session setup, OS-specific prompts, and login automatically via Unicon (for pyATS) or similar.

**3. Topology awareness** — you can query which interfaces are connected to which device, making it useful for link-level tests like port-channel membership or MACsec peer verification.

**4. Credential management** — one place to update credentials for your whole lab. Can also be integrated with Vault or environment variables for security.

**5. Multi-OS normalization** — the same `device.parse('show interfaces')` call works on Arista EOS, IOS-XR, and NXOS, with Genie normalizing the parsed output.

## In a lab setup (EVE-NG / GNS3)

For your EVE-NG lab, the testbed would point to the management IPs of your virtual nodes. The workflow looks like this:

1. Spin up your EVE-NG topology (Arista vEOS, IOS-XRv, etc.)

## What's in each file

**`testbed.yaml`** — inventory for two IOS-XRv PE routers on `192.168.100.0/24`. The only things you need to edit before running are the two management IPs and the device key names (they must match the router's configured hostname exactly, since Unicon uses the hostname to detect the CLI prompt).

**`check_lab.py`** — a 300-line pyATS script that runs all four checks against every device in the testbed and prints a colour-coded summary. The four checks are:

| Check | Command | What it flags |
|---|---|---|
| Interface status | `show interfaces` | Anything not `oper=up / line=up` |
| Bundle-Ether membership | `show bundle` | Members missing `SA` (Selected Active) flag |
| MACsec sessions | `show macsec summary` + detail | Anything not `Secured` |
| BGP neighbors | `show bgp summary` | Any neighbor not `Established` |

**`README.md`** — the full quickstart including the IOS-XRv SSH config snippet.

## Getting started in 5 commands

```bash
# 1. Install
pip install pyats[full] genie

# 2. Edit the two IPs in testbed.yaml, then validate
pyats validate testbed testbed.yaml

# 3. Set credentials
export NET_USER=admin && export NET_PASS=admin

# 4. Run all checks
python check_lab.py

# 5. Or target one device and save a JSON report
python check_lab.py --device XRv-PE1 --report report.json
```

A few things worth noting for your environment: `learn_hostname=True` is set by default in the connect call, so if your lab router's hostname doesn't exactly match the YAML key it will still connect rather than throwing a `PromptNotFound` error. The MACsec check also pulls `show macsec detail interface <intf>` for any interface that comes back `Secured`, giving you the full AN, cipher suite, and key expiry info in the JSON report.
